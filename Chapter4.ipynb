# Exercise 4.1 
Suppose $M(t), 0 \leq t \leq T$, is a martingale with respect to some filtration $\mathcal{F}(t), 0 \leq t \leq T.$ Let $\Delta(t), 0 \leq t \leq T,$ be a simple process adapte to $\mathcal{F}(t)$ (i.e., there is a partition $\Pi = \{t_0, t_1, \ldots, t_n\}$ of $[0, T]$ such that, for every $j, \Delta(t_j)$ is $\mathcal{F}(t_j)$-measurable and $\Delta(t)$ is constant in $t$ on each subinterval $[t_j, t_{j+1})$). For $t \in [t_k, t_{k+1}),$ define the stochastic integral
$$I(t) = \sum_{j=0}^{k-1} \Delta(t_j)[M(t_{j+1}) - M(t_j)] + \Delta(t_k)[M(t) - M(t_k)].$$
We think of $M(t)$ as the price of an asset at time $t$ and $\Delta(t_j)$ as the number of share of the asset held by an investor between $t_j$ and $t_{j+1}$. Then $I(t)$ is the capital gains that accure to the investor between times $0$ and $t$. Show that $I(t), 0 \leq t \leq T$, is a martingale.

**Proof:**

Let $s < t$ with $s \in [t_l, t_{l+1})$ and $t \in [t_k, t_{k+1})$ where $l < k$. Decompose $I(t)$ as:

$$I(t) = \underbrace{\sum_{j=0}^{l-1}\Delta(t_j)[M(t_{j+1})-M(t_j)]}_{\text{(A) past}} + \underbrace{\Delta(t_l)[M(t_{l+1})-M(t_l)]}_{\text{(B) straddles } s} + \underbrace{\sum_{j=l+1}^{k-1}\Delta(t_j)[M(t_{j+1})-M(t_j)]}_{\text{(C) future}} + \underbrace{\Delta(t_k)[M(t)-M(t_k)]}_{\text{(D) current}}$$

**Term (A):** Every term is $\mathcal{F}(s)$-measurable, so:

$$\mathbb{E}\left[\sum_{j=0}^{l-1}\Delta(t_j)[M(t_{j+1})-M(t_j)]\bigg|\mathcal{F}(s)\right] = \sum_{j=0}^{l-1}\Delta(t_j)[M(t_{j+1})-M(t_j)]$$

**Term (B):** Since $\Delta(t_l)$ and $M(t_l)$ are $\mathcal{F}(s)$-measurable, and $M$ is a martingale:

$$\mathbb{E}[\Delta(t_l)[M(t_{l+1})-M(t_l)]|\mathcal{F}(s)] = \Delta(t_l)[\mathbb{E}[M(t_{l+1})|\mathcal{F}(s)] - M(t_l)] = \Delta(t_l)[M(s) - M(t_l)]$$

**Term (C):** For $l+1 \leq j \leq k-1$, by iterated conditioning and the martingale property of $M$:

$$\mathbb{E}[\Delta(t_j)[M(t_{j+1})-M(t_j)]|\mathcal{F}(s)] = \mathbb{E}\left[\Delta(t_j)\underbrace{[\mathbb{E}[M(t_{j+1})|\mathcal{F}(t_j)]-M(t_j)]}_{=0}\bigg|\mathcal{F}(s)\right] = 0$$

**Term (D):** Similarly, by iterated conditioning:

$$\mathbb{E}[\Delta(t_k)[M(t)-M(t_k)]|\mathcal{F}(s)] = \mathbb{E}\left[\Delta(t_k)\underbrace{[\mathbb{E}[M(t)|\mathcal{F}(t_k)]-M(t_k)]}_{=0}\bigg|\mathcal{F}(s)\right] = 0$$

**Conclusion:**

$$\mathbb{E}[I(t)|\mathcal{F}(s)] = \sum_{j=0}^{l-1}\Delta(t_j)[M(t_{j+1})-M(t_j)] + \Delta(t_l)[M(s)-M(t_l)] = I(s) \quad \blacksquare$$

# Exercise 4.2.
Let $W(t), 0 \leq t \leq T$, be a Browninan motion, and let $\mathcal{F}(t), 0 \leq t \leq T,$ be an associated filtration. Let $\Delta(t), 0 \leq t \leq T,$ be a nonrandom simple process (i.e., there is a partition $\Pi = \{t_0, t_1, \ldots, t_n\}$ of $[0,T]$ such that for every $j, \Delta(t_j)$ is a nonrandom quantity and $\Delta(t) = \Delta(t_j)$ is constant in $t$ on the subinterval $[t_j, t_{j+1})$). For $t \in [t_k, t_{k+1}],$ define the stochastic integral
$$I(t) = \sum_{j=0}^{k-1} \Delta(t_j) [W(t_{j+1} - W(t_j)] + \Delta(t_k)[W(t) - W(t_k)].$$
(i) Show that whenever $0 \leq s <t \leq T,$ the increment $I(t)-I(s)$ is independent of $\mathcal{F}(s).$ (Simplification: If $s$ is between two partition points, we can always insert $s$ as an extra partition point. Then we can relabel the partition points so that they are still called $t_0, t_1, \ldots, t_n,$ but with a larger value of $n$ and now with $s = t_k$ for some value of $k$. Of course, we must set $\Delta(s) = \Delta(t_{k-1})$ so that $\Delta$ takes the same value on the interval $[s, t_{k+1})$ as on the interval $[t_{k-1}, s).$ Similarly, we can insert $t$ as an extra partition point if it is not already one. Consequently, to show that $I(t) - I(s)$ is independent of $\mathcal{F}(s)$ for all $0 \leq s < t \leq T,$ it suffices to show that $I(t_k) - I(t_l)$ is independent of $\mathcal{F}(t_l)$ whenever $t_k$ and $t_l$ are two partition points with $t_l < t_k.$ This is all you need to do.)

(ii) Show that whenever $0 \leq s < t \leq T,$ the increment $I(t) - I(s)$ is a normally distributed random variable with mean zero and variance $\int_{s}^{t}\Delta^2(u)du.$

(iii) Use (i) and (ii) to show that $I(t), 0 \leq t \leq T,$ is a martingale.

(iv) Show that $I^2(t) - \int_{0}^{t} \Delta^2(u)du, 0 \leq t \leq T,$ is a martingale.


### (i) Independence of increments

By the simplification in the problem, it suffices to show that $I(t_k) - I(t_l)$ is independent of $\mathcal{F}(t_l)$ for partition points $t_l < t_k$.

$$I(t_k) - I(t_l) = \sum_{j=l}^{k-1}\Delta(t_j)[W(t_{j+1}) - W(t_j)]$$

Since $\Delta(t_j)$ is nonrandom and $W(t_{j+1}) - W(t_j)$ is independent of $\mathcal{F}(t_l)$ for all $j \geq l$, each term is independent of $\mathcal{F}(t_l)$. Therefore $I(t_k) - I(t_l)$ is independent of $\mathcal{F}(t_l)$. $\blacksquare$

### (ii) Normal distribution of increments

Define the process:

$$M_u(t) = \exp\left\{uI(t) - \frac{u^2}{2}\int_0^t \Delta^2(s)ds\right\}$$

This is a martingale (generalized geometric Brownian motion). Let $0 \leq t_1 < t_2 \leq T$. Since $M_u$ is a martingale with $M_u(0) = 1$:

$$1 = \mathbb{E}[M_{u_1}(t_1)] = \mathbb{E}\left[\exp\left\{u_1 I(t_1) - \frac{u_1^2}{2}\int_0^{t_1}\Delta^2(s)ds\right\}\right]$$

Multiplying $M_{u_2}(t_2)/M_{u_2}(t_1)$ (which equals $\exp\{u_2(I(t_2)-I(t_1)) - \frac{u_2^2}{2}\int_{t_1}^{t_2}\Delta^2(s)ds\}$) and taking expectations:

$$\mathbb{E}\left[\exp\{u_1 I(t_1) + u_2(I(t_2)-I(t_1))\}\right] = \exp\left\{\frac{u_1^2}{2}\int_0^{t_1}\Delta^2(s)ds\right\} \cdot \exp\left\{\frac{u_2^2}{2}\int_{t_1}^{t_2}\Delta^2(s)ds\right\}$$

This implies:
- $I(t_1)$ is normally distributed with mean $0$ and variance $\int_0^{t_1}\Delta^2(s)ds$
- $I(t_2) - I(t_1)$ is normally distributed with mean $0$ and variance $\int_{t_1}^{t_2}\Delta^2(s)ds$
- $I(t_1)$ and $I(t_2) - I(t_1)$ are independent $\blacksquare$

### (iii) $I(t)$ is a martingale

$$\mathbb{E}[I(t)|\mathcal{F}(s)] = \mathbb{E}[I(t)-I(s)+I(s)|\mathcal{F}(s)]$$

$$= \mathbb{E}[I(t)-I(s)] + I(s)$$

where we used: (a) independence of $I(t)-I(s)$ from $\mathcal{F}(s)$ by (i), and (b) $\mathbb{E}[I(t)-I(s)] = 0$ by (ii). Therefore:

$$\mathbb{E}[I(t)|\mathcal{F}(s)] = 0 + I(s) = I(s) \quad \blacksquare$$

### (iv) $I^2(t) - \int_0^t \Delta^2(u)du$ is a martingale

$$\mathbb{E}\left[I^2(t) - \int_0^t \Delta^2(u)du\bigg|\mathcal{F}(s)\right]$$

$$= \mathbb{E}\left[(I(t)-I(s))^2 + 2I(t)I(s) - I^2(s) - \int_0^t \Delta^2(u)du\bigg|\mathcal{F}(s)\right]$$

- $\mathbb{E}[(I(t)-I(s))^2] = \int_s^t \Delta^2(u)du$ by (ii)
- $\mathbb{E}[2I(t)I(s)|\mathcal{F}(s)] = 2I(s)\mathbb{E}[I(t)|\mathcal{F}(s)] = 2I^2(s)$ by taking-out property and (iii)
- $I^2(s)$ and $\int_0^t\Delta^2(u)du$ are $\mathcal{F}(s)$-measurable

Therefore:

$$= \int_s^t \Delta^2(u)du + 2I^2(s) - I^2(s) - \int_0^t \Delta^2(u)du = I^2(s) - \int_0^s \Delta^2(u)du \quad \blacksquare$$

# Exercise 4.3.
We now consider a case in which $\Delta(t)$ in Exercise 4.2 is simple but random. In particular, let $t_0 = 0, t_1 = s. t_2 = t,$ and let $\Delta(0)$ be nonrandom and $\Delta(s) = W(s)$. Which of the following assertions is true?

(i) $I(t)-I(s)$ is independent of $\mathcal{F}(s)$.

(ii) $I(t) - I(s)$ is normally distributed. (Hint: Check if the forth moment is three times the square of the variance; see Exercise 3.3 of Chapter 3.)

(iii) $\mathbb{E}[I(t)|\mathcal{F}(s)] = I(s).$

(iv) $\mathbb{E}[I^2(t) - \int_{0}^{t} \Delta^2(u)du |\mathcal{F}(s)] = I^2(s) - \int_{0}^{s} \Delta^2(u)du.$

### (i) Is $I(t) - I(s)$ independent of $\mathcal{F}(s)$? **False** 

Since $W(s)$ is $\mathcal{F}(s)$-measurable, $I(t) - I(s) = W(s)[W(t)-W(s)]$ depends on $\mathcal{F}(s)$. Therefore $I(t) - I(s)$ is not independent of $\mathcal{F}(s)$.

Note: In general, for $s \in [t_l, t_{l+1})$:

$$I(t) - I(s) = \underbrace{-\Delta(t_l)[W(s)-W(t_l)]}_{\mathcal{F}(s)\text{-measurable}} + \sum_{j=l+1}^{k-1}\Delta(t_j)[W(t_{j+1})-W(t_j)] + \Delta(t_k)[W(t)-W(t_k)]$$

The first term is $\mathcal{F}(s)$-measurable, so independence fails.

### (ii) Is $I(t) - I(s)$ normally distributed? **False** 

By Exercise 3.3, a normally distributed random variable satisfies $\mathbb{E}[X^4] = 3(\text{Var}(X))^2$.

**Variance:**

$$\mathbb{E}[(I(t)-I(s))^2] = \mathbb{E}[W^2(s)]\cdot\mathbb{E}[(W(t)-W(s))^2] = s(t-s)$$

**Fourth moment:**

$$\mathbb{E}[(I(t)-I(s))^4] = \mathbb{E}[W^4(s)]\cdot\mathbb{E}[(W(t)-W(s))^4] = 3s^2\cdot 3(t-s)^2 = 9s^2(t-s)^2$$

**Check:**

$$3(\text{Var})^2 = 3s^2(t-s)^2 \neq 9s^2(t-s)^2$$

Therefore $I(t) - I(s)$ is not normally distributed.

### (iii) $\mathbb{E}[I(t)|\mathcal{F}(s)] = I(s)$? **True** 

$$\mathbb{E}[I(t)|\mathcal{F}(s)] = \mathbb{E}[I(s) + W(s)(W(t)-W(s))|\mathcal{F}(s)]$$

$$= I(s) + W(s)\underbrace{\mathbb{E}[W(t)-W(s)|\mathcal{F}(s)]}_{=0} = I(s) \quad \blacksquare$$

### (iv) $\mathbb{E}\left[I^2(t) - \int_0^t\Delta^2(u)du\bigg|\mathcal{F}(s)\right] = I^2(s) - \int_0^s\Delta^2(u)du$? **True** 

**Computing $\mathbb{E}[I^2(t)|\mathcal{F}(s)]$:**

$$\mathbb{E}[I^2(t)|\mathcal{F}(s)] = \mathbb{E}[(I(s) + W(s)(W(t)-W(s)))^2|\mathcal{F}(s)]$$

$$= I^2(s) + 2I(s)W(s)\underbrace{\mathbb{E}[W(t)-W(s)|\mathcal{F}(s)]}_{=0} + W^2(s)\underbrace{\mathbb{E}[(W(t)-W(s))^2|\mathcal{F}(s)]}_{=t-s}$$

$$= I^2(s) + W^2(s)(t-s)$$

**Computing $\mathbb{E}\left[\int_0^t\Delta^2(u)du\big|\mathcal{F}(s)\right]$:**

$$\mathbb{E}\left[\int_0^t\Delta^2(u)du\bigg|\mathcal{F}(s)\right] = \int_0^s\Delta^2(u)du + \mathbb{E}\left[\int_s^t W^2(s)du\bigg|\mathcal{F}(s)\right] = \int_0^s\Delta^2(u)du + W^2(s)(t-s)$$

**Conclusion:**

$$\mathbb{E}\left[I^2(t) - \int_0^t\Delta^2(u)du\bigg|\mathcal{F}(s)\right] = I^2(s) + W^2(s)(t-s) - \int_0^s\Delta^2(u)du - W^2(s)(t-s) = I^2(s) - \int_0^s\Delta^2(u)du \quad \blacksquare$$

# Exercise 4.4 (Stratonovich integral).
Let $W(t), t \geq 0$, be a Brownian motion. Let $T$ be a fixed positive number and let $\Pi = \{t_0, t_1, \ldots, t_n\}$ be a partition of $[0,T]$ (i.e., $0 = t_0 < t_1 < \cdots < t_n = T$). For each $j$, define $t_j^* = \frac{t_j + t_{j+1}}{2}$ to be the midpoint of the interval $[t_j, t_{j+1}]$.

(i) Define the half-sample quadratic variation corresponding to $\Pi$ to be 
$$Q_{\Pi/2} = \sum_{j=0}^{n-1} (W(t_j^*) - W(t_j))^2.$$
Show that $Q_{\Pi/2} = \frac{1}{2} T $ as $||\Pi|| \to 0$. (Hint: It suffices to show that $\mathbb{E}Q_{\Pi/2} = \frac{1}{2}T$ and $\lim_{||\Pi|| \to 0} \text{Var}(Q_{\Pi/2}) = 0.$)

(ii) Define the Stranovich integral of $W(t)$ with respect to $W(t)$ to be
$$\int_{0}^{T} W(t) \circ dW(t) = \lim_{||\Pi|| \to 0} \sum_{j=0}^{n-1} W(t_j^*)(W(t_{j+1}) - W(t_j)). \tag{4.10.1}$$
In contrast to the Ito integral $\int_{0}^{T} W(t)dW(t) = \frac{1}{2} W^2(T) - \frac{1}{2}T$ of (4.3.4), which evaluates the integrand at the left endpoint of each subinterval $[t_j, t_{j+1}]$, here we evaluate the integrand at the midpoint $t_j^*$. Show that
$$\int_{0}^{T} W(t)\circ dW(t) = \frac{1}{2} W^2(T).$$
(Hint: Write the approximating sum in (4.10.1) as the sum of an approxicimating sum for the Ito integral $\int_{0}^{T} W(t) dW(t)$ and $Q_{\Pi/2}.$ The approximating sum for the Ito integral is the one corresponding to the partiion $0 = t_0 < t_0^* < t_1 < t_1^* < \cdots < t_{n-1}^* < t_n = T,$ not the partition $\Pi.$)

### (i) Half-sample quadratic variation

**$\mathbb{E}Q_{\Pi/2} = \frac{1}{2}T$:**

$$\mathbb{E}Q_{\Pi/2} = \sum_{j=0}^{n-1}\mathbb{E}[(W(t_j^*)-W(t_j))^2] = \sum_{j=0}^{n-1}(t_j^*-t_j) = \sum_{j=0}^{n-1}\frac{t_{j+1}-t_j}{2} = \frac{T}{2} \quad \blacksquare$$

**$\text{Var}(Q_{\Pi/2}) \to 0$:**

Since $(W(t_j^*)-W(t_j))^2$ are independent across $j$, and using $\mathbb{E}[X^4] = 3\sigma^4$ for normal $X$:

$$\text{Var}(Q_{\Pi/2}) = \sum_{j=0}^{n-1}\text{Var}((W(t_j^*)-W(t_j))^2)$$

$$= \sum_{j=0}^{n-1}\left[3\left(\frac{t_{j+1}-t_j}{2}\right)^2 - \left(\frac{t_{j+1}-t_j}{2}\right)^2\right] = 2\sum_{j=0}^{n-1}\left(\frac{t_{j+1}-t_j}{2}\right)^2$$

$$\leq \frac{||\Pi||}{2}\sum_{j=0}^{n-1}(t_{j+1}-t_j) = \frac{||\Pi||T}{2} \to 0 \text{ as } ||\Pi|| \to 0 \quad \blacksquare$$

Therefore $Q_{\Pi/2} \to \frac{T}{2}$ as $||\Pi|| \to 0$.

### (ii) Stratonovich integral equals $\frac{1}{2}W^2(T)$

**Decomposing $Q_{\Pi/2}$:**

$$Q_{\Pi/2} = \sum_{j=0}^{n-1}(W(t_j^*)-W(t_j))^2$$

$$= \sum_{j=0}^{n-1}\left[W^2(t_j^*) - 2W(t_j^*)W(t_j) + W^2(t_j)\right]$$

Rearranging:

$$\sum_{j=0}^{n-1}W(t_j^*)(W(t_{j+1})-W(t_j)) = Q_{\Pi/2} + \sum_{j=0}^{n-1}\left[W(t_j^*)(W(t_{j+1})-W(t_j^*)) + W(t_j)(W(t_j^*)-W(t_j))\right]$$

**Identifying the terms:**

- **LHS**: approximating sum for the Stratonovich integral $\int_0^T W(t)\circ dW(t)$
- **RHS first term**: $Q_{\Pi/2} \to \frac{T}{2}$
- **RHS second term**: approximating sum for the Ito integral with respect to the refined partition $\{t_0, t_0^*, t_1, t_1^*, \ldots, t_{n-1}^*, t_n\}$, which converges to $\int_0^T W(t)dW(t) = \frac{1}{2}W^2(T) - \frac{1}{2}T$

**Taking $||\Pi|| \to 0$:**

$$\int_0^T W(t)\circ dW(t) = \frac{T}{2} + \frac{1}{2}W^2(T) - \frac{1}{2}T = \frac{1}{2}W^2(T) \quad \blacksquare$$

Let $f(x) = x^4$, so $f'(x) = 4x^3$ and $f''(x) = 12x^2$. Using $dW(t)dW(t) = dt$:

$$d(W^4(t)) = 4W^3(t)dW(t) + \frac{1}{2}\cdot 12W^2(t)dt = 4W^3(t)dW(t) + 6W^2(t)dt$$

Integrating from $0$ to $T$:

### (ii) $\mathbb{E}W^4(T) = 3T^2$

Taking expectations:

$$\mathbb{E}W^4(T) = \mathbb{E}\left[\int_0^T 4W^3(t)dW(t)\right] + \mathbb{E}\left[\int_0^T 6W^2(t)dt\right]$$

The Ito integral has expectation zero (martingale property), and by Fubini's theorem:

$$= 0 + 6\int_0^T \mathbb{E}[W^2(t)]dt = 6\int_0^T t\, dt = 6\cdot\frac{T^2}{2} = 3T^2 \quad \blacksquare$$

$$d(W^6(t)) = 6W^5(t)dW(t) + 15W^4(t)dt$$

$$\mathbb{E}W^6(T) = 0 + 15\int_0^T \mathbb{E}[W^4(t)]dt = 15\int_0^T 3t^2\, dt = 15\cdot T^3 = 15T^3 \quad \blacksquare$$

# Exercise 4.10 (Self-financing trading).
The fundamental idea behind no-arbitrage pricing is to reproduce the payoff of a derivative security by trading in the underlying asset (which we call a stock) and the money market account. In descrete time, we let $X_k$ denote the value of the hedging portfolio at time $k$ and let $\Delta/k$ denot the number of shares of stock held between times $k$ and $k + 1$. Then, at time$k$, after rebalancing (i.e., moving from a position of $\Delta_{k-1}$ to a position $\Delta_k$ in the stock), the amount in the money market account is $X_k - S_k \Delta_k.$ The value of the portfolio at time $k + 1$ is
$$X_{k+1} = \Delta_k S_{k+1} + (1 + r) (X_k - \Delta_k S_k).$$
This formula can be rearranged to become
$$X_{k+1} - X_k = \Delta_k(S_{k+1} - S_k) + r(X_k - \Delta_k S_k),$$
which says that the gain between time $k$ and time $k + 1$ is the sum of the capital gain on the stock holdings, $\Delta_k (S_{k+1} - S_k),$ and the interest earnings on the money market account, $r(X_k - \Delta_k S_k).$ The continuous-time analogue of (4.10.8) is 
$$dX(t) = \Delta(t) dS(t) + r(X(t) - \Delta(t) S(t)) dt.$$
Alternatively, one could define the value of a share of the money market account at time $k$ to be 
$$M_k = ( 1+ r)^k$$
and formulate the discrete-time model with two processes, $\Delta_k$ as before and $\Gamma_k$ denoting the number of shares of the money market account held at time $k$ after rebalancing. Then
$$X_k = \Delta_k S_k + \Gamma_k M_k,$$
so that (4.10.7) becomes
$$X_{k+1} = \Delta_k S_{k+1} + (1+r)\gamma_k M_k = \Delta_k S_{k+1} + \Gamma_k M_{k+1}.$$
Subtracting (4.10.10) from (4.10.11), we obtain in place of (4.10.8) the equation 
$$X_{k+1} - X_k = \Delta_k (S_{k+1} - S_k) + \Gamma_k (M_{k+1} - M_k),$$
which says that the gain between time $k$ and $k+1$ is the sum of the capital gain on stock holdings, $\Delta_k(S_{k+1} - S_k),$ and the earnings from the money market investment, $\Gamma_k (M_{k+1} - M_k).$

But $\Delta_k$ and $\Gamma_k$ cannot be chosen arbitrarily. The agent arrives at time $k+1$ with some portfolio of $\Delta_k$ shares of stock and $\Gamma_k$ shares of the money market account and then rebalances. In terms of $\Delta_k$ and $\Gamma_k$, the value of the portfolio upon arrival at time $k+1$ is given by (4.10.11). After rebalancing, it is
$$X_{k+1} = \Delta_{k+1}S_{k+1} + \Gamma_{k+1} M_{k+1}.$$
Setting these two values equal, we obtain the discrete-time self-financing condition
$$S_{k+1}(\Delta_{k+1} - \Delta_k) + M_{k+1} (\Gamma_{k+1} - \Gamma_k) = 0.$$
The first term is the cost of rebalancing in the stock, and the second is the cost of rebalancing in the money market account. If the sum of these two terms is not zero, then money must either be put into the position or can be taken out as a by-product of rebalancing. The point is that when the two processes $\Delta_k$ and $\Gamma_k$ are used to describe the evolution of the portfolio value $X_k$, then two equations, (4.10.12) and (4.10.13) are required rather than the single equation (4.10.8) when only the process $\Delta_k$ is used.

Finally, we note that we may rewrite the discrete-time self-financing condition (4.10.13) as
$$S_k (\Delta_{k+1} - \Delta_k) + (S_{k+1} - S_k)(\Delta_{k+1} - \Delta_k) $$
$$ + M_k(\Gamma_{k+1} - \Gamma_k) + (M_{k+1} - M_k)(\Gamma_{k+1} - \Gamma_k) = 0.$$
This is suggestive of the continuous-time self-financing condition
$$S(t)d\Delta(t) + dS(t) d\Delta(t) + M(t) d\Gamma(t) + dM(t)d\Gamma(t) = 0,$$

$$\lim_{\epsilon \downarrow 0} \frac{1}{\epsilon} \mathbb{E}[(Y_i(t_0 + \epsilon) - Y_i(t_0))(Y_j(t_0 + \epsilon) - Y_j(t_0))|\mathcal{F}(t_0)] = \rho_{ij}(t_0)\sigma_i(t_0)\sigma_j(t_0).$$
To handle the other terms that arise from expanding (4.10.36), you will need (4.10.31) and the fact that
$$\lim_{\epsilon \downarrow 0} \mathbb{E}[|Y_i(t_0 + \epsilon) - Y_1(t_0)||\mathcal{F}(t_0)] = 0. \tag{4.10.37}$$
You may use (4.10.37) without proving it.

(v) Show that, conditioned on $\mathcal{F}(t_0),$ the pair of random vairables
$$(X_1(t_0 + \epsilon) - X_1(t_0), X_2(t_0 + \epsilon) - X_2(t_0))$$
has variances and covariance
$$V_i(\epsilon) = \mathbb{E}\left[(X_i(t_0 + \epsilon) - X_i(t_0))^2|\mathcal{F}(t_0) \right] - M_i^2(\epsilon)$$
$$ = \sigma_i^2(t_0)\epsilon + o(\epsilon) \text{ for } i = 1, 2, $$
$$C(\epsilon) = \mathbb{E}[(X_1(t_0 +\epsilon) - X_1(t_0))(X_2(t_0 + \epsilon) - X_2(t_0))| \mathcal{F}(t_0)] - M_1(\epsilon) M_2(\epsilon)$$
$$ = \rho(t_0) \sigma_1(t_0) \sigma_2(t_0) \epsilon + o(\epsilon).$$
(vi) Show that 
$$\lim_{\epsilon \downarrow 0} \frac{C(\epsilon)}{\sqrt{V_1(\epsilon)V_2(\epsilon)}} = \rho(t_0).$$
In other words, for small values of $\epsilon > 0,$ conditioned on $\mathcal{F}(t_0),$ the correlation between the increments $X_1(t_0 + \epsilon) - X_1(t_0)$ ans $X_2(t_0 + \epsilon) - X_2(t_0)$ is approximately equal to $\rho(t_0)$, and this approximation becomes exavt as $\epsilon \downarrow 0.$

### (i) $\mathbb{E}[(B_1(t_0+\epsilon)-B_1(t_0))(B_2(t_0+\epsilon)-B_2(t_0))|\mathcal{F}(t_0)] = \rho\epsilon$

By Itô's product rule:

$$d(B_1 B_2) = B_1\,dB_2 + B_2\,dB_1 + dB_1\,dB_2 = B_1\,dB_2 + B_2\,dB_1 + \rho\,dt$$

Integrating from $t_0$ to $t_0+\epsilon$ and taking conditional expectations (Ito integrals have zero expectation):

$$\mathbb{E}[B_1(t_0+\epsilon)B_2(t_0+\epsilon)|\mathcal{F}(t_0)] = B_1(t_0)B_2(t_0) + \rho\epsilon$$

Therefore:

$$\mathbb{E}[(B_1(t_0+\epsilon)-B_1(t_0))(B_2(t_0+\epsilon)-B_2(t_0))|\mathcal{F}(t_0)]$$

$$= \mathbb{E}[B_1(t_0+\epsilon)B_2(t_0+\epsilon)|\mathcal{F}(t_0)] - B_1(t_0)B_2(t_0) = \rho\epsilon \quad \blacksquare$$

### (ii) Means, variances, and covariance (constant coefficients)

Since $X_i(t_0+\epsilon) - X_i(t_0) = \mathcal{\Theta}_i\epsilon + \sigma_i(B_i(t_0+\epsilon)-B_i(t_0))$:

**Means:**

$$M_i(\epsilon) = \mathbb{E}[X_i(t_0+\epsilon)-X_i(t_0)|\mathcal{F}(t_0)] = \mathcal{\Theta}_i\epsilon$$

**Variances:**

$$\mathbb{E}[(X_i(t_0+\epsilon)-X_i(t_0))^2|\mathcal{F}(t_0)] = \mathcal{\Theta}_i^2\epsilon^2 + \sigma_i^2\epsilon$$

$$V_i(\epsilon) = \mathcal{\Theta}_i^2\epsilon^2 + \sigma_i^2\epsilon - \mathcal{\Theta}_i^2\epsilon^2 = \sigma_i^2\epsilon$$

**Covariance:** Expanding the product and using (i):

$$\mathbb{E}[(X_1(t_0+\epsilon)-X_1(t_0))(X_2(t_0+\epsilon)-X_2(t_0))|\mathcal{F}(t_0)] = \mathcal{\Theta}_1\mathcal{\Theta}_2\epsilon^2 + \rho\sigma_1\sigma_2\epsilon$$

$$C(\epsilon) = \mathcal{\Theta}_1\mathcal{\Theta}_2\epsilon^2 + \rho\sigma_1\sigma_2\epsilon - \mathcal{\Theta}_1\mathcal{\Theta}_2\epsilon^2 = \rho\sigma_1\sigma_2\epsilon$$

**Correlation:**

$$\frac{C(\epsilon)}{\sqrt{V_1(\epsilon)V_2(\epsilon)}} = \frac{\rho\sigma_1\sigma_2\epsilon}{\sigma_1\sigma_2\epsilon} = \rho \quad \blacksquare$$

### (iii) Conditional mean formula (general case)

Since the Ito integral is a martingale:

$$M_i(\epsilon) = \mathbb{E}[X_i(t_0+\epsilon)-X_i(t_0)|\mathcal{F}(t_0)] = \mathbb{E}\left[\int_{t_0}^{t_0+\epsilon}\mathcal{\Theta}_i(u)du\bigg|\mathcal{F}(t_0)\right]$$

By the conditional DCT (using $|\mathcal{\Theta}_i(u)| \leq M$ as dominating function):

$$\lim_{\epsilon\downarrow 0}\frac{1}{\epsilon}M_i(\epsilon) = \mathbb{E}\left[\lim_{\epsilon\downarrow 0}\frac{1}{\epsilon}\int_{t_0}^{t_0+\epsilon}\mathcal{\Theta}_i(u)du\bigg|\mathcal{F}(t_0)\right] = \mathbb{E}[\mathcal{\Theta}_i(t_0)|\mathcal{F}(t_0)] = \mathcal{\Theta}_i(t_0) \quad \blacksquare$$

### (iv) $D_{ij}(\epsilon) = \rho_{ij}(t_0)\sigma_i(t_0)\sigma_j(t_0)\epsilon + o(\epsilon)$

Define martingales $Y_i(t) = \int_0^t\sigma_i(u)dB_i(u)$. By Itô's product rule:

$$d(Y_i Y_j) = Y_i\,dY_j + Y_j\,dY_i + dY_i\,dY_j = Y_i\sigma_j\,dB_j + Y_j\sigma_i\,dB_i + \sigma_i\sigma_j\rho_{ij}\,dt$$

Taking conditional expectations (Ito integrals vanish):

$$\mathbb{E}[Y_i(t_0+\epsilon)Y_j(t_0+\epsilon)|\mathcal{F}(t_0)] = Y_i(t_0)Y_j(t_0) + \int_{t_0}^{t_0+\epsilon}\sigma_i(u)\sigma_j(u)\rho_{ij}(u)du$$

Since the cross terms involving $\int_{t_0}^{t_0+\epsilon}\mathcal{\Theta}_i(u)du$ are $o(\epsilon)$ by (iii) and (4.10.37):

$$D_{ij}(\epsilon) = \int_{t_0}^{t_0+\epsilon}\sigma_i(u)\sigma_j(u)\rho_{ij}(u)du$$

By the fundamental theorem of calculus:

$$\lim_{\epsilon\downarrow 0}\frac{1}{\epsilon}D_{ij}(\epsilon) = \sigma_i(t_0)\sigma_j(t_0)\rho_{ij}(t_0)$$

Therefore $D_{ij}(\epsilon) = \rho_{ij}(t_0)\sigma_i(t_0)\sigma_j(t_0)\epsilon + o(\epsilon)$. $\blacksquare$

### (v) Variances and covariance (general case)

Setting $i = j = 1$ and $i = j = 2$ in (iv):

$$V_i(\epsilon) = D_{ii}(\epsilon) = \sigma_i^2(t_0)\epsilon + o(\epsilon) \quad \blacksquare$$

Setting $i = 1, j = 2$ in (iv):

$$C(\epsilon) = D_{12}(\epsilon) = \rho(t_0)\sigma_1(t_0)\sigma_2(t_0)\epsilon + o(\epsilon) \quad \blacksquare$$

### (vi) Instantaneous correlation

$$\frac{C(\epsilon)}{\sqrt{V_1(\epsilon)V_2(\epsilon)}} = \frac{\rho(t_0)\sigma_1(t_0)\sigma_2(t_0)\epsilon + o(\epsilon)}{\sqrt{\sigma_1^2(t_0)\sigma_2^2(t_0)\epsilon^2 + o(\epsilon^2)}}$$

$$= \frac{\rho(t_0)\sigma_1(t_0)\sigma_2(t_0) + \frac{o(\epsilon)}{\epsilon}}{\sigma_1(t_0)\sigma_2(t_0) + \frac{o(\epsilon)}{\epsilon}} \xrightarrow{\epsilon\downarrow 0} \rho(t_0) \quad \blacksquare$$

Therefore, for small $\epsilon > 0$, conditioned on $\mathcal{F}(t_0)$, the correlation between the increments $X_1(t_0+\epsilon)-X_1(t_0)$ and $X_2(t_0+\epsilon)-X_2(t_0)$ is approximately $\rho(t_0)$, becoming exact as $\epsilon\downarrow 0$.

# Exercise 4.19. 
Let $W(t)$ be a Brownian motion, and define
$$B(t) = \int_{0}^{t} \text{sign}(W(s))dW(s),$$
where 
$$\text{sign}(x) = \begin{cases} 1 & \text{ if } x \geq 0, \\ -1 & \text{ if } x < 0. \end{cases}$$
(i) Show that $B(t)$ is a Brownian motion.

(ii) Use Itô's product rule to compute $d[B(t)W(t)].$ Integrate both sides of the resulting equation and take expectations. Show that $\mathbb{E}[B(t)W(t)] = 0$ (i.e., $B(t)$ and $W(t)$ are uncorrelated).

(iii) Verify that
$$dW^2(t) = 2W(t)dW(t) + dt.$$
(iv) Use Itô's product rule to compute $d[B(t)W^2(t)].$ Integrate both sides of the resulting equation and take expectations to conclude that
$$\mathbb{E}[B(t)W^2(t)] \neq \mathbb{E}B(t) \cdot \mathbb{E}W^2(t).$$

### (i) $B(t)$ is a Brownian motion

$$dB(t) = \text{sign}(W(t))dW(t)$$

$$dB\,dB = \text{sign}^2(W(t))dW\,dW = dt$$

By Lévy's theorem, $B(t)$ is a Brownian motion. $\blacksquare$

### (ii) $\mathbb{E}[B(t)W(t)] = 0$

By Itô's product rule:

$$d[B(t)W(t)] = B(t)dW(t) + W(t)dB(t) + dB\,dW$$

$$= B(t)dW(t) + W(t)\text{sign}(W(t))dW(t) + \text{sign}(W(t))dt$$

Integrating:

$$B(t)W(t) = \int_0^t \text{sign}(W(u))du + \int_0^t B(u)dW(u) + \int_0^t |W(u)|dW(u)$$

Taking expectations (Itô integrals vanish) and using $\mathbb{E}[\text{sign}(W(u))] = 0$ (by symmetry of $W(u)$):

$$\mathbb{E}[B(t)W(t)] = \int_0^t \mathbb{E}[\text{sign}(W(u))]du = \int_0^t 0\,du = 0 \quad \blacksquare$$

### (iii) $dW^2(t) = 2W(t)dW(t) + dt$

By Itô's lemma with $f(x) = x^2$:

$$dW^2(t) = 2W(t)dW(t) + \frac{1}{2}\cdot 2\cdot dW\,dW = 2W(t)dW(t) + dt \quad \blacksquare$$

### (iv) $\mathbb{E}[B(t)W^2(t)] \neq \mathbb{E}[B(t)]\cdot\mathbb{E}[W^2(t)]$

By Itô's product rule using (iii):

$$d[B(t)W^2(t)] = B(t)dW^2(t) + W^2(t)dB(t) + dB\,dW^2$$

$$= 2B(t)W(t)dW(t) + B(t)dt + W^2(t)\text{sign}(W(t))dW(t) + 2W(t)\text{sign}(W(t))dt$$

Integrating and taking expectations (Itô integrals vanish):

$$\mathbb{E}[B(t)W^2(t)] = \int_0^t\mathbb{E}[B(u)]du + 2\int_0^t\mathbb{E}[W(u)\text{sign}(W(u))]du$$

**First term:** $\mathbb{E}[B(u)] = 0$ (martingale), so this is $0$.

**Second term:** $W(u)\text{sign}(W(u)) = |W(u)|$, and since $W(u)\sim\mathcal{N}(0,u)$:

$$\mathbb{E}[|W(u)|] = \sqrt{\frac{2u}{\pi}}$$

Therefore:

$$\mathbb{E}[B(t)W^2(t)] = 2\int_0^t\sqrt{\frac{2u}{\pi}}du = 2\sqrt{\frac{2}{\pi}}\cdot\frac{2}{3}t^{3/2} = \frac{4}{3}\sqrt{\frac{2}{\pi}}t^{3/2} \neq 0$$

However:

$$\mathbb{E}[B(t)]\cdot\mathbb{E}[W^2(t)] = 0 \cdot t = 0$$

Since $\mathbb{E}[B(t)W^2(t)] \neq \mathbb{E}[B(t)]\cdot\mathbb{E}[W^2(t)]$, $B(t)$ and $W(t)$ are not independent. $\blacksquare$

**Remark:** Although $B(t)$ and $W(t)$ are uncorrelated (shown in (ii)), they are not independent. This is because uncorrelatedness only implies independence for jointly normal random variables, and the pair $(B(t), W(t))$ is not jointly normal.

- $x < K$: for large $n$, $f'_n(x) = 0$
- $x > K$: for large $n$, $f'_n(x) = 1$
- $x = K$: $f'_n(K) = \frac{1}{2}$ for all $n$ $\blacksquare$

### (v) $L_K(T) = 0$ if path stays below $K$

If $W(t) < K$ for all $t \in [0,T]$, then $\exists\, \epsilon > 0$ such that $W(t) < K - \epsilon$ for all $t$. For $n$ large enough ($\frac{1}{2n} < \epsilon$):

$$\mathbb{I}_{(K-\frac{1}{2n}, K+\frac{1}{2n})}(W(t)) = 0 \quad \forall t \in [0,T]$$

Therefore:

$$L_K(T) = \lim_{n\to\infty} n\int_0^T \mathbb{I}_{(K-\frac{1}{2n}, K+\frac{1}{2n})}(W(t))dt = 0 \quad \blacksquare$$

### (vi) $L_K(T) = 0$ a.s. is impossible

From (4.10.44) with $W(0) = 0$ and $K > 0$:

$$L_K(T) = (W(T)-K)^+ - \int_0^T \mathbb{I}_{(K,\infty)}(W(t))dW(t)$$

Suppose $L_K(T) = 0$ a.s. Taking expectations (Itô integral has zero expectation):

$$0 = \mathbb{E}[L_K(T)] = \mathbb{E}[(W(T)-K)^+]$$

However, $\mathbb{E}[(W(T)-K)^+] > 0$. Contradiction. Therefore $L_K(T) > 0$ for some paths. $\blacksquare$

# Exercise 4.21 (Stop-loss start-gain paradox). 
Let $S(t)$ be a geometric Brownian motion with mean rate of return zero. In other words,
$$dS(t) = \sigma S(t) dW(t),$$
where the volatility $\sigma$ is constant. We assume the interest rate is $r = 0$.

Suppose we want to hedge a short position in a European call with strike price $K$ and expiration date $T$. We assume that the call is initially out of the money (i.e., $S(0) < K$). Starting with zero capital ($X(0) = 0$), we could try the following portfolio strategy: own on eshare of the stock whenever its price striclty exxceeds $K$, and own zero shares whenber its price is $K$ or less. In other words, we use the hedging portfolio process
$$\Delta(t) = \mathbb{I}_{(K, \infty)} (S(t)).$$
The value of this hedge follows the stochastic differential equation
$$dX(t) = \Delta(t)dS(t) + r(X(t) - \Delta(t)X(t))dt,$$
and since $r = 0$ and $X(0) = 0$, we have
$$X(T) = \sigma \int_{0}^{T} \mathbb{I}_{(K, \infty)} (S(t))S(t)dW(t). \tag{4.10.46}$$
Executing this hedge requires us to borrow from the money market to buy a share of stock whenever the stock price rises across level $K$ and sell the share, repaying the money market debt, when it falls back across level $K$. (Recall that we have taken the interest rate to be zero. The situation we are describing can also occur with a nonzero interest rate, but it is more complicated to set up.) At expiration, if the stock price $S(T)$ is below $K$, there would appear to have been an even number of crossings of the level $K$, half in the up direction and half in the down direction, so that we would have bought and sold the stock repeatedly, each time at the same price $K$, and at the final time have no stock and zero debt to the money market. In other words, if $S(T) < K$, then $X(T) = 0$. On the other hand, if at the final time $S(T)$ is above $K$, we have bought the stock one more time than we sold it, so that we end with one share of stock and a debt of $K$ to the money market. Hence, if $S(T) > K,$ we have $X(T) = S(T) - K.$ If at the final time $S(T) = K,$ then we either own a share of stock valued at K and have a money market debt $K$ or we have sold the stock and have zero money market debt. In either case, $X(T) = 0.$ According to this argument, regardless of the final stock price, we have $X(T) = (S(T) - K)^+.$ This kind of hedging is called stop-loss start-gain strategy.

(i)Discuss the practical aspects of implementing the stop-loss start-gain strategy described above. Can it be done?

(ii)Apart from the practical aspects, does the mathematics of continuous­ time stochastic calculus suggest that the stop-loss start-gain strategy can be implemented? In other words, with $X(T)$ defined by (4.10.46), is it really true that $X(T) = (S(T) - K)^+$?

### (i) Practical aspects

The stop-loss start-gain strategy cannot be implemented in practice for the following reasons:

**Infinite number of transactions:** Brownian motion (and GBM) crosses any level $K$ infinitely many times on $[0,T]$. This means the strategy requires buying and selling the stock infinitely many times, which is physically impossible.

**Infinite transaction costs:** Even if we ignore the impossibility of infinite transactions, any nonzero transaction cost per trade would result in infinite total costs, making the strategy economically unviable.

### (ii) Mathematical analysis

**No — the mathematics shows $X(T) \neq (S(T)-K)^+$.**

By the corrected Itô-Doeblin formula for $f(x) = (x-K)^+$ (Exercise 4.20), which accounts for the local time at level $K$:

$$(S(T)-K)^+ = (S(0)-K)^+ + \int_0^T \mathbb{I}_{(K,\infty)}(S(t))dS(t) + \frac{1}{2}L_K(T)$$

Since $X(0) = 0$ and $r = 0$:

$$X(T) = \int_0^T \mathbb{I}_{(K,\infty)}(S(t))dS(t)$$

Since $S(0) < K$, we have $(S(0)-K)^+ = 0$, so:

$$(S(T)-K)^+ = X(T) + \frac{1}{2}L_K(T)$$

Therefore:

$$X(T) = (S(T)-K)^+ - \frac{1}{2}L_K(T)$$

By Exercise 4.20(vi), $L_K(T) > 0$ for some paths of the Brownian motion. Therefore:

$$X(T) \neq (S(T)-K)^+$$

The stop-loss start-gain strategy fails to replicate the call payoff — it falls short by exactly $\frac{1}{2}L_K(T)$, the local time of the stock price at level $K$. $\blacksquare$
