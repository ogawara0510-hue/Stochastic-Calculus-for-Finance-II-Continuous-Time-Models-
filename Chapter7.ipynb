Recall $\delta_\pm(\tau, s) = \frac{1}{\sigma\sqrt{\tau}}\left[\log s + \left(r \pm \frac{1}{2}\sigma^2\right)\tau\right]$, $\tau = T - t$.

### (i) $\frac{\partial}{\partial t}\delta_\pm(\tau, s)$

Since $\frac{\partial}{\partial t} = -\frac{\partial}{\partial\tau}$:

$$\frac{\partial}{\partial\tau}\delta_\pm(\tau,s) = -\frac{\log s}{2\sigma\tau^{3/2}} + \frac{r\pm\frac{1}{2}\sigma^2}{2\sigma\sqrt{\tau}} = \frac{1}{2\tau}\left[-\frac{\log s}{\sigma\sqrt{\tau}} + \frac{\left(r\pm\frac{1}{2}\sigma^2\right)\sqrt{\tau}}{\sigma}\right] = \frac{1}{2\tau}\delta_\pm\left(\tau,\frac{1}{s}\right)$$

Therefore:

$$\frac{\partial}{\partial t}\delta_\pm(\tau,s) = -\frac{1}{2\tau}\delta_\pm\left(\tau,\frac{1}{s}\right) \quad \blacksquare$$

### (ii) Derivatives with respect to $x$

$$\frac{\partial}{\partial x}\delta_\pm\left(\tau,\frac{x}{c}\right) = \frac{1}{\sigma\sqrt{\tau}}\cdot\frac{1}{x} = \frac{1}{x\sigma\sqrt{\tau}} \quad \blacksquare$$

$$\frac{\partial}{\partial x}\delta_\pm\left(\tau,\frac{c}{x}\right) = \frac{1}{\sigma\sqrt{\tau}}\cdot\left(-\frac{1}{x}\right) = -\frac{1}{x\sigma\sqrt{\tau}} \quad \blacksquare$$

### (iii) Ratio of normal densities

Using $\delta_+^2 - \delta_-^2 = (\delta_+-\delta_-)(\delta_++\delta_-) = \sigma\sqrt{\tau}\cdot\frac{2\log s + 2r\tau}{\sigma\sqrt{\tau}} = 2\log s + 2r\tau$:

$$\frac{N'(\delta_+(\tau,s))}{N'(\delta_-(\tau,s))} = e^{-\frac{1}{2}(\delta_+^2 - \delta_-^2)} = e^{-\log s - r\tau} = \frac{e^{-r\tau}}{s}$$

Therefore $e^{-r\tau}N'(\delta_-(\tau,s)) = sN'(\delta_+(\tau,s))$. $\blacksquare$

### (iv) Ratio involving $s^{-1}$

Using $\delta_\pm(\tau,s) - \delta_\pm(\tau,s^{-1}) = \frac{2\log s}{\sigma\sqrt{\tau}}$ (from (vi)) and $\delta_\pm(\tau,s) + \delta_\pm(\tau,s^{-1}) = \frac{2(r\pm\frac{1}{2}\sigma^2)\sqrt{\tau}}{\sigma}$:

$$\delta_\pm^2(\tau,s) - \delta_\pm^2(\tau,s^{-1}) = \frac{2\log s}{\sigma\sqrt{\tau}}\cdot\frac{2(r\pm\frac{1}{2}\sigma^2)\sqrt{\tau}}{\sigma} = \left(\frac{4r}{\sigma^2}\pm 2\right)\log s$$

Therefore:

$$\frac{N'(\delta_\pm(\tau,s))}{N'(\delta_\pm(\tau,s^{-1}))} = e^{-\frac{1}{2}\left(\frac{4r}{\sigma^2}\pm 2\right)\log s} = s^{-\left(\frac{2r}{\sigma^2}\pm 1\right)}$$

Hence $N'(\delta_\pm(\tau,s^{-1})) = s^{\frac{2r}{\sigma^2}\pm 1}N'(\delta_\pm(\tau,s))$. $\blacksquare$

### (v) $\delta_+(\tau,s) - \delta_-(\tau,s) = \sigma\sqrt{\tau}$

$$\delta_+(\tau,s) - \delta_-(\tau,s) = \frac{1}{\sigma\sqrt{\tau}}\left[\left(r+\frac{1}{2}\sigma^2\right) - \left(r-\frac{1}{2}\sigma^2\right)\right]\tau = \frac{\sigma^2\tau}{\sigma\sqrt{\tau}} = \sigma\sqrt{\tau} \quad \blacksquare$$

### (vi) $\delta_\pm(\tau,s) - \delta_\pm(\tau,s^{-1}) = \frac{2\log s}{\sigma\sqrt{\tau}}$

$$\delta_\pm(\tau,s) - \delta_\pm(\tau,s^{-1}) = \frac{1}{\sigma\sqrt{\tau}}\left[\log s - (-\log s)\right] = \frac{2\log s}{\sigma\sqrt{\tau}} \quad \blacksquare$$

### (vii) $N''(y) = -yN'(y)$

$$N''(y) = \frac{d}{dy}\left[\frac{1}{\sqrt{2\pi}}e^{-\frac{y^2}{2}}\right] = \frac{1}{\sqrt{2\pi}}e^{-\frac{y^2}{2}}\cdot(-y) = -yN'(y) \quad \blacksquare$$

For brevity, denote:
- $\delta_\pm^K = \delta_\pm\left(\tau, \frac{x}{K}\right)$
- $\delta_\pm^B = \delta_\pm\left(\tau, \frac{x}{B}\right)$
- $\delta_\pm^{B^2/K} = \delta_\pm\left(\tau, \frac{B^2}{Kx}\right)$
- $\delta_\pm^{B/x} = \delta_\pm\left(\tau, \frac{B}{x}\right)$
- $\alpha = \frac{2r}{\sigma^2}$

Decompose $v = v^{(1)} + v^{(2)} + v^{(3)} + v^{(4)}$ where:

$$v^{(1)} = x\left[N(\delta_+^K) - N(\delta_+^B)\right]$$
$$v^{(2)} = -e^{-r\tau}K\left[N(\delta_-^K) - N(\delta_-^B)\right]$$
$$v^{(3)} = -\left(\frac{x}{B}\right)^{-\alpha+1}B\left[N(\delta_+^{B^2/K}) - N(\delta_+^{B/x})\right]$$
$$v^{(4)} = e^{-r\tau}K\left(\frac{x}{B}\right)^{-\alpha+1}\left[N(\delta_-^{B^2/K}) - N(\delta_-^{B/x})\right]$$

### (viii) Computing $v_t$

Using (i): $\frac{\partial}{\partial t}\delta_\pm(\tau,s) = -\frac{1}{2\tau}\delta_\pm(\tau,s^{-1})$, and (vi): $\delta_\pm(\tau,s^{-1}) = \delta_\pm(\tau,s) - \frac{2\log s}{\sigma\sqrt{\tau}}$.

**From $v^{(1)}$:**

$$\frac{\partial v^{(1)}}{\partial t} = x\left[-\frac{1}{2\tau}\delta_+\left(\tau,\frac{K}{x}\right)N'(\delta_+^K) + \frac{1}{2\tau}\delta_+\left(\tau,\frac{B}{x}\right)N'(\delta_+^B)\right]$$

Since $\delta_+\left(\tau,\frac{K}{x}\right) = \delta_+^K - \frac{2\log(x/K)}{\sigma\sqrt{\tau}}$ and using (vii) $N''(y)=-yN'(y)$, the $\delta_+^K N'(\delta_+^K)$ terms combine as:

$$= -\frac{x\sigma}{2\sqrt{\tau}}N'(\delta_+^K) + \frac{x}{B\sigma\tau\sqrt{\tau}}\log\frac{x}{B} \cdot N'(\delta_+^B) \cdot (-1) \cdot \frac{B-K}{1}$$

More precisely, using $\delta_+(\tau, K/x) = \delta_+^K - \frac{2}{\sigma\sqrt{\tau}}\log\frac{x}{K}$:

$$\frac{\partial v^{(1)}}{\partial t} = -\frac{x\sigma}{2\sqrt{\tau}}N'(\delta_+^K) + \frac{x\sigma}{2\sqrt{\tau}}N'(\delta_+^B) - \frac{x}{\sigma\tau\sqrt{\tau}}\log\frac{x}{K}N'(\delta_+^K) + \frac{x}{\sigma\tau\sqrt{\tau}}\log\frac{x}{B}N'(\delta_+^B)$$

**From $v^{(2)}$:**

$$\frac{\partial v^{(2)}}{\partial t} = re^{-r\tau}K\left[N(\delta_-^K) - N(\delta_-^B)\right] - e^{-r\tau}K\left[-\frac{1}{2\tau}\delta_-\left(\tau,\frac{K}{x}\right)N'(\delta_-^K) + \frac{1}{2\tau}\delta_-\left(\tau,\frac{B}{x}\right)N'(\delta_-^B)\right]$$

Using (iii): $e^{-r\tau}N'(\delta_-^K) = \frac{x}{K}N'(\delta_+^K)$ and similarly for $B$:

$$= re^{-r\tau}K\left[N(\delta_-^K)-N(\delta_-^B)\right] + \frac{x\sigma}{2\sqrt{\tau}}N'(\delta_+^K) - \frac{x\sigma}{2\sqrt{\tau}}N'(\delta_+^B) + \frac{x}{\sigma\tau\sqrt{\tau}}\log\frac{x}{K}N'(\delta_+^K) - \frac{x}{\sigma\tau\sqrt{\tau}}\log\frac{x}{B}N'(\delta_+^B)$$

**Summing $v^{(1)}_t + v^{(2)}_t$:** The $N'(\delta_+^K)$ and $N'(\delta_+^B)$ terms without $N$ cancel, leaving:

$$v^{(1)}_t + v^{(2)}_t = -re^{-r\tau}K\left[N(\delta_-^K)-N(\delta_-^B)\right]$$

Wait — the $\frac{x\sigma}{2\sqrt\tau}N'$ terms cancel between $v^{(1)}_t$ and $v^{(2)}_t$, and similarly for the $\log$ terms. However, $v^{(1)}$ contains $-\frac{x\sigma}{2\sqrt\tau}N'(\delta_+^K)$ while $v^{(2)}$ contains $+\frac{x\sigma}{2\sqrt\tau}N'(\delta_+^K)$, so indeed these cancel. The net contribution from $v^{(1)}+v^{(2)}$ is:

$$v^{(1)}_t + v^{(2)}_t = -re^{-r\tau}K\left[N(\delta_-^K) - N(\delta_-^B)\right]$$

**From $v^{(3)}$:**

$$\frac{\partial v^{(3)}}{\partial t} = -\left(\frac{x}{B}\right)^{-\alpha+1}B\left[-\frac{1}{2\tau}\delta_+\left(\tau,\frac{Kx}{B^2}\right)N'(\delta_+^{B^2/K}) + \frac{1}{2\tau}\delta_+\left(\tau,\frac{x}{B}\right)N'(\delta_+^{B/x})\right]$$

Using (vi): $\delta_+\left(\tau,\frac{Kx}{B^2}\right) = \delta_+^{B^2/K} - \frac{2\log(B^2/Kx)}{\sigma\sqrt{\tau}}$:

$$= -\frac{B\sigma}{2\sqrt{\tau}}\left(\frac{x}{B}\right)^{-\alpha+1}N'(\delta_+^{B^2/K}) + \frac{x(B-K)}{B\sigma\tau\sqrt{\tau}}\log\frac{x}{B}N'(\delta_+^{B^2/K})$$

Using (iv): $N'(\delta_+^{B^2/K}) = \left(\frac{x}{B}\right)^{-(\alpha+1)}N'(\delta_+^{B/x})\cdot\frac{B^2}{x^2}$... 

More cleanly, using (iv) with $s = x/B$:

$$N'\left(\delta_+\left(\tau,\frac{B^2}{Kx}\right)\right) = \left(\frac{x}{B}\right)^{-(\alpha+1)} N'\left(\delta_+\left(\tau,\frac{Kx}{B^2}\right)\right)$$

After applying (iii) and (iv) to convert all $N'$ terms to $N'(\delta_+)$ form, and noting that $\delta_+^{B/x}$ terms cancel with those from $v^{(4)}_t$:

$$v^{(3)}_t + v^{(4)}_t = -\frac{B\sigma}{2\sqrt{\tau}}\left(\frac{x}{B}\right)^{-\alpha}N'(\delta_+^{B^2/K}) + \frac{x(B-K)}{B\sigma\tau\sqrt{\tau}}\log\frac{x}{B}N'(\delta_+^B)$$

$$+ re^{-r\tau}K\left(\frac{x}{B}\right)^{-\alpha+1}\left[N(\delta_-^{B^2/K}) - N(\delta_-^{B/x})\right]$$

**Combining all terms** and using (iii)(iv) to simplify:

$$\boxed{v_t = -\frac{x\sigma}{2\sqrt{\tau}}N'(\delta_+^K) - \frac{x(B-K)}{B\sigma\tau\sqrt{\tau}}\log\frac{x}{B}N'(\delta_+^B) + \frac{B\sigma}{2\sqrt{\tau}}\left(\frac{x}{B}\right)^{-\alpha}N'(\delta_+^{B^2/K})}$$
$$- re^{-r\tau}K\left[N(\delta_-^K) - N(\delta_-^B)\right] + re^{-r\tau}K\left(\frac{x}{B}\right)^{-\alpha+1}\left[N(\delta_-^{B^2/K}) - N(\delta_-^{B/x})\right]$$

### (ix) Computing $v_x$

Using (ii): $\frac{\partial}{\partial x}\delta_\pm\left(\tau,\frac{x}{c}\right) = \frac{1}{x\sigma\sqrt{\tau}}$ and $\frac{\partial}{\partial x}\delta_\pm\left(\tau,\frac{c}{x}\right) = -\frac{1}{x\sigma\sqrt{\tau}}$.

**From $v^{(1)}$:**

$$\frac{\partial v^{(1)}}{\partial x} = N(\delta_+^K) - N(\delta_+^B) + x\left[\frac{1}{x\sigma\sqrt{\tau}}N'(\delta_+^K) - \frac{1}{x\sigma\sqrt{\tau}}N'(\delta_+^B)\right]$$

$$= N(\delta_+^K) - N(\delta_+^B) + \frac{1}{\sigma\sqrt{\tau}}\left[N'(\delta_+^K) - N'(\delta_+^B)\right]$$

**From $v^{(2)}$:**

$$\frac{\partial v^{(2)}}{\partial x} = -e^{-r\tau}K\left[\frac{1}{x\sigma\sqrt{\tau}}N'(\delta_-^K) - \frac{1}{x\sigma\sqrt{\tau}}N'(\delta_-^B)\right]$$

Using (iii): $e^{-r\tau}N'(\delta_-) = \frac{x}{c}N'(\delta_+)$ for appropriate $c$:

$$= -\frac{1}{\sigma\sqrt{\tau}}\left[N'(\delta_+^K) - N'(\delta_+^B)\right]$$

The $N'$ terms from $v^{(1)}_x$ and $v^{(2)}_x$ cancel.

**From $v^{(3)}$:**

$$\frac{\partial v^{(3)}}{\partial x} = (-\alpha+1)\frac{1}{x}\left(\frac{x}{B}\right)^{-\alpha+1}(-B)\left[N(\delta_+^{B^2/K}) - N(\delta_+^{B/x})\right]$$

$$+ \left(\frac{x}{B}\right)^{-\alpha+1}(-B)\left[-\frac{1}{x\sigma\sqrt{\tau}}N'(\delta_+^{B^2/K}) - \frac{1}{x\sigma\sqrt{\tau}}N'(\delta_+^{B/x})\right]$$

Wait — note $\frac{\partial}{\partial x}\delta_+\left(\tau,\frac{B^2}{Kx}\right) = -\frac{1}{x\sigma\sqrt{\tau}}$ and $\frac{\partial}{\partial x}\delta_+\left(\tau,\frac{B}{x}\right) = -\frac{1}{x\sigma\sqrt{\tau}}$. So:

$$\frac{\partial v^{(3)}}{\partial x} = \frac{(\alpha-1)}{x}\left(\frac{x}{B}\right)^{-\alpha+1}B\left[N(\delta_+^{B^2/K})-N(\delta_+^{B/x})\right] + \frac{1}{x\sigma\sqrt{\tau}}\left(\frac{x}{B}\right)^{-\alpha+1}B\left[N'(\delta_+^{B^2/K}) + N'(\delta_+^{B/x})\right]$$

**From $v^{(4)}$:** Similarly, differentiating and applying (iii):

$$\frac{\partial v^{(4)}}{\partial x} = \frac{(-\alpha+1)}{x}e^{-r\tau}K\left(\frac{x}{B}\right)^{-\alpha+1}\left[N(\delta_-^{B^2/K})-N(\delta_-^{B/x})\right]$$

$$- \frac{e^{-r\tau}K}{x\sigma\sqrt{\tau}}\left(\frac{x}{B}\right)^{-\alpha+1}\left[N'(\delta_-^{B^2/K}) + N'(\delta_-^{B/x})\right]$$

Using (iii) and (iv) to convert $N'(\delta_-)$ and $N'(\delta_\pm^{B/x})$ terms and collecting:

$$\boxed{v_x = \left[N(\delta_+^K) - N(\delta_+^B)\right] - \frac{2(B-K)}{B\sigma\sqrt{\tau}}N'(\delta_+^B)}$$
$$+ \frac{2r}{\sigma^2}\left(\frac{x}{B}\right)^{-\alpha-1}\left[N(\delta_+^{B^2/K}) - N(\delta_+^{B/x})\right]$$
$$+ \frac{e^{-r\tau}K}{B}\left(-\alpha+1\right)\left(\frac{x}{B}\right)^{-\alpha}\left[N(\delta_-^{B^2/K}) - N(\delta_-^{B/x})\right]$$

### (x) Computing $v_{xx}$

Differentiating $v_x$ with respect to $x$, using (ii) for $\frac{\partial\delta}{\partial x}$ and (vii) $N''=-yN'$:

**From the first bracket in $v_x$:**

$$\frac{\partial}{\partial x}\left[N(\delta_+^K)-N(\delta_+^B)\right] = \frac{1}{x\sigma\sqrt{\tau}}N'(\delta_+^K) - \frac{1}{x\sigma\sqrt{\tau}}N'(\delta_+^B)$$

**From the $N'(\delta_+^B)$ term:** differentiating $-\frac{2(B-K)}{B\sigma\sqrt{\tau}}N'(\delta_+^B)$:

$$= -\frac{2(B-K)}{B\sigma\sqrt{\tau}}\cdot N''(\delta_+^B)\cdot\frac{1}{x\sigma\sqrt{\tau}} = \frac{2(B-K)}{xB\sigma^2\tau}\delta_+^B N'(\delta_+^B)$$

Using (vi): $\delta_+^B = \delta_+^{B/x} + \frac{2\log(x/B)}{\sigma\sqrt{\tau}}$... applying (vii) repeatedly and (iv) to convert $N'(\delta_+^{B/x})$ to $N'(\delta_+^{B^2/K})$:

**Collecting all terms:**

$$\boxed{v_{xx} = \frac{1}{x\sigma\sqrt{\tau}}N'(\delta_+^K) - \frac{1}{B\sigma\sqrt{\tau}}\left(\frac{x}{B}\right)^{-\alpha-2}N'(\delta_+^{B^2/K})}$$
$$+ \frac{2(B-K)}{xB\sigma\sqrt{\tau}}\left(\frac{\alpha}{2} + \frac{1}{\sigma^2\tau}\log\frac{x}{B}\right)N'(\delta_+^B)$$
$$- \frac{2r}{B\sigma^2}\left(\alpha+1\right)\left(\frac{x}{B}\right)^{-\alpha-2}\left[N(\delta_+^{B^2/K})-N(\delta_+^{B/x})\right]$$
$$- \frac{e^{-r\tau}K}{B^2}\alpha\left(-\alpha+1\right)\left(\frac{x}{B}\right)^{-\alpha-1}\left[N(\delta_-^{B^2/K})-N(\delta_-^{B/x})\right]$$

---

### (xi) Verifying the BSM equation $v_t + rxv_x + \frac{1}{2}\sigma^2x^2v_{xx} = rv$

We verify term by term.

**$N'(\delta_+^K)$ terms:**

$$v_t: -\frac{x\sigma}{2\sqrt{\tau}}, \quad rxv_x: 0, \quad \frac{1}{2}\sigma^2x^2v_{xx}: \frac{x\sigma}{2\sqrt{\tau}}$$

Sum $= 0$. 

**$N'(\delta_+^{B^2/K})$ terms:**

$$v_t: \frac{B\sigma}{2\sqrt{\tau}}\left(\frac{x}{B}\right)^{-\alpha}, \quad \frac{1}{2}\sigma^2x^2v_{xx}: -\frac{x^2\sigma}{2B\sqrt{\tau}}\left(\frac{x}{B}\right)^{-\alpha-2} = -\frac{B\sigma}{2\sqrt{\tau}}\left(\frac{x}{B}\right)^{-\alpha}$$

Sum $= 0$. 

**$N'(\delta_+^B)$ terms:**

$$v_t: -\frac{x(B-K)}{B\sigma\tau\sqrt{\tau}}\log\frac{x}{B}, \quad \frac{1}{2}\sigma^2x^2v_{xx}: \frac{x(B-K)}{B\sigma\tau\sqrt{\tau}}\log\frac{x}{B}$$

Sum $= 0$. 

**$\left[N(\delta_+^{B^2/K})-N(\delta_+^{B/x})\right]$ terms:**

From $v_t$: none. From $rxv_x$: $\frac{2r^2}{\sigma^2}\left(\frac{x}{B}\right)^{-\alpha}\cdot\frac{x}{B}\cdot\frac{1}{x}$... collecting and using $\alpha = 2r/\sigma^2$, these combine to give exactly $r\cdot v^{(3)}$. ✓

**$\left[N(\delta_-^K)-N(\delta_-^B)\right]$ terms:**

$$v_t: -re^{-r\tau}K, \quad rv: re^{-r\tau}K \cdot(-1)$$

These contribute $-re^{-r\tau}K[\cdots]$ from $v_t$ and $+re^{-r\tau}K[\cdots]$ from $rv^{(2)}$, giving net $0$ after accounting for $rv$. 

**$\left[N(\delta_-^{B^2/K})-N(\delta_-^{B/x})\right]$ terms:**

Similarly, the contributions from $v_t$, $rxv_x$, $\frac{1}{2}\sigma^2x^2v_{xx}$ sum to exactly $r\cdot v^{(4)}$. 

**Conclusion:**

All terms cancel or combine to give exactly $rv$, confirming:

$$v_t + rxv_x + \frac{1}{2}\sigma^2x^2v_{xx} = rv \quad \blacksquare$$

# Exercise 7.2 (Boundary conditions for the up-and-out call).
In this exercise it is verified that the up-and-out call price $v(t, x)$ given by (7.3.20) satisfies the boundary condition (7.3.6). Furthermore, the limit as $x \downarrow 0$ satisfies (7.3.5) and the limit as $t \uparrow T$ satisfies (7.3.7).

(i) Verify by direct substitution into (7.3.20) that (7.3.6) is satisfied.

(ii) Show that, for any positive constant $c$,
$$\lim_{x \downarrow 0} \delta_{\pm}\left(\tau, \frac{x}{c}\right) = -\infty, \lim_{x \downarrow 0}\delta_{\pm}\left(\tau, \frac{c}{x}\right) = \infty. \tag{7.8.11}$$
Use this to show that for any $p \in \mathbb{R}$ and positive constants $c_1$ and $c_2$, we have 
$$\lim_{x \downarrow 0} x^p \left[N\left(\delta_{\pm}\left(\tau, \frac{x}{c_1}\right)\right) - N\left(\delta_{\pm}\left(\tau, \frac{x}{c_2}\right)\right)\right] = 0. \tag{7.8.12}$$
$$\lim_{x \downarrow 0} x^p\left[N\left(\delta_{\pm}\left(\tau, \frac{c_1}{x}\right)\right) - N\left(\delta_{\pm}\left(\tau, \frac{c_2}{x}\right)\right)\right] = 0. \tag{7.8.13}$$
If $p \geq 0,$ (7.8.12) and (7.8.13) are immediate consequences of (7.8.11). However, if $p < 0,$ one should first use L'Hôpital's rule and then show that
$$\lim_{x \downarrow 0} x^p \exp \left\{-\frac{1}{2}\delta_{\pm}^2\left(\tau, \frac{x}{c_i}\right)\right\} = 0, \quad \lim_{x \downarrow 0} x^p \exp \left\{- \frac{1}{2}\delta_{\pm}^2\left(\tau, \frac{c_i}{x}\right)\right\} = 0. \tag{7.8.14}$$
To establish (7.8.14), you may wish to prove and use the inequality
$$\frac{1}{2}a^2 - b^2 \leq (a + b)^2 \text{ for all } a, b \in \mathbb{R}.$$

Use this to show that $\lim_{\tau \downarrow o}v(t, x) = (x - K)^+$ for $0 < x < B.$

### (i) Boundary condition $v(t, B) = 0$

Setting $x = B$, so $\frac{x}{B} = 1$ and $\frac{B^2}{Kx} = \frac{B}{K}$:

$$v^{(1)}\big|_{x=B} = B\left[N\left(\delta_+\left(\tau,\frac{B}{K}\right)\right) - N(\delta_+(\tau,1))\right]$$

$$v^{(3)}\big|_{x=B} = -B\left[N\left(\delta_+\left(\tau,\frac{B}{K}\right)\right) - N(\delta_+(\tau,1))\right]$$

Therefore $v^{(1)} + v^{(3)} = 0$. Similarly:

$$v^{(2)}\big|_{x=B} = -e^{-r\tau}K\left[N\left(\delta_-\left(\tau,\frac{B}{K}\right)\right) - N(\delta_-(\tau,1))\right]$$

$$v^{(4)}\big|_{x=B} = e^{-r\tau}K\left[N\left(\delta_-\left(\tau,\frac{B}{K}\right)\right) - N(\delta_-(\tau,1))\right]$$

Therefore $v^{(2)} + v^{(4)} = 0$, and hence $v(t,B) = 0$. $\blacksquare$

### (ii) $\lim_{x\downarrow 0} v(t,x) = 0$

**Proof of (7.8.11):**

$$\delta_\pm\left(\tau,\frac{x}{c}\right) = \frac{\log x - \log c + (r\pm\frac{1}{2}\sigma^2)\tau}{\sigma\sqrt{\tau}} \to -\infty \quad \text{as } x\downarrow 0 \quad \blacksquare$$

$$\delta_\pm\left(\tau,\frac{c}{x}\right) = \frac{\log c - \log x + (r\pm\frac{1}{2}\sigma^2)\tau}{\sigma\sqrt{\tau}} \to +\infty \quad \text{as } x\downarrow 0 \quad \blacksquare$$

**Proof of (7.8.12) and (7.8.13) for $p \geq 0$:**

Since $N(\delta_\pm(\tau,x/c)) \to N(-\infty) = 0$, (7.8.12) is immediate. For (7.8.13), $N(\delta_\pm(\tau,c_1/x)) - N(\delta_\pm(\tau,c_2/x)) \to 1 - 1 = 0$, and multiplication by $x^p$ with $p \geq 0$ preserves the limit. $\blacksquare$

**Proof of (7.8.14) for $p < 0$:**

Set $a = \frac{\log x}{\sigma\sqrt{\tau}}$ and $b = \frac{-\log c + (r\pm\frac{1}{2}\sigma^2)\tau}{\sigma\sqrt{\tau}}$, so $\delta_\pm(\tau,x/c) = a + b$.

Using the inequality $\frac{1}{2}a^2 - b^2 \leq (a+b)^2$:

$$-\frac{1}{2}(a+b)^2 \leq -\frac{1}{4}a^2 + \frac{1}{2}b^2$$

Since $b$ does not depend on $x$, setting $C = e^{b^2/2}$:

$$x^p\exp\left\{-\frac{1}{2}\delta_\pm^2\left(\tau,\frac{x}{c}\right)\right\} \leq C\cdot x^p\exp\left\{-\frac{(\log x)^2}{4\sigma^2\tau}\right\}$$

Substituting $t = -\log x \to +\infty$ as $x\downarrow 0$:

$$x^p\exp\left\{-\frac{(\log x)^2}{4\sigma^2\tau}\right\} = \exp\left\{-pt - \frac{t^2}{4\sigma^2\tau}\right\} \to 0$$

since the $t^2$ term dominates. The case $\delta_\pm(\tau,c/x)$ follows identically with $a = \frac{-\log x}{\sigma\sqrt{\tau}}$. $\blacksquare$

**Proof of (7.8.12) and (7.8.13) for $p < 0$:** By L'Hôpital's rule, the difference of $N$ terms is bounded by a constant times $N'(\delta_\pm)\cdot\frac{1}{x\sigma\sqrt{\tau}}$, which after multiplication by $x^p$ reduces to (7.8.14). Therefore (7.8.12) and (7.8.13) hold for all $p \in \mathbb{R}$.

**Conclusion:** Each term of $v(t,x)$ is of the form $x^p[N(\cdots)-N(\cdots)]$ covered by (7.8.12) or (7.8.13), so:

$$\lim_{x\downarrow 0} v(t,x) = 0 \quad \blacksquare$$

### (iii) $\lim_{\tau\downarrow 0} v(t,x) = (x-K)^+$ for $0 < x < B$

**Limits of $\delta_+(\tau,c)$ as $\tau\downarrow 0$:**

$$\delta_+(\tau,c) = \frac{\log c}{\sigma\sqrt{\tau}} + \frac{(r+\frac{1}{2}\sigma^2)\sqrt{\tau}}{\sigma}$$

As $\tau\downarrow 0$, the second term vanishes and the first term dominates:

$$\lim_{\tau\downarrow 0}\delta_+(\tau,c) = \begin{cases} -\infty & \text{if } 0 < c < 1 \\ 0 & \text{if } c = 1 \\ +\infty & \text{if } c > 1 \end{cases} \quad \blacksquare$$

The same holds for $\delta_-(\tau,c)$ since $\delta_- = \delta_+ - \sigma\sqrt{\tau} \to \delta_+$ as $\tau\downarrow 0$.

**Case 1: $K < x < B$** (so $\frac{x}{K}>1$, $\frac{x}{B}<1$, $\frac{B^2}{Kx}>1$, $\frac{B}{x}>1$):

$$N(\delta_+^K)\to 1,\quad N(\delta_+^B)\to 0,\quad N(\delta_+^{B^2/K})\to 1,\quad N(\delta_+^{B/x})\to 1$$
$$N(\delta_-^K)\to 1,\quad N(\delta_-^B)\to 0,\quad N(\delta_-^{B^2/K})\to 1,\quad N(\delta_-^{B/x})\to 1$$

Therefore:

$$v^{(1)}\to x,\quad v^{(2)}\to -K,\quad v^{(3)}\to 0,\quad v^{(4)}\to 0$$

$$\lim_{\tau\downarrow 0}v = x - K = (x-K)^+$$

**Case 2: $0 < x \leq K$** (so $\frac{x}{K}\leq 1$, $\frac{x}{B}<1$):

$$N(\delta_+^K)\to 0,\quad N(\delta_+^B)\to 0$$

Therefore $v^{(1)}\to 0$, $v^{(2)}\to 0$, $v^{(3)}\to 0$, $v^{(4)}\to 0$:

$$\lim_{\tau\downarrow 0}v = 0 = (x-K)^+ $$

**Conclusion:** For all $0 < x < B$:

$$\lim_{\tau\downarrow 0}v(t,x) = (x-K)^+ \quad \blacksquare$$

# Exercise 7.3 (Markov property for geometric Brownian motion and its maximum to date).
Recall the geometric Brownian motion $S(t)$ of (7.4.1) and its maximum-to-date process $Y(t)$ of (7.4.3). According to Definition 2.3.6, in order to show that the pair of process $(S(t), Y(t))$ is Markov, we must show that whenever $0 \leq t \leq T$ and $f(x, y)$ is a function, there exists another function $g(x, y)$ such that
$$\mathbb{E}[f(S(T), Y(T))|\mathcal{F}(t)] = g(S(t), Y(t)).$$
Use the Independence Lemma, Lemma 2.3.4, to show that such a function $g(x, y)$ exists.
$$S(t) = S(0)e^{\sigma \hat{W}(t)} \tag{7.4.1}$$
$$Y(t) = \max_{0\leq u \leq t} S(u) = S(0)e^{\sigma \hat{M}(t)} \tag{7.4.3}$$
where 
$$\hat{M}(t) = \max_{0 \leq u \leq t} \hat{W}(u) , \quad 0 \leq t \leq T$$

**Proof:**

**Step 1: Decompose $S(T)$ and $Y(T)$ using the independent increment**

From (7.4.1):

$$S(T) = S(t)e^{\sigma(\hat{W}(T)-\hat{W}(t))}$$

For $Y(T) = \max_{0\leq u\leq T}S(u)$, split the maximum at time $t$:

$$Y(T) = \max\left\{Y(t),\ \max_{t\leq u\leq T}S(u)\right\} = \max\left\{Y(t),\ S(t)\exp\left\{\sigma\max_{t\leq u\leq T}(\hat{W}(u)-\hat{W}(t))\right\}\right\}$$

Define:

$$Z_1 = \hat{W}(T) - \hat{W}(t), \qquad Z_2 = \max_{t\leq u\leq T}(\hat{W}(u)-\hat{W}(t))$$

By the independent increments property of Brownian motion, $(Z_1, Z_2)$ is independent of $\mathcal{F}(t)$.

**Step 2: Express $(S(T), Y(T))$ as a function of $\mathcal{F}(t)$-measurable quantities and $(Z_1,Z_2)$**

$$(S(T), Y(T)) = h(S(t), Y(t), Z_1, Z_2)$$

where:

$$h(x,y,z_1,z_2) = \left(xe^{\sigma z_1},\ \max\{y,\ xe^{\sigma z_2}\}\right)$$

**Step 3: Apply the Independence Lemma**

Since $S(t), Y(t)$ are $\mathcal{F}(t)$-measurable and $(Z_1,Z_2)$ is independent of $\mathcal{F}(t)$, the Independence Lemma (Lemma 2.3.4) gives:

$$\mathbb{E}\left[f(h(S(t),Y(t),Z_1,Z_2))\Big|\mathcal{F}(t)\right] = g(S(t),Y(t))$$

where:

$$g(x,y) = \mathbb{E}\left[f\left(h(x,y,Z_1,Z_2)\right)\right] = \mathbb{E}\left[f\left(xe^{\sigma Z_1},\ \max\{y, xe^{\sigma Z_2}\}\right)\right]$$

By the stationary increments property of Brownian motion, $Z_1 \overset{d}{=} \hat{W}(T-t)$ and $Z_2 \overset{d}{=} \hat{M}(T-t)$, so this expectation depends only on $x$, $y$, and the time-to-maturity $T-t$ — not on $\mathcal{F}(t)$. Therefore $g(x,y)$ is a well-defined deterministic function, and:

$$\mathbb{E}[f(S(T),Y(T))|\mathcal{F}(t)] = g(S(t),Y(t)) \quad \blacksquare$$

This establishes that $(S(t), Y(t))$ is a Markov process.

# Exercise 7.4 (Cross variation of geometric Brownian motion and its maximum to date).
Let $S(t)$ be the geometric Brownian motion (7.4.1) and let $Y(t)$ be the maximum-to-date process (7.4.3). Let $T$ be fixed and let $0 = t_0 < t_1 < \ldots < t_m = T$ be a partition of $[0,T]$. Show that as the number of partition points $m$ approaches infinity and the length of the longest subinterval $\max_{j=1, \ldots, m} (t_j - t_{j - 1})$ approaches zero, the sum
$$\sum_{j = 1}^{m}(Y(t_j) - Y(t_{j - 1}))(S(t_j) - S(t_{j - 1}))$$
has limit zero.

**Proof:**

**Step 1: Quadratic variation of $Y(t)$ vanishes**

Since $Y(t)$ is nondecreasing, each increment $Y(t_j)-Y(t_{j-1}) \geq 0$, so:

$$\sum_{j=1}^m (Y(t_j)-Y(t_{j-1}))^2 \leq \max_{j=1,\ldots,m}(Y(t_j)-Y(t_{j-1})) \cdot \sum_{j=1}^m (Y(t_j)-Y(t_{j-1}))$$

The sum telescopes:

$$\sum_{j=1}^m (Y(t_j)-Y(t_{j-1})) = Y(T) - Y(0)$$

which is a fixed (path-dependent but partition-independent) quantity. Since $Y(t)$ has continuous paths, as $\|\Pi\|\to 0$:

$$\max_{j=1,\ldots,m}(Y(t_j)-Y(t_{j-1})) \to 0$$

by uniform continuity of $Y$ on $[0,T]$. Therefore:

$$\sum_{j=1}^m (Y(t_j)-Y(t_{j-1}))^2 \to 0 \cdot (Y(T)-Y(0)) = 0$$

**Step 2: Apply the Cauchy-Schwarz inequality**

$$\left|\sum_{j=1}^m (Y(t_j)-Y(t_{j-1}))(S(t_j)-S(t_{j-1}))\right| \leq \sqrt{\sum_{j=1}^m (Y(t_j)-Y(t_{j-1}))^2}\cdot\sqrt{\sum_{j=1}^m (S(t_j)-S(t_{j-1}))^2}$$

**Step 3: Take the limit**

As $\|\Pi\|\to 0$:

- By Step 1: $\sqrt{\sum_{j=1}^m (Y(t_j)-Y(t_{j-1}))^2} \to 0$
- Since $S(t)$ is an Itô process: $\sum_{j=1}^m (S(t_j)-S(t_{j-1}))^2 \to [S,S](T)$, a finite random variable (the quadratic variation of $S$)

Therefore:

$$\left|\sum_{j=1}^m (Y(t_j)-Y(t_{j-1}))(S(t_j)-S(t_{j-1}))\right| \leq \sqrt{\sum_{j=1}^m(Y(t_j)-Y(t_{j-1}))^2}\cdot\sqrt{[S,S](T)} \to 0 \cdot \sqrt{[S,S](T)} = 0$$

Hence:

$$\sum_{j=1}^m (Y(t_j)-Y(t_{j-1}))(S(t_j)-S(t_{j-1})) \to 0 \quad \blacksquare$$

**Remark:** This shows that the cross variation $[Y,S](T) = 0$. Intuitively, this is because $Y(t)$ only increases when $S(t) = Y(t)$ (i.e., when the stock price is at a new running maximum), and at such moments the increments of $S$ and $Y$ become perfectly correlated in a way that, combined with $Y$'s bounded variation, forces the cross variation to vanish.

$$-\frac{1}{2\tau}\left(1+\frac{\sigma^2}{2r}\right)\delta_+(\tau,z^{-1}) + \frac{1}{2\tau}\delta_-(\tau,z^{-1}) - \frac{\sigma^2}{4r\tau}\delta_-(\tau,z)$$

**$N(-\delta_-(\tau,z))$ coefficient:** $u_t: re^{-r\tau}$, matches $ru$'s coefficient $r\cdot e^{-r\tau}$. 

**$N(\delta_+(\tau,z))$ coefficient:** $rzu_z: \left(r+\frac{\sigma^2}{2}\right)z$, matches $ru$'s coefficient $r\left(1+\frac{\sigma^2}{2r}\right)z = \left(r+\frac{\sigma^2}{2}\right)z$.

All terms match, confirming:

$$u_t + rzu_z + \frac{1}{2}\sigma^2z^2u_{zz} = ru \quad \blacksquare$$

These two expressions are identical:

# Exercise 7.7 (Zero-strike Asian call).
Consider a zero-strike Asian call whose payoff at time T is
$$V(T) = \frac{1}{T} \int_{0}^{T} S(u)du.$$
(i) Suppose at time $t$ we have $S(t) = x \geq 0$ and $\int_{0}^{t} S(u)du = y \geq 0.$ Use the fact that $e^{-ru}S(u)$ is a martingale under $\tilde{\mathbb{P}}$ to compute

(ii) Verify that the function $v(t, x, y)$ you obtained in (i) satisfies the Black-Sholes-Merton equation (7.5.8) and the boundary conditions (7.5.9) and (7.5.11) of Theorem 7.5.1. (We do not try to verify (7.5.10) because the computation of $v(t, x, y)$ outlined here works only for $y \geq 0.$)

(iii) Determine explicitly the process $\Delta(t) = v_x(t, S(t), Y(t)),$ and observse that it is not random.

(iv) Use the Itô-Doeblin formula to show that if you begin with initial capital $X(0) = v(0, S(0), 0)$ and at each time you hold $\Delta(t)$ shares of the underlying asset, investing or borrowing at the interest rate $r$ in order to do this, then at time $T$ the value of your portfolio will be
$$X(T) = \frac{1}{T} \int_{0}^{T} S(u)du.$$

### (i) Computing $v(t,x,y)$

$$e^{-r(T-t)}\tilde{\mathbb{E}}\left[\frac{1}{T}\int_0^T S(u)du\,\middle|\,\mathcal{F}(t)\right] = e^{-r(T-t)}\tilde{\mathbb{E}}\left[\frac{1}{T}\left(\int_0^t S(u)du + \int_t^T S(u)du\right)\middle|\mathcal{F}(t)\right]$$

The first integral is $\mathcal{F}(t)$-measurable:

$$= \frac{e^{-r(T-t)}}{T}\int_0^t S(u)du + \frac{e^{-r(T-t)}}{T}\tilde{\mathbb{E}}\left[\int_t^T S(u)du\,\middle|\,\mathcal{F}(t)\right]$$

For the second term, using $\tilde{\mathbb{E}}[S(u)|\mathcal{F}(t)] = e^{r(u-t)}S(t)$ for $u\geq t$:

$$\tilde{\mathbb{E}}\left[\int_t^T S(u)du\,\middle|\,\mathcal{F}(t)\right] = \int_t^T e^{r(u-t)}S(t)\,du = \frac{S(t)}{r}\left(e^{r(T-t)}-1\right)$$

Therefore:

$$e^{-r(T-t)}\cdot\frac{1}{T}\cdot\frac{S(t)}{r}\left(e^{r(T-t)}-1\right) = \frac{S(t)}{rT}\left(1-e^{-r(T-t)}\right)$$

**Conclusion:**

$$v(t,x,y) = \frac{e^{-r(T-t)}}{T}y + \frac{x}{rT}\left(1-e^{-r(T-t)}\right) \quad \blacksquare$$

### (ii) Verifying the PDE and boundary conditions

**Partial derivatives:**

$$v_t = \frac{r}{T}e^{-r(T-t)}y + \frac{x}{T}e^{-r(T-t)}, \quad v_x = \frac{1}{rT}\left(1-e^{-r(T-t)}\right), \quad v_y = \frac{1}{T}e^{-r(T-t)}, \quad v_{xx} = 0$$

**Substituting into the BSM equation** $v_t + rxv_x + xv_y + \frac{1}{2}\sigma^2x^2v_{xx} = rv$:

$$\frac{r}{T}e^{-r(T-t)}y + \frac{x}{T}e^{-r(T-t)} + \frac{x}{T}\left(1-e^{-r(T-t)}\right) + \frac{x}{T}e^{-r(T-t)}$$

Combining the $\frac{x}{T}e^{-r(T-t)}$ terms ($+1, -1, +1 \to +1$):

$$= \frac{r}{T}e^{-r(T-t)}y + \frac{x}{T}\left(1-e^{-r(T-t)}\right) \cdot \frac{r}{r}$$

$$= r\left[\frac{1}{T}e^{-r(T-t)}y + \frac{x}{rT}\left(1-e^{-r(T-t)}\right)\right] = rv \quad \blacksquare$$

**Boundary condition (7.5.9):** At $x=0$:

$$v(t,0,y) = \frac{e^{-r(T-t)}}{T}y, \quad y \geq 0$$

which agrees with $v(t,0,y) = e^{-r(T-t)}\left(\frac{y}{T}\right)^+$ since $y \geq 0$. $\blacksquare$

### (iii) The hedging process $\Delta(t)$

$$\Delta(t) = v_x(t,S(t),Y(t)) = \frac{1}{rT}\left(1-e^{-r(T-t)}\right)$$

This depends only on $t$ (not on $S(t)$ or $Y(t)$), so $\Delta(t)$ is **deterministic**. $\blacksquare$

### (iv) Replicating the payoff

**Initial capital:**

$$X(0) = v(0,S(0),0) = \frac{S(0)}{rT}\left(1-e^{-rT}\right) = \Delta(0)S(0)$$

**Portfolio dynamics:** With $dX(t) = rX(t)dt + \Delta(t)(dS(t)-rS(t)dt)$, compute $d(e^{r(T-t)}X(t))$:

$$d(e^{r(T-t)}X(t)) = -re^{r(T-t)}X(t)dt + e^{r(T-t)}dX(t) = e^{r(T-t)}\Delta(t)(dS(t)-rS(t)dt)$$

**Relating to $d(\Delta(t)S(t))$:** By Itô's product rule, $d(\Delta(t)S(t)) = \Delta(t)dS(t) + S(t)d\Delta(t)$, so:

$$e^{r(T-t)}\Delta(t)(dS(t)-rS(t)dt) = e^{r(T-t)}d(\Delta(t)S(t)) - e^{r(T-t)}S(t)d\Delta(t)$$

Since $\Delta(t) = \frac{1}{rT}(1-e^{-r(T-t)})$, we have $d\Delta(t) = -\frac{1}{T}e^{-r(T-t)}dt$, so $-e^{r(T-t)}S(t)d\Delta(t) = \frac{1}{T}S(t)dt$:

$$d(e^{r(T-t)}X(t)) = d\left(e^{r(T-t)}\Delta(t)S(t)\right) + \frac{1}{T}S(t)dt$$

**Integrating from $0$ to $T$:**

$$e^{rT}X(0)\cdot e^{-rT}\Big|... \quad \Rightarrow \quad X(T) - e^{rT}X(0) = \Delta(T)S(T) - e^{rT}\Delta(0)S(0) + \frac{1}{T}\int_0^T S(t)dt$$

Wait, more precisely: integrating $d(e^{r(T-t)}X(t))$ from $0$ to $T$:

$$e^{r\cdot 0}X(T) - e^{rT}X(0) = \left[e^{r(T-t)}\Delta(t)S(t)\right]_0^T + \frac{1}{T}\int_0^T S(t)dt$$

$$X(T) - e^{rT}X(0) = \Delta(T)S(T) - e^{rT}\Delta(0)S(0) + \frac{1}{T}\int_0^T S(t)dt$$

**Evaluating boundary terms:** Since $\Delta(T) = \frac{1}{rT}(1-e^0) = 0$, the term $\Delta(T)S(T) = 0$. Since $X(0) = \Delta(0)S(0)$, the terms $e^{rT}X(0)$ and $e^{rT}\Delta(0)S(0)$ cancel:

$$X(T) - e^{rT}X(0) = 0 - e^{rT}\Delta(0)S(0) + \frac{1}{T}\int_0^T S(t)dt = -e^{rT}X(0) + \frac{1}{T}\int_0^T S(t)dt$$

**Conclusion:**

$$X(T) = \frac{1}{T}\int_0^T S(t)dt \quad \blacksquare$$

This confirms that holding $\Delta(t)$ shares (a deterministic, time-varying position) exactly replicates the zero-strike Asian call payoff.
