### (i) Expectations of $Y_1(t)$ and $Y_2(t)$

Since Itô integrals are martingales with zero expectation, taking $\tilde{\mathbb{E}}$ of (10.2.43)-(10.2.46):

**From (10.2.43):**

$$\tilde{\mathbb{E}}Y_1(t) = e^{-\lambda_1 t}Y_1(0) \quad \blacksquare$$

**Case $\lambda_1 \neq \lambda_2$, from (10.2.44):**

$$\tilde{\mathbb{E}}Y_2(t) = \frac{\lambda_{21}}{\lambda_1-\lambda_2}\left(e^{-\lambda_1 t}-e^{-\lambda_2 t}\right)Y_1(0) + e^{-\lambda_2 t}Y_2(0) \quad \blacksquare$$

**Case $\lambda_1 = \lambda_2$, from (10.2.46):**

$$\tilde{\mathbb{E}}Y_2(t) = -\lambda_{21}te^{-\lambda_1 t}Y_1(0) + e^{-\lambda_1 t}Y_2(0) \quad \blacksquare$$

### (ii) Second moments of the Itô integrals

**$\tilde{\mathbb{E}}I_1^2(t)$:** By Itô isometry, $I_1(t) \sim \mathcal{N}\left(0, \int_0^t e^{2\lambda_1 u}du\right)$:

$$\tilde{\mathbb{E}}I_1^2(t) = \int_0^t e^{2\lambda_1 u}du = \frac{1}{2\lambda_1}\left(e^{2\lambda_1 t}-1\right)$$

**$\tilde{\mathbb{E}}[I_1(t)I_2(t)]$:** By Itô's product rule,

$$d(I_1 I_2) = I_1 e^{\lambda_2 t}d\tilde{W}_1 + I_2 e^{\lambda_1 t}d\tilde{W}_1 + e^{(\lambda_1+\lambda_2)t}dt$$

Taking expectations (Itô integrals vanish):

$$\tilde{\mathbb{E}}[I_1(t)I_2(t)] = \int_0^t e^{(\lambda_1+\lambda_2)u}du = \frac{1}{\lambda_1+\lambda_2}\left(e^{(\lambda_1+\lambda_2)t}-1\right)$$

**$\tilde{\mathbb{E}}[I_1(t)I_3(t)]$:** Since $I_1$ is driven by $\tilde{W}_1$ and $I_3$ by $\tilde{W}_2$ (independent), the cross variation vanishes:

$$d(I_1 I_3) = I_1 e^{\lambda_2 t}d\tilde{W}_2 + I_3 e^{\lambda_1 t}d\tilde{W}_1$$

$$\tilde{\mathbb{E}}[I_1(t)I_3(t)] = 0$$

**$\tilde{\mathbb{E}}[I_1(t)I_4(t)]$:** By Itô's product rule, the cross variation is $te^{2\lambda_1 t}dt$:

$$\tilde{\mathbb{E}}[I_1(t)I_4(t)] = \int_0^t ue^{2\lambda_1 u}du = \frac{1}{2\lambda_1}\left[\left(t-\frac{1}{2\lambda_1}\right)e^{2\lambda_1 t}+\frac{1}{2\lambda_1}\right]$$

**$\tilde{\mathbb{E}}I_4^2(t)$:** By Itô isometry:

$$\tilde{\mathbb{E}}I_4^2(t) = \int_0^t u^2 e^{2\lambda_1 u}du = \frac{1}{2\lambda_1}\left[e^{2\lambda_1 t}\left(t^2-\frac{t}{\lambda_1}+\frac{1}{2\lambda_1^2}\right)-\frac{1}{2\lambda_1^2}\right]$$

### (iii) Cross-time covariance $\tilde{\mathbb{E}}[I_1(s)I_2(t)]$ for $0 \leq s < t$

Following the hint, define:

$$J_1(t) = \int_0^t e^{\lambda_1 u}\mathbb{I}_{\{u\leq s\}}d\tilde{W}_1(u)$$

Note that $J_1(t) = I_1(s)$ for $t \geq s$. Since both $J_1$ and $I_2$ are Itô integrals with respect to the same Brownian motion $\tilde{W}_1$, by Itô isometry:

$$\tilde{\mathbb{E}}[I_1(s)I_2(t)] = \tilde{\mathbb{E}}[J_1(t)I_2(t)] = \int_0^t e^{\lambda_1 u}\mathbb{I}_{\{u\leq s\}}\cdot e^{\lambda_2 u}\,du$$

$$= \int_0^s e^{(\lambda_1+\lambda_2)u}du = \frac{1}{\lambda_1+\lambda_2}\left(e^{(\lambda_1+\lambda_2)s}-1\right) \quad \blacksquare$$

**Remark:** Note that this is nonzero because $I_1$ and $I_2$ are driven by the same Brownian motion $\tilde{W}_1$. In contrast, $\tilde{\mathbb{E}}[I_1(t)I_3(t)] = 0$ since $I_3$ is driven by the independent Brownian motion $\tilde{W}_2$.

# Exercise 10.2 (Ordinary differential equations for the mixed affine-yield model).
In the mixed model of Subsection 10.2.3, as in the two-factor Vascik model and the two-factor Cox-Ingersoll-Ross model, zero-coupon bond prices have the affince-yield form
$$f(t, y_1, y_2) = e^{-y_1C_1(T - t) - y_2C_2(T - t) - A(T - t)},$$
where $C_1(0) = C_2(0) = A(0) = 0.$

(i) Find the partial differential equation satisfied by $f(t, y_1, y_2).$

(ii) Show that $C_1, C_2,$ and $A$ satisfy the system of ordinary differential equations
$$C'_1 = -\lambda_1 C_1 - \frac{1}{2}C_1^2 - \sigma_{21} C_1 C_2 - (1 + \beta)C_2^2 + \sigma_1, \tag{10.7.4}$$
$$C''_2 = -\lambda_2 C_2 + \delta_2, \tag{10.7.5}$$
$$A' = \mu C_1 - \frac{1}{2} \alpha C_2^2 + \delta_0. \tag{10.7.6}$$

### Setup

$$dY_1(t) = (\mu - \lambda_1 Y_1(t))dt + \sqrt{Y_1(t)}\,d\tilde{W}_1(t)$$

$$dY_2(t) = -\lambda_2 Y_2(t)dt + \sigma_{21}\sqrt{Y_1(t)}\,d\tilde{W}_1(t) + \sqrt{\alpha+\beta Y_1(t)}\,d\tilde{W}_2(t)$$

$$R(t) = \delta_0 + \delta_1 Y_1(t) + \delta_2 Y_2(t)$$

### (i) The PDE satisfied by $f(t,y_1,y_2)$

Since $D(t)B(t,T) = D(t)f(t,Y_1(t),Y_2(t))$ is a martingale under $\tilde{\mathbb{P}}$, computing its differential:

$$d(D(t)f) = D(t)\left[-Rf\,dt + f_t\,dt + f_{y_1}dY_1 + f_{y_2}dY_2 + \frac{1}{2}f_{y_1y_1}(dY_1)^2 + f_{y_1y_2}dY_1dY_2 + \frac{1}{2}f_{y_2y_2}(dY_2)^2\right]$$

**Quadratic variations:**

$$(dY_1)^2 = y_1\,dt, \quad (dY_2)^2 = \left(\sigma_{21}^2 y_1 + \alpha + \beta y_1\right)dt, \quad dY_1\,dY_2 = \sigma_{21}y_1\,dt$$

Setting the $dt$ term to zero gives the PDE:

$$f_t + (\mu-\lambda_1 y_1)f_{y_1} - \lambda_2 y_2 f_{y_2} + \frac{1}{2}y_1 f_{y_1y_1} + \sigma_{21}y_1 f_{y_1y_2} + \frac{1}{2}\left(\sigma_{21}^2 y_1 + \alpha + \beta y_1\right)f_{y_2y_2}$$

$$= (\delta_0 + \delta_1 y_1 + \delta_2 y_2)f \quad \blacksquare$$

### (ii) Deriving the system of ODEs

For $f(t,y_1,y_2) = e^{-y_1C_1(T-t)-y_2C_2(T-t)-A(T-t)}$, the partial derivatives are:

$$f_t = \left(y_1C_1' + y_2C_2' + A'\right)f, \quad f_{y_i} = -C_i f$$

$$f_{y_1y_1} = C_1^2 f, \quad f_{y_2y_2} = C_2^2 f, \quad f_{y_1y_2} = C_1C_2 f$$

Substituting into the PDE and dividing by $f$:

$$y_1C_1' + y_2C_2' + A' - (\mu-\lambda_1y_1)C_1 + \lambda_2y_2C_2 + \frac{1}{2}y_1C_1^2 + \sigma_{21}y_1C_1C_2 + \frac{1}{2}\left(\sigma_{21}^2y_1+\alpha+\beta y_1\right)C_2^2$$

$$= \delta_0 + \delta_1y_1 + \delta_2y_2$$

Since this must hold for all $y_1, y_2$, we match coefficients:

**Coefficient of $y_1$:**

$$C_1' + \lambda_1C_1 + \frac{1}{2}C_1^2 + \sigma_{21}C_1C_2 + \frac{1}{2}(\sigma_{21}^2+\beta)C_2^2 = \delta_1$$

$$C_1' = -\lambda_1C_1 - \frac{1}{2}C_1^2 - \sigma_{21}C_1C_2 - \frac{1}{2}(\sigma_{21}^2+\beta)C_2^2 + \delta_1 \quad \blacksquare$$

**Coefficient of $y_2$:**

$$C_2' + \lambda_2C_2 = \delta_2 \implies C_2' = -\lambda_2C_2 + \delta_2 \quad \blacksquare$$

**Constant terms:**

$$A' - \mu C_1 + \frac{1}{2}\alpha C_2^2 = \delta_0 \implies A' = \mu C_1 - \frac{1}{2}\alpha C_2^2 + \delta_0 \quad \blacksquare$$

# Exercise 10.3 (Callibration of the two-factor Vasicek model).
Consider the canonical two-factor Vascek model (10.2.4), (10.2.5), but replace the interest rate equation (10.2.6) by
$$R(t) = \delta_0(t) + \delta_1 Y_1(t) + \delta_2 Y_2(t) \tag{10.7.7}$$
where $\delta_1$ and $\delta_2$ are constant but $\delta_0(t)$ is a nonrandom function of time. Assume that for each $T$ there is a zero-coupon bond maturing at time $T$. The price of this bond at time $t \in [0, T]$ is 
$$B(t, T) = \tilde{\mathbb{E}}[e^{-\int_{t}^{T} R(u)du} | \mathcal{F}(t)].$$
Because the pair of process $(Y_1(t), Y_2(t))$ is Markov, there must exist some functyion $f(t, T, y_1, y_2)$ such that $B(t, T) = f(t, T, Y_1(t), Y_2(t)).$ (We indicate the dependence of $f$ on the maturity $T$ because, unlike in Subsection 10.2.1, here we shall consider more than one value of $T$.)

(i) The function $f(t, T, y_1, y_2)$ is of the affine-yield form
$$f(t, T, y_1, y_2) = e^{-y_1C_1(t, T) - y_2 C_2(t, T) - A(t, T)}. \tag{10.7.8}$$
Holding $T$ fixed, derive a system of ordinary differential equations for $\frac{d}{dt}C_1(t, T), \frac{d}{dt}C_2(t, T),$ and $\frac{d}{dt}A(t, T).$

(ii) Using the terminal conditions $C_1(T, T) = C_2(T, T) = 0,$ solve the equations in (i) for $C_1(t, T)$ and $C_2(t, T).$ (As in Subsectin 10.2.1, the functions $C_1$ and $C_2$ depend on $t$ and $T$ only through the difference $\tau = T - t$; however, the function $A$ discussed in part (iii) below depends on $t$ and $T$ separately.)

(iii) Using the terminal condition $A(T,T) = 0$, write a formula for $A(t, T)$ as an integral involving $C_1(u, T), C_2(u, T),$ and $\delta_0(u).$ You do not need to evaluate this integral.

(iv) Assume that the model parameters $\lambda_1 > 0, \lambda_2 >0, \lambda_{21}, \delta_1,$ and $\delta_2$ and the initial conditions $Y_1(0)$ and $Y_2(0)$ are given. We wish to choose a function $\delta_0$ so that the zero-coupon bond prices given by the model match the bond prices given by the market at the initial time zero. In other words, we want to choose a function $\delta(T), T \geq 0,$ so that
$$f(0, T, Y_1(0), Y_2(0)) = B(0, T), \quad T \geq 0.$$
In this part of the exercise, we regard both $t$ and $T$ as variables and use the notation $\frac{\partial}{\partial t}$ to indicate the derivative with respect to $t$ when $T$ is held fixed and the notation $\frac{\partial}{\partial T}$ to indicate the derivative with respext to $T$ when $t$ is held fixed. Give a formula for $\delta_0(T)$ in terms of $\frac{\partial}{\partial T} A(0, T)$ in two ways, using (10.7.8) and also using the formula obtained in (iii). Because $C_i(t, T)$ depends only on $t$ and $T$ through $\tau = T - t,$ there are functions $\bar{C_1}(\tau)$ such that $\bar{C_i}(\tau) = \bar{C_1}(t, T), i = 1, 2.$ Then
$$\frac{\partial}{\partial t} C_i(t, T) = -\bar{C'_i}(\tau), \quad \frac{\partial}{\partial T}C_i(t, T) = \bar{C'_i}(\tau),$$
where $'$ denotes differentiation with respect to $\tau$. This shows that 
$$\frac{\partial}{\partial T}C_i(t, T) = -\frac{\partial}{\partial t} C_i(t, T), i = 1, 2, \tag{10.7.9}$$
a fact that you will need.)

### Setup

$$dY_1(t) = -\lambda_1 Y_1(t)dt + d\tilde{W}_1(t), \quad dY_2(t) = -\lambda_{21}Y_1(t)dt - \lambda_2 Y_2(t)dt + d\tilde{W}_2(t)$$

$$R(t) = \delta_0(t) + \delta_1 Y_1(t) + \delta_2 Y_2(t)$$

### (i) The system of ODEs

Since $D(t)B(t,T) = D(t)f(t,T,Y_1(t),Y_2(t))$ is a $\tilde{\mathbb{P}}$-martingale, setting the $dt$ term of its differential to zero:

$$-Rf + f_t - \lambda_1 y_1 f_{y_1} - (\lambda_{21}y_1+\lambda_2 y_2)f_{y_2} + \frac{1}{2}f_{y_1y_1} + \frac{1}{2}f_{y_2y_2} = 0$$

For $f = e^{-y_1C_1(t,T)-y_2C_2(t,T)-A(t,T)}$ we have $f_t = -\left(y_1\frac{dC_1}{dt}+y_2\frac{dC_2}{dt}+\frac{dA}{dt}\right)f$, $f_{y_i} = -C_i f$, $f_{y_iy_i} = C_i^2 f$. Dividing by $f$ and matching coefficients:

**Coefficient of $y_1$:**
$$\frac{d}{dt}C_1(t,T) = \lambda_1 C_1(t,T) + \lambda_{21}C_2(t,T) - \delta_1$$

**Coefficient of $y_2$:**
$$\frac{d}{dt}C_2(t,T) = \lambda_2 C_2(t,T) - \delta_2$$

**Constant terms:**
$$\frac{d}{dt}A(t,T) = \frac{1}{2}C_1^2(t,T) + \frac{1}{2}C_2^2(t,T) - \delta_0(t) \quad \blacksquare$$

### (ii) Solving for $C_1$ and $C_2$

Write $\tau = T-t$.

**$C_2$:** Multiplying by the integrating factor $e^{-\lambda_2 t}$:

$$\frac{d}{dt}\left(C_2(t,T)e^{-\lambda_2 t}\right) = -\delta_2 e^{-\lambda_2 t}$$

Integrating from $t$ to $T$ and using $C_2(T,T)=0$:

$$C_2(t,T) = \frac{\delta_2}{\lambda_2}\left(1-e^{-\lambda_2\tau}\right)$$

**$C_1$ (assuming $\lambda_1 \neq \lambda_2$):** Substituting $C_2$ and using the integrating factor $e^{-\lambda_1 t}$:

$$\frac{d}{dt}\left(C_1 e^{-\lambda_1 t}\right) = e^{-\lambda_1 t}\left[\frac{\lambda_{21}\delta_2}{\lambda_2}-\delta_1\right] - \frac{\lambda_{21}\delta_2}{\lambda_2}e^{-\lambda_2 T}e^{(\lambda_2-\lambda_1)t}$$

Integrating from $t$ to $T$ with $C_1(T,T)=0$:

$$C_1(t,T) = \left[\frac{\lambda_{21}\delta_2}{\lambda_1(\lambda_2-\lambda_1)}-\frac{\delta_1}{\lambda_1}\right]e^{-\lambda_1\tau} - \frac{\lambda_{21}\delta_2}{\lambda_2(\lambda_2-\lambda_1)}e^{-\lambda_2\tau} - \frac{1}{\lambda_1}\left(\frac{\lambda_{21}\delta_2}{\lambda_2}-\delta_1\right) \quad \blacksquare$$

**Verification:** At $\tau=0$ the three coefficients sum to zero, confirming $C_1(T,T)=0$.

### (iii) Formula for $A(t,T)$

Integrating the ODE for $A$ from $t$ to $T$ and using $A(T,T)=0$:

$$A(t,T) = \int_t^T \delta_0(u)du - \frac{1}{2}\int_t^T C_1^2(u,T)du - \frac{1}{2}\int_t^T C_2^2(u,T)du \quad \blacksquare$$

### (iv) Calibrating $\delta_0$

Write $y_1 = Y_1(0)$, $y_2 = Y_2(0)$.

**First expression (from (iii)):** Setting $t=0$ and differentiating with respect to $T$. Note that $C_i(u,T)$ depends on $T$, so those terms contribute; the boundary terms vanish since $C_i(T,T)=0$:

$$\frac{\partial}{\partial T}A(0,T) = \delta_0(T) - \int_0^T C_1\frac{\partial C_1}{\partial T}du - \int_0^T C_2\frac{\partial C_2}{\partial T}du$$

Using (10.7.9), $\frac{\partial C_i(u,T)}{\partial T} = -\frac{\partial C_i(u,T)}{\partial u}$:

$$\int_0^T C_i\frac{\partial C_i}{\partial T}du = -\frac{1}{2}\Big[C_i^2(u,T)\Big]_{u=0}^{u=T} = \frac{1}{2}C_i^2(0,T)$$

Therefore:

$$\boxed{\delta_0(T) = \frac{\partial}{\partial T}A(0,T) + \frac{1}{2}C_1^2(0,T) + \frac{1}{2}C_2^2(0,T)}$$

**Second expression (from (10.7.8)):** Taking logarithms of $f(0,T,y_1,y_2) = B(0,T)$:

$$\log B(0,T) = -y_1C_1(0,T) - y_2C_2(0,T) - A(0,T)$$

Differentiating with respect to $T$ and applying (10.7.9):

$$\frac{\partial}{\partial T}\log B(0,T) = y_1\frac{\partial C_1(0,T)}{\partial t} + y_2\frac{\partial C_2(0,T)}{\partial t} - \frac{\partial}{\partial T}A(0,T)$$

Substituting the ODEs from (i):

$$= y_1\left[\lambda_1 C_1 + \lambda_{21}C_2 - \delta_1\right] + y_2\left[\lambda_2 C_2 - \delta_2\right] - \frac{\partial}{\partial T}A(0,T)$$

Hence:

$$\boxed{\frac{\partial}{\partial T}A(0,T) = \lambda_1 y_1 C_1(0,T) + (\lambda_{21}y_1+\lambda_2 y_2)C_2(0,T) - \delta_1 y_1 - \delta_2 y_2 - \frac{\partial}{\partial T}\log B(0,T)}$$

**Combining the two:**

$$\delta_0(T) = \lambda_1 y_1 C_1(0,T) + (\lambda_{21}y_1+\lambda_2 y_2)C_2(0,T) - \delta_1 y_1 - \delta_2 y_2 - \frac{\partial}{\partial T}\log B(0,T) + \frac{1}{2}C_1^2(0,T) + \frac{1}{2}C_2^2(0,T) \quad \blacksquare$$

Since $C_1(0,T)$ and $C_2(0,T)$ are known explicitly from (ii) and $\frac{\partial}{\partial T}\log B(0,T)$ is obtained from the observed market bond prices (it equals $-f(0,T)$, the initial instantaneous forward rate curve), this determines $\delta_0(T)$ for every $T \geq 0$.

**Consistency check:** As $T\to 0$, $C_i(0,0)=0$ and $-\frac{\partial}{\partial T}\log B(0,T)\big|_{T=0} = R(0)$, giving $\delta_0(0) = R(0)-\delta_1y_1-\delta_2y_2$, as required by (10.7.7).

$$dU(t) = -\lambda_1 U(t) dt + \sigma_1 d\tilde{B}_2(t), \tag{10.7.10}$$
$$dR(t) = [\theta(t) + U(t) - \lambda_2 R(t)]dt + \sigma_2 d\tilde{B}_1(t), \tag{10.7.11}$$
where $\lambda_1, \lambda_2, \sigma_1,$ and $\sigma_2$ are positive constants, $\theta(t)$ is a nonrandom function, and $\tilde{B}_1(t)$ and $\tilde{B}_2(t)$ are correlated Brownian motions with $d\tilde{B}_1(t)d\tilde{B}_2't( = \rho dt$ for some $\rho \in (-1, 1).$ In this exercise, we discuss how to reduce this to the two-fator Vasicek model of Subsection 10.2.1, except that, instead of (10.2.6), the interest rate is given by (10.7.7), in which $\delta_0(t)$ is a nonrandom function of time

(i) Define
$$X(t) = \begin{bmatrix} U(t) \\ R(t) \end{bmatrix}, \quad K = \begin{bmatrix} \lambda_1 & 0 \\ -1 & \lambda_2 \end{bmatrix}, \quad \Sigma = \begin{bmatrix} \sigma_1 & 0 \\ 0 & \sigma_2 \end{bmatrix}$$
$$\Theta(t) = \begin{bmatrix} 0 \\ \theta(t) \end{bmatrix}, \tilde{B}(t) = \begin{bmatrix} \tilde{B}_1(t) \\ \tilde{B}_2(t) \end{bmatrix}$$
so that (10.7.10) and (10.7.11) can be written in vector notation as
$$dX(t) = \Theta(t)dt - K X(t)dt + \Sigma d\tilde{B}(t). \tag{10.7.12}$$
Now set
$$\hat{X}(t) = X(t) - e^{-Kt} \int_{0}^{t} e^{Ku} \Theta(u) du.$$
Show that
$$d\hat{X}(t) = -K \hat{X}(t) dt + \Sigma d\tilde{B}(t). \tag{10.7.13}$$
(ii) With
$$C = \begin{bmatrix} \frac{1}{\sigma1} & 0 \\ -\frac{\rho}{\sigma_1 \sqrt{1 - \rho^2}} & \frac{1}{\sigma_2 \sqrt{1 - \rho^2}} \end{bmatrix},$$
define $Y(t) = C\hat{X}(t), \tilde{W}(t) = C \Sigma \tilde{B}(t).$ Show that the components of $\tilde{W}_1(t)$ and $\tilde{W}_2(t)$ are independend Brownian motions and 
$$dY(t) = -\Lambda Y(t) + d\tilde{W}(t), \tag{10.7.14}$$
where
$$\Lambda = CKC^{-1} = \begin{bmatrix} \lambda_1 & 0 \\ \frac{\rho \sigma_2(\lambda_2 - \lambda_1) - \sigma_1}{\sigma_2 \sqrt{1 - \rho^2}} & \lambda_2 \end{bmatrix}.$$
Equation (10.7.14) is the vector form of the canonical two-factor Vasicek equation (10.2.4) and (10.2.5).

(iii) Obtain a formula for $R(t)$ of the form (10.7.7). What are $\delta_0(t), \delta_1,$ and $\delta_2$?

### (i) The shifted process $\hat{X}(t)$

Differentiating $\hat{X}(t) = X(t) - e^{-Kt}\int_0^t e^{Ku}\Theta(u)du$ using the product rule (the integral term is of bounded variation, so no Itô correction):

$$d\hat{X}(t) = dX(t) + Ke^{-Kt}\left(\int_0^t e^{Ku}\Theta(u)du\right)dt - e^{-Kt}e^{Kt}\Theta(t)dt$$

Substituting (10.7.12):

$$= \Theta(t)dt - KX(t)dt + \Sigma d\tilde{B}(t) + Ke^{-Kt}\left(\int_0^t e^{Ku}\Theta(u)du\right)dt - \Theta(t)dt$$

The $\Theta(t)dt$ terms cancel, and the remaining drift terms combine:

$$= -K\left[X(t) - e^{-Kt}\int_0^t e^{Ku}\Theta(u)du\right]dt + \Sigma d\tilde{B}(t) = -K\hat{X}(t)dt + \Sigma d\tilde{B}(t) \quad \blacksquare$$

### (ii) Decorrelation and the canonical form

**Computing $\tilde{W}(t) = C\Sigma\tilde{B}(t)$:**

$$C\Sigma = \begin{bmatrix} \frac{1}{\sigma_1} & 0 \\ -\frac{\rho}{\sigma_1\sqrt{1-\rho^2}} & \frac{1}{\sigma_2\sqrt{1-\rho^2}}\end{bmatrix}\begin{bmatrix}\sigma_1 & 0 \\ 0 & \sigma_2\end{bmatrix} = \begin{bmatrix} 1 & 0 \\ -\frac{\rho}{\sqrt{1-\rho^2}} & \frac{1}{\sqrt{1-\rho^2}}\end{bmatrix}$$

so that:

$$\tilde{W}_1(t) = \tilde{B}_1(t), \quad \tilde{W}_2(t) = \frac{-\rho\tilde{B}_1(t)+\tilde{B}_2(t)}{\sqrt{1-\rho^2}}$$

**Verification via Lévy's theorem:** $\tilde{W}_1(0)=\tilde{W}_2(0)=0$, both are continuous martingales, and using $d\tilde{B}_1 d\tilde{B}_2 = \rho\,dt$:

$$d\tilde{W}_1 d\tilde{W}_1 = dt$$

$$d\tilde{W}_2 d\tilde{W}_2 = \frac{\rho^2 - 2\rho\cdot\rho + 1}{1-\rho^2}dt = dt$$

$$d\tilde{W}_1 d\tilde{W}_2 = \frac{-\rho + \rho}{\sqrt{1-\rho^2}}dt = 0$$

By the two-dimensional Lévy theorem, $\tilde{W}_1$ and $\tilde{W}_2$ are independent Brownian motions. $\blacksquare$

**Deriving (10.7.14):** Applying $C$ to (10.7.13):

$$dY(t) = C\,d\hat{X}(t) = -CK\hat{X}(t)dt + C\Sigma\,d\tilde{B}(t) = -CKC^{-1}\underbrace{C\hat{X}(t)}_{Y(t)}dt + d\tilde{W}(t)$$

$$= -\Lambda Y(t)dt + d\tilde{W}(t) \quad \blacksquare$$

### (iii) Recovering $R(t)$ in the form (10.7.7)

From the definition of $\hat{X}$:

$$X(t) = C^{-1}Y(t) + e^{-Kt}\int_0^t e^{Ku}\Theta(u)du$$

**The matrix $C^{-1}$:**

$$C^{-1} = \begin{bmatrix} \sigma_1 & 0 \\ \rho\sigma_2 & \sigma_2\sqrt{1-\rho^2}\end{bmatrix}$$

**The matrix exponentials:** Since $K$ is lower-triangular with diagonal $(\lambda_1,\lambda_2)$, for $\lambda_1 \neq \lambda_2$:

$$e^{Ku} = \begin{bmatrix} e^{\lambda_1 u} & 0 \\ \frac{e^{\lambda_2 u}-e^{\lambda_1 u}}{\lambda_2-\lambda_1} & e^{\lambda_2 u}\end{bmatrix}, \quad e^{-Kt} = \begin{bmatrix} e^{-\lambda_1 t} & 0 \\ \frac{e^{-\lambda_1 t}-e^{-\lambda_2 t}}{\lambda_2-\lambda_1} & e^{-\lambda_2 t}\end{bmatrix}$$

Since $\Theta(u) = (0,\ \theta(u))^\top$, only the second column of $e^{Ku}$ contributes:

$$e^{Ku}\Theta(u) = \begin{bmatrix} 0 \\ e^{\lambda_2 u}\theta(u)\end{bmatrix}, \quad e^{-Kt}\int_0^t e^{Ku}\Theta(u)du = \begin{bmatrix} 0 \\ e^{-\lambda_2 t}\int_0^t e^{\lambda_2 u}\theta(u)du\end{bmatrix}$$

**Taking the second component of $X(t)$:**

$$R(t) = \underbrace{e^{-\lambda_2 t}\int_0^t e^{\lambda_2 u}\theta(u)du}_{\delta_0(t)} + \underbrace{\rho\sigma_2}_{\delta_1}Y_1(t) + \underbrace{\sigma_2\sqrt{1-\rho^2}}_{\delta_2}Y_2(t)$$

Therefore:

$$\delta_0(t) = e^{-\lambda_2 t}\int_0^t e^{\lambda_2 u}\theta(u)du, \quad \delta_1 = \rho\sigma_2, \quad \delta_2 = \sigma_2\sqrt{1-\rho^2} \quad \blacksquare$$

**Sanity check:** At $t=0$, $\delta_0(0)=0$ and $Y(0)=C\hat{X}(0)=CX(0)$, so $R(0) = \rho\sigma_2 Y_1(0)+\sigma_2\sqrt{1-\rho^2}Y_2(0)$, which is the second component of $C^{-1}CX(0) = X(0)$, as required.

# Exercise 10.5(Correlation between long rate and short rate in the one-factor Vasicek model).
The one-faxtor Vasicek model is the one-factor Hull-White model of Example 6.5.1 with constant parameters,
$$dR(t) = (a - bR(t))dt + \sigma d\tilde{W}(t), \tag{10.7.15}$$
where $a, b,$ and $\sigma$ are positive constants and $\tilde{W}(t)$ is a one-dimensional Brownian motion. In this model, the price at time $t \in [0, T]$ of the zero-coupon bond maturing at time T is 
$$B(t, T) = e^{-C(t,T)R(t) - A(t, T)},$$
where $C(t, T)$ and $A(t,T)$ are given by (6.5.10) and (6.5.11):
$$C(t, T) = \int_{t}^{T} e^{-\int_{t}^{s}bdv}ds = \frac{1}{b} (1 - e^{-b(T - t)}),$$
$$A(t, T) = \int_{t}^{T} \left(aC(s, T) - \frac{1}{2} \sigma_2 C^2(s, T)\right) ds$$
$$ = \frac{2ab - \sigma^2}{2b^2} (T - t) + \frac{\sigma^2 - ab}{b^3} (1 - e^{-b(T - t)}) - \frac{\sigma^2}{4b^3}(1 - e^{-2b(T - t)}).$$
In the spirit of the discussion of the short rate and long rate in Subsection 10.2.1, we fix a positive relative maturity $\bar{\tau}$ and define the long rate $L(t)$ at time $t$ by (10.2.30):
$$L(t) = -\frac{1}{\bar{\tau}} \log B(t, t+ \bar{\tau}).$$
Show that changes in $L(t)$ and $R(t)$ are perfectly correlated (i.e., for any $0 \leq t_1 \leq t_2,$ the correlation coefficient between $L(t_2) - L(t_1)$ and $R(t_2) - R(t_1)$ is one). This characteristic of one-factor models caused the development of models with more than one factor.

### Step 1: $L(t)$ is an affine function of $R(t)$

Since $b$ is constant, $C(t,T)$ and $A(t,T)$ depend on $t$ and $T$ only through $\tau = T-t$. Writing $C(\bar\tau) = C(t,t+\bar\tau)$ and $A(\bar\tau) = A(t,t+\bar\tau)$:

$$B(t,t+\bar\tau) = e^{-C(\bar\tau)R(t)-A(\bar\tau)}$$

Taking logarithms in the definition (10.2.30):

$$L(t) = -\frac{1}{\bar\tau}\log B(t,t+\bar\tau) = \frac{C(\bar\tau)}{\bar\tau}R(t) + \frac{A(\bar\tau)}{\bar\tau}$$

Note that $\frac{C(\bar\tau)}{\bar\tau}$ and $\frac{A(\bar\tau)}{\bar\tau}$ are **constants** (independent of $t$), and:

$$\frac{C(\bar\tau)}{\bar\tau} = \frac{1-e^{-b\bar\tau}}{b\bar\tau} > 0$$

since $b > 0$ and $\bar\tau > 0$.

### Step 2: The increments are proportional

Setting $\kappa = \frac{C(\bar\tau)}{\bar\tau} > 0$, the constant term $\frac{A(\bar\tau)}{\bar\tau}$ cancels in the difference:

$$L(t_2) - L(t_1) = \kappa\left(R(t_2)-R(t_1)\right)$$

### Step 3: Computing the correlation

Write $\Delta R = R(t_2)-R(t_1)$ and $\Delta L = \kappa\,\Delta R$. Then:

$$\text{Var}(\Delta L) = \kappa^2\text{Var}(\Delta R)$$

$$\text{Cov}(\Delta L, \Delta R) = \mathbb{E}\left[\left(\kappa\Delta R - \kappa\mathbb{E}\Delta R\right)\left(\Delta R - \mathbb{E}\Delta R\right)\right] = \kappa\,\text{Var}(\Delta R)$$

Therefore:

$$\rho(\Delta L,\Delta R) = \frac{\text{Cov}(\Delta L,\Delta R)}{\sqrt{\text{Var}(\Delta L)}\sqrt{\text{Var}(\Delta R)}} = \frac{\kappa\,\text{Var}(\Delta R)}{|\kappa|\,\text{Var}(\Delta R)} = \frac{\kappa}{|\kappa|} = 1$$

using $\kappa > 0$ and $\text{Var}(\Delta R) > 0$ (which holds since $R$ is driven by a nondegenerate Brownian motion). $\blacksquare$

### Remark: Why this motivates multifactor models

The correlation is one because $L(t)$ is a **deterministic affine function** of $R(t)$ — there is only one source of randomness ($\tilde{W}$) in the model, so every point on the yield curve is a deterministic transformation of the single state variable $R(t)$.

Consequences that contradict market observation:

- The yield curve can only shift, never change shape independently: a $1$bp move in $R$ forces exactly a $\kappa$ bp move in $L$
- Steepening, flattening, and butterfly moves cannot be represented
- Hedging is degenerate: any single bond hedges any other perfectly

Since $\kappa = \frac{1-e^{-b\bar\tau}}{b\bar\tau} \in (0,1)$ and decreases in $\bar\tau$, longer rates move less than the short rate — but always in lockstep. Real yield curves show correlations well below one between short and long rates, which is precisely what the two-factor models of Section 10.2 are designed to capture.

### (i) Reversing the order of integration

Starting from (10.7.20):

$$-\int_t^T f(t,v)dv = -\int_t^T f(0,v)dv - \int_t^T\left[\int_0^t\alpha(u,v)du\right]dv - \int_t^T\left[\int_0^t\sigma(u,v)dW(u)\right]dv$$

**Second term (Fubini for Riemann integrals):**

$$\int_t^T\int_0^t\alpha(u,v)\,du\,dv = \int_0^t\left[\int_t^T\alpha(u,v)dv\right]du = \int_0^t\hat\alpha(u,t,T)\,du$$

**Third term (stochastic Fubini):** Reversing a Riemann and an Itô integral is justified in the appendix of [83]:

$$\int_t^T\int_0^t\sigma(u,v)\,dW(u)\,dv = \int_0^t\left[\int_t^T\sigma(u,v)dv\right]dW(u) = \int_0^t\hat\sigma(u,t,T)\,dW(u)$$

Substituting both:

$$-\int_t^T f(t,v)dv = -\int_t^T f(0,v)dv - \int_0^t\hat\alpha(u,t,T)du - \int_0^t\hat\sigma(u,t,T)dW(u) \quad \blacksquare$$

### (ii) Differentiating with respect to $t$

Each term contains $t$ in more than one place, so each contributes two pieces.

**First term:** $t$ appears only as the lower limit:

$$d\left(-\int_t^T f(0,v)dv\right) = f(0,t)\,dt$$

**Second term:** $t$ appears as the upper limit of $\int_0^t\cdots du$ **and** inside $\hat\alpha(u,t,T)$:

$$d\left(-\int_0^t\hat\alpha(u,t,T)du\right) = -\hat\alpha(t,t,T)\,dt - \left[\int_0^t\frac{\partial}{\partial t}\hat\alpha(u,t,T)du\right]dt$$

Since $\hat\alpha(u,t,T) = \int_t^T\alpha(u,v)dv$, differentiating the lower limit gives $\frac{\partial}{\partial t}\hat\alpha(u,t,T) = -\alpha(u,t)$, so:

$$= -\hat\alpha(t,t,T)\,dt + \left[\int_0^t\alpha(u,t)du\right]dt$$

**Third term:** similarly, $t$ appears as the upper limit of the Itô integral **and** inside $\hat\sigma(u,t,T)$:

$$d\left(-\int_0^t\hat\sigma(u,t,T)dW(u)\right) = -\hat\sigma(t,t,T)\,dW(t) + \left[\int_0^t\sigma(u,t)dW(u)\right]dt$$

**Combining all three:**

$$d\left(-\int_t^T f(t,v)dv\right) = \left[f(0,t) + \int_0^t\alpha(u,t)du + \int_0^t\sigma(u,t)dW(u)\right]dt - \hat\alpha(t,t,T)\,dt - \hat\sigma(t,t,T)\,dW(t)$$

### (iii) Agreement with (10.3.10)

The bracketed $dt$ coefficient is exactly (10.3.5) evaluated at $v = t$:

$$f(0,t) + \int_0^t\alpha(u,t)du + \int_0^t\sigma(u,t)dW(u) = f(t,t) = R(t)$$

using the definition of the short rate $R(t) = f(t,t)$. Writing:

$$\alpha^*(t,T) = \hat\alpha(t,t,T) = \int_t^T\alpha(t,v)dv, \quad \sigma^*(t,T) = \hat\sigma(t,t,T) = \int_t^T\sigma(t,v)dv$$

we obtain:

$$d\left(-\int_t^T f(t,v)dv\right) = R(t)\,dt - \alpha^*(t,T)\,dt - \sigma^*(t,T)\,dW(t)$$

which is (10.3.10). $\blacksquare$

# Exercise 10.9 (Multifactor HJM model).
Suppose the Heath-Jarrow-Morton model is driven by a d-dimensional Brownian motion, so that $\sigma(t, T)$ is also a d-dimensional vector and the forward rate dynamics are given by
$$df(t, T) =\alpha(t, T)dt + \sum_{j=1}^{d}\sigma_j(t, T)dW_j(t).$$
(i) Show that (10.3.16) becomes
$$\alpha(t, T) = \sum_{j=1}^{d}\sigma_j(t, T)[\sigma_j^*(t, T) + \Theta_j(t)].$$
(ii) Suppose there is an adapted, d-dimensional process
$$\Theta(t) = (\Theta_1(t), \ldots, \Theta_d(t))$$
satisfying this equation for all $0 \leq t \leq T \leq \bar{T}.$ Show that if there are maturities $T_1, \ldots, T_d$ such that the $d \times d$ matrix $(\sigma_j(t, T_i))_{i,j}$ is nonsingular, then $\Theta(t)$ is unique.

### (i) The multifactor no-arbitrage condition

**Bond price dynamics.** From Exercise 10.8 extended to $d$ factors:

$$d\left(-\int_t^T f(t,v)dv\right) = R(t)dt - \alpha^*(t,T)dt - \sum_{j=1}^d \sigma_j^*(t,T)dW_j(t)$$

where $\alpha^*(t,T) = \int_t^T\alpha(t,v)dv$ and $\sigma_j^*(t,T) = \int_t^T\sigma_j(t,v)dv$.

Applying Itô's lemma to $B(t,T) = \exp\left\{-\int_t^T f(t,v)dv\right\}$, and using the independence of $W_1,\dots,W_d$ so that the quadratic variation is $\sum_j(\sigma_j^*)^2dt$:

$$dB(t,T) = B(t,T)\left[\left(R(t) - \alpha^*(t,T) + \frac{1}{2}\sum_{j=1}^d\left(\sigma_j^*(t,T)\right)^2\right)dt - \sum_{j=1}^d\sigma_j^*(t,T)dW_j(t)\right]$$

**Discounting.** With $D(t) = e^{-\int_0^t R(u)du}$:

$$d\left(D(t)B(t,T)\right) = D(t)B(t,T)\left[\left(-\alpha^*(t,T)+\frac{1}{2}\sum_{j=1}^d\left(\sigma_j^*\right)^2\right)dt - \sum_{j=1}^d\sigma_j^*\,dW_j(t)\right]$$

**No-arbitrage requirement.** There must exist an adapted $\Theta(t) = (\Theta_1(t),\dots,\Theta_d(t))$ such that, with $d\tilde{W}_j(t) = dW_j(t) + \Theta_j(t)dt$ defining a risk-neutral measure via Girsanov:

$$d\left(D(t)B(t,T)\right) = -D(t)B(t,T)\sum_{j=1}^d\sigma_j^*(t,T)\,d\tilde{W}_j(t)$$

Matching the $dt$ terms:

$$-\alpha^*(t,T) + \frac{1}{2}\sum_{j=1}^d\left(\sigma_j^*(t,T)\right)^2 = -\sum_{j=1}^d\sigma_j^*(t,T)\Theta_j(t) \tag{$\ast$}$$

**Differentiating with respect to $T$.** Using $\frac{\partial}{\partial T}\alpha^*(t,T) = \alpha(t,T)$ and $\frac{\partial}{\partial T}\sigma_j^*(t,T) = \sigma_j(t,T)$:

$$-\alpha(t,T) + \sum_{j=1}^d\sigma_j^*(t,T)\sigma_j(t,T) = -\sum_{j=1}^d\sigma_j(t,T)\Theta_j(t)$$

$$\alpha(t,T) = \sum_{j=1}^d\sigma_j(t,T)\left[\sigma_j^*(t,T) + \Theta_j(t)\right] \quad \blacksquare$$

### (ii) Uniqueness of $\Theta(t)$

Evaluate the condition from (i) at the $d$ maturities $T_1,\dots,T_d$ and move the known terms to the left:

$$\underbrace{\alpha(t,T_i) - \sum_{j=1}^d\sigma_j(t,T_i)\sigma_j^*(t,T_i)}_{=:b_i(t)} = \sum_{j=1}^d\sigma_j(t,T_i)\Theta_j(t), \quad i=1,\dots,d$$

In matrix form, with $M(t) = \left(\sigma_j(t,T_i)\right)_{i,j}$:

$$M(t)\,\Theta(t) = b(t)$$

The left-hand data $b(t)$ is determined entirely by the model coefficients $\alpha$ and $\sigma_j$. Since $M(t)$ is nonsingular by hypothesis:

$$\Theta(t) = M(t)^{-1}b(t)$$

is uniquely determined. Any $\Theta(t)$ satisfying the condition for all $0\leq t\leq T\leq\bar{T}$ must in particular satisfy it at $T_1,\dots,T_d$, hence must equal this vector. $\blacksquare$

# Exercise 10.11.
Let $\delta > 0$ be given. Consider an interest rate swap paying a fixed interest rate $K$ and receiving backset LIBOR $L(T_{j-1}, T_{j-1})$ on a principal of $1$ at each of the payment dates $T_j = \delta j, j = 1, 2, \ldots, n+1.$ Show thta the value of the swap is 
$$\delta K \sum_{j=1}^{n+1} B(0, T_j) - \delta \sum_{j=1}^{n+1}B(0, T_j)L(0, T_{j-1}). \tag{10.7.22}$$
Remark 10.7.1. The swap rate is defined to be the value of $K$ that makes the initial value of the swap equal to zero. Thus, the swap rate is 
$$K = \frac{\sum_{j=1}^{n+1}B(0, T_j)L(0, T_{j-1})}{\sum_{j=1}^{n+1}B(0, T_j)}. \tag{10.7.23}$$

### Valuing the fixed leg

The payment $\delta K$ at $T_j$ is deterministic, so its time-zero value is $\delta K B(0,T_j)$. Summing:

$$\text{Fixed leg} = \delta K\sum_{j=1}^{n+1}B(0,T_j)$$

### Valuing the floating leg

Each floating payment is a **backset LIBOR** contract: the rate $L(T_{j-1},T_{j-1})$ is set at $T_{j-1}$ and paid at $T_j = T_{j-1}+\delta$. By the backset LIBOR pricing formula, its time-zero value is:

$$\delta B(0,T_j)L(0,T_{j-1})$$

**Verification of the backset formula.** By risk-neutral pricing, using $L(T_{j-1},T_{j-1}) = \frac{1}{\delta}\left(\frac{1}{B(T_{j-1},T_j)}-1\right)$:

$$\tilde{\mathbb{E}}\left[D(T_j)\,\delta L(T_{j-1},T_{j-1})\right] = \tilde{\mathbb{E}}\left[D(T_j)\left(\frac{1}{B(T_{j-1},T_j)}-1\right)\right]$$

For the first term, condition on $\mathcal{F}(T_{j-1})$ and use $\tilde{\mathbb{E}}\left[\frac{D(T_j)}{D(T_{j-1})}\middle|\mathcal{F}(T_{j-1})\right] = B(T_{j-1},T_j)$:

$$\tilde{\mathbb{E}}\left[\frac{D(T_j)}{B(T_{j-1},T_j)}\right] = \tilde{\mathbb{E}}\left[D(T_{j-1})\right] = B(0,T_{j-1})$$

The second term is $-B(0,T_j)$. Hence the value is $B(0,T_{j-1})-B(0,T_j)$, and by the definition of forward LIBOR:

$$L(0,T_{j-1}) = \frac{B(0,T_{j-1})-B(0,T_j)}{\delta B(0,T_j)} \implies B(0,T_{j-1})-B(0,T_j) = \delta B(0,T_j)L(0,T_{j-1}) \quad\checkmark$$

Summing over $j$:

$$\text{Floating leg} = \delta\sum_{j=1}^{n+1}B(0,T_j)L(0,T_{j-1}) = B(0,T_0) - B(0,T_{n+1})$$

(the telescoping form — the floating leg is worth the difference of two zero-coupon bonds).

### Value of the swap

$$\text{Value} = \text{Fixed} - \text{Floating} = \delta K\sum_{j=1}^{n+1}B(0,T_j) - \delta\sum_{j=1}^{n+1}B(0,T_j)L(0,T_{j-1}) \quad \blacksquare$$

### Remark 10.7.1: the swap rate

Factoring out $\delta\sum_j B(0,T_j) > 0$:

$$\text{Value} = \delta\left(\sum_{j=1}^{n+1}B(0,T_j)\right)\left[K - \frac{\sum_{j=1}^{n+1}B(0,T_j)L(0,T_{j-1})}{\sum_{j=1}^{n+1}B(0,T_j)}\right]$$

The bracket vanishes exactly when $K$ equals (10.7.23), so the swap rate is the **weighted average of the forward LIBOR rates**, with weights $\frac{B(0,T_j)}{\sum_k B(0,T_k)}$ given by the discount factors.

Equivalently, using the telescoping form of the floating leg:

$$K = \frac{B(0,T_0)-B(0,T_{n+1})}{\delta\sum_{j=1}^{n+1}B(0,T_j)}$$

The denominator $\delta\sum_j B(0,T_j)$ is the **annuity** (or PV01) and serves as the numéraire for the swap measure used in pricing swaptions.

# Exercise 10.12.
In the proof of Theorem 10.4.1, we showed by an arbitrage argument that the value at time $0$ of a payment of backset LIBOR $L(T, T)$ at time $T + \delta$ is $B(0, T+ \delta)L(0, T).$ The risk-neutral price of this payment, computed at time zero, is
$$\tilde{\mathbb{E}}[D(T + \delta)L(T, T)].$$
Use the definitions
$$L(T, T) = \frac{1 - B(T, T+\delta)}{\delta B(T, T+\delta)},$$
$$B(0, T+\delta) = \tilde{\mathbb{E}}[D(T+ \delta)],$$ 
and the properties of conditional expectations to show that
$$\tilde{\mathbb{E}}[D(T + \delta)L(T, T)] = B(0, T+\delta)L(0, T).$$

**Proof:**

By the tower property, conditioning on $\mathcal{F}(T)$:

$$\tilde{\mathbb{E}}\left[D(T+\delta)L(T,T)\right] = \tilde{\mathbb{E}}\left[\tilde{\mathbb{E}}\left[D(T+\delta)L(T,T)\,\middle|\,\mathcal{F}(T)\right]\right]$$

Since $L(T,T)$ is $\mathcal{F}(T)$-measurable, it factors out of the inner expectation:

$$= \tilde{\mathbb{E}}\left[L(T,T)\,\tilde{\mathbb{E}}\left[D(T+\delta)\,\middle|\,\mathcal{F}(T)\right]\right]$$

By the risk-neutral pricing formula for the zero-coupon bond, $\tilde{\mathbb{E}}\left[D(T+\delta)\middle|\mathcal{F}(T)\right] = D(T)B(T,T+\delta)$, and $D(T)=1$ is not assumed — but writing it out:

$$= \tilde{\mathbb{E}}\left[L(T,T)\,D(T)B(T,T+\delta)\right]$$

Substituting $L(T,T) = \dfrac{1-B(T,T+\delta)}{\delta B(T,T+\delta)}$, the factor $B(T,T+\delta)$ cancels:

$$= \frac{1}{\delta}\,\tilde{\mathbb{E}}\left[D(T)\left(1-B(T,T+\delta)\right)\right] = \frac{1}{\delta}\left(\tilde{\mathbb{E}}\left[D(T)\right] - \tilde{\mathbb{E}}\left[D(T)B(T,T+\delta)\right]\right)$$

For the first term, $\tilde{\mathbb{E}}[D(T)] = B(0,T)$. For the second, apply the tower property again:

$$\tilde{\mathbb{E}}\left[D(T)B(T,T+\delta)\right] = \tilde{\mathbb{E}}\left[\tilde{\mathbb{E}}\left[D(T+\delta)\,\middle|\,\mathcal{F}(T)\right]\right] = \tilde{\mathbb{E}}\left[D(T+\delta)\right] = B(0,T+\delta)$$

Therefore:

$$\tilde{\mathbb{E}}\left[D(T+\delta)L(T,T)\right] = \frac{B(0,T)-B(0,T+\delta)}{\delta}$$

Finally, by the definition of forward LIBOR:

$$L(0,T) = \frac{B(0,T)-B(0,T+\delta)}{\delta B(0,T+\delta)} \implies \frac{B(0,T)-B(0,T+\delta)}{\delta} = B(0,T+\delta)L(0,T)$$

Hence:

$$\tilde{\mathbb{E}}\left[D(T+\delta)L(T,T)\right] = B(0,T+\delta)L(0,T) \quad \blacksquare$$
