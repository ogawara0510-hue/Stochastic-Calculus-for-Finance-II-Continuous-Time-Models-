# Exercise 9.1.
This exercise provides an alternate proof of the main assertion of Theorem 9.2.2.

(i) Use Lemma 5.2.2 to prove Remark 9.2.5.

(ii) Let $S(t)$ and $N(t)$ be prices of two assets, denominated in a common currency, and assume $N(t)$ is always strictly positive. Let $\tilde{\mathbb{P}}$ be the risk-neutral measure under which the discounted asset prices $D(t)S(t)$ and $D(t)N(t)$ are martingales. Apply Remark 9.2.5 to show that $S^{(N)}(t) = \frac{S(t)}{N(t)}$ is a martingale under $\tilde{\mathbb{P}}^{(N)}$ defined by (9.2.6).

### (i) Proof of Remark 9.2.5

**Goal:** Show that if $M_1(t)$ is a martingale under $\mathbb{P}$, then $\frac{M_1(t)}{M_2(t)}$ is a martingale under $\mathbb{P}^{(M_2)}$.

Since $M_2(t)$ is a $\mathbb{P}$-martingale with $M_2(t) > 0$ and $M_2(0) = 1$, it is a valid Radon-Nikodým derivative process, defining:

$$\mathbb{P}^{(M_2)}(A) = \int_A M_2(T)\,d\mathbb{P}, \quad A \in \mathcal{F}(T)$$

By Lemma 5.2.2, for $0 \leq s \leq t \leq T$:

$$\mathbb{E}^{(M_2)}\left[\frac{M_1(t)}{M_2(t)}\bigg|\mathcal{F}(s)\right] = \frac{1}{M_2(s)}\mathbb{E}\left[\frac{M_1(t)}{M_2(t)}\cdot M_2(t)\bigg|\mathcal{F}(s)\right] = \frac{1}{M_2(s)}\mathbb{E}\left[M_1(t)\big|\mathcal{F}(s)\right] = \frac{M_1(s)}{M_2(s)}$$

where the last equality uses the fact that $M_1(t)$ is a $\mathbb{P}$-martingale. Therefore $\frac{M_1(t)}{M_2(t)}$ is a martingale under $\mathbb{P}^{(M_2)}$. $\blacksquare$

### (ii) $S^{(N)}(t) = \frac{S(t)}{N(t)}$ is a martingale under $\tilde{\mathbb{P}}^{(N)}$

Define:

$$M_1(t) = \frac{D(t)S(t)}{N(0)}, \quad M_2(t) = \frac{D(t)N(t)}{N(0)}$$

Since $D(t)S(t)$ and $D(t)N(t)$ are martingales under $\tilde{\mathbb{P}}$, so are $M_1(t)$ and $M_2(t)$. Furthermore, $M_2(t) > 0$ (since $D(t) > 0$ and $N(t) > 0$) and $M_2(0) = \frac{D(0)N(0)}{N(0)} = 1$.

By Remark 9.2.5, $\frac{M_1(t)}{M_2(t)}$ is a martingale under $\tilde{\mathbb{P}}^{(N)}$ defined by $M_2$. Since:

$$\frac{M_1(t)}{M_2(t)} = \frac{D(t)S(t)/N(0)}{D(t)N(t)/N(0)} = \frac{S(t)}{N(t)} = S^{(N)}(t)$$

Therefore $S^{(N)}(t) = \frac{S(t)}{N(t)}$ is a martingale under $\tilde{\mathbb{P}}^{(N)}$. $\blacksquare$

# Exercise 9.2 (Portfolios under change of numeraire).
Consider two assets with prices $S(t)$ and $N(t)$ given by
$$S(t) = S(0) \exp\left\{\sigma\tilde{W}(t) + \left(r - \frac{1}{2} \sigma^2\right)t\right\},$$
$$N(t) = N(0)\exp\left\{\nu\tilde{W}(t) + \left(r - \frac{1}{2} \nu^2 \right)t\right\},$$
where $\tilde{W}(t)$ is a one-dimensional Brownian motion under the risk-neutral measure $\tilde{\mathbb{P}}$ and the volatilities $\sigma > 0$ and $\nu > 0$ are constant, as is the interest rate $r$. We define a third asset, the money market account, whose price per share at time $t$ is $M(t) = e^{rt}.$

Let us now denominate prices in terms of the numeraire $N$, so that the redenominated first asset price is 
$$\hat{S}(t) = \frac{S(t)}{N(t)}$$
and the redenominated money market account price is 
$$\hat{M}(t) = \frac{M(t)}{N(t)}.$$
According to Theorem 9.2.2, $d\hat{S}(t) = (\sigma - \nu) \hat{S}(t) d\hat{W}(t),$ where $\hat{W}(t) = \tilde{W}(t) - \nu t.$

(i) Compute the differential of $\frac{1}{N(t)}.$

(ii) Compute the differential of $\hat{M}(t)$, expressing it in terms of $d\hat{W}(t).$

Consider a portfolio that at each time $t$ holds $\Delta(t)$ shares of the first asset and finances this by investing in or borrowing from the money market. According to the usual formula, the differential of the value $X(t)$ of this portfolio is 
$$dX(t) = \Delta(t)dS(t) + r(X(t) - \Delta(t)S(t))dt.$$
We define
$$\Gamma(t) = \frac{X(t) - \Delta(t) S(t)}{M(t)}$$
to be the number of shares of money market account held by this portfolio at time $t$ and can then rewrite the differential of $X(t)$ as
$$dX(t) = \Delta(t)dS(t) + \Gamma(t)dM(t). \tag{9.7.1}$$
Note also that by the definition of $\Gamma(t),$ we have
$$X(t) = \Delta(t)S(t) + \Gamma(t)M(t). \tag{9.7.2}$$
We redenominate the portfolio value, defining 
$$\hat{X}(t) = \frac{X(t)}{N(t)}, \tag{9.7.3}$$
so that (dividing (9.7.2) by $N(t)$) we have
$$\hat{X}(t) = \Delta(t)\hat{S}(t) + \Gamma(t) \hat{M}(t). \tag{9.7.4}$$

(iii) Use stochastic calculus to show that
$$d\hat{X}(t) = \Delta(t)d\hat{S}(t) + \Gamma(t)d\hat{M}(t).$$
This equation is the counterpart in the neq numeraire of equation (9.7.1) and says that the change in $\hat{X}(t)$ is solely due to changes in the prices of the assets held by the portfolio. (Hint: Start from equation (9.7.3) and use (9.7.1) and (9.7.4) along the way.)

### (i) Differential of $\frac{1}{N(t)}$

Since $dN(t) = rN(t)dt + \nu N(t)d\tilde{W}(t)$, applying Itô's lemma to $f(x) = \frac{1}{x}$:

$$d\frac{1}{N(t)} = -\frac{1}{N^2}dN + \frac{1}{N^3}(dN)^2 = \frac{1}{N(t)}\left[(\nu^2-r)dt - \nu d\tilde{W}(t)\right] \quad \blacksquare$$

### (ii) Differential of $\hat{M}(t)$

By Itô's product rule applied to $\hat{M}(t) = M(t)\cdot\frac{1}{N(t)}$, using $dM(t) = rM(t)dt$ (no diffusion term, so $dM\cdot d\frac{1}{N} = 0$):

$$d\hat{M}(t) = \frac{1}{N(t)}dM(t) + M(t)d\frac{1}{N(t)} = \hat{M}(t)\left[rdt + (\nu^2-r)dt - \nu d\tilde{W}(t)\right] = \hat{M}(t)\left[\nu^2 dt - \nu d\tilde{W}(t)\right]$$

Substituting $d\tilde{W}(t) = d\hat{W}(t) + \nu dt$:

$$d\hat{M}(t) = \hat{M}(t)\left[\nu^2 dt - \nu(d\hat{W}(t)+\nu dt)\right] = -\nu\hat{M}(t)d\hat{W}(t) \quad \blacksquare$$

### (iii) $d\hat{X}(t) = \Delta(t)d\hat{S}(t) + \Gamma(t)d\hat{M}(t)$

Applying Itô's product rule to $\hat{X}(t) = X(t)\cdot\frac{1}{N(t)}$:

$$d\hat{X}(t) = \frac{1}{N(t)}dX(t) + X(t)d\frac{1}{N(t)} + dX(t)\,d\frac{1}{N(t)}$$

**Cross variation term:** From (9.7.1), $dX = \Delta\,dS + \Gamma\,dM$. Since $dM\cdot d\frac{1}{N} = 0$ and $dS\cdot d\frac{1}{N} = \sigma S\cdot(-\nu)\frac{1}{N}dt = -\nu\sigma\hat{S}\,dt$:

$$dX\cdot d\frac{1}{N} = -\nu\sigma\Delta(t)\hat{S}(t)dt$$

**Second term:** Using (9.7.4), $X(t) = \Delta S + \Gamma M$:

$$X(t)d\frac{1}{N(t)} = \left(\Delta\hat{S}+\Gamma\hat{M}\right)\left[(\nu^2-r)dt - \nu d\tilde{W}\right]$$

**Collecting the $\Delta$ terms:**

$$\frac{\Delta\,dS}{N} + \Delta\hat{S}\left[(\nu^2-r)dt-\nu d\tilde{W}\right] - \nu\sigma\Delta\hat{S}\,dt$$

$$= \Delta\hat{S}\left[r\,dt+\sigma d\tilde{W} + (\nu^2-r)dt - \nu d\tilde{W} - \nu\sigma\,dt\right]$$

$$= \Delta\hat{S}(\sigma-\nu)\left[d\tilde{W}-\nu\,dt\right] = \Delta\hat{S}(\sigma-\nu)d\hat{W}(t) = \Delta(t)d\hat{S}(t)$$

**Collecting the $\Gamma$ terms:**

$$\frac{\Gamma\,dM}{N} + \Gamma\hat{M}\left[(\nu^2-r)dt - \nu d\tilde{W}\right] = \Gamma\hat{M}\left[r\,dt+(\nu^2-r)dt-\nu d\tilde{W}\right]$$

$$= \Gamma\hat{M}\left[\nu^2\,dt-\nu d\tilde{W}\right] = \Gamma\hat{M}(-\nu)d\hat{W}(t) = \Gamma(t)d\hat{M}(t)$$

**Conclusion:**

$$d\hat{X}(t) = \Delta(t)d\hat{S}(t) + \Gamma(t)d\hat{M}(t) \quad \blacksquare$$

This shows that the change in the redenominated portfolio value $\hat{X}(t)$ is solely due to changes in the redenominated asset prices $\hat{S}(t)$ and $\hat{M}(t)$, confirming that the self-financing property is preserved under change of numéraire.

# Exercise 9.3 (Change in volatility caused by change of numeraire).
Let $S(t)$ and $N(t)$ be the prices of two assets, denominated in a common currency, and let $\sigma$ and $\nu$ denote their volatilities, which we assume are constant. We assume also that the interest rate $r$ is constant. Then
$$dS(t) = rS(t)dt + \sigma S(t)d\tilde{W}_1(t),$$
$$dN(t) = rN(t)dt + \nu N(t) d\tilde{W}_3(t),$$
where $\tilde{W}_1(t)$ and $\tilde{W}_3(t)$ are Brownian motions under the risk-neutral measure $\tilde{mathbb{P}}$. We assume these Brownian motions are correlated, with $d\tilde{W}_1(t)d\tilde{W}_3(t) = \rho dt$ for some constant $\rho.$

(i) Show that $S^{(N)}(t) = \frac{S(t)}{N(t)}$ has volatility $\gamma = \sqrt{\sigma^2 - 2\rho \sigma \nu + \nu^2}.$ In other words, show that there exists a Brownian motion $\tilde{W}$ under $\tilde{\mathbb{P}}$ such that 
$$\frac{dS^{(N)}(t)}{S^{(N)}(t)} = (\text{Something})dt + \gamma\tilde{W}_4(t).$$

(ii) Show how to construct a Brownian motion $\tilde{W}_2(t)$ under $\tilde{\mathbb{P}}$ that is independent of $\tilde{W}_1(t)$ such that $dN(t)$ may be written as 
$$dN(t) = rN(t)dt + \nu N(t)[\rho d\tilde{W}_1(t) + \sqrt{1 - \rho^2} d\tilde{W}_2(t)].$$

(iii) Using Theorem 9.2.2, determine the volatility vector of $S^{(N)}(t).$ In other words, find a vector $(v_1, v_2)$ such that
$$dS^{(N)}(t) = S^{(N)}(t)[v_1d\tilde{W}_1^{(N)} (t) + v_2 d\tilde{W}_2^{(N)}(t)],$$
where $\tilde{W}_1(t)$ and $\tilde{W}_2(t)$ are independent Brownian motions under $\tilde{\mathbb{P}}^{(N)}.$ Show that
$$\sqrt{v_1^2 + v_2^2} = \sqrt{\sigma^2 - 2\rho \sigma \nu + \nu^2}.$$

### (i) Volatility of $S^{(N)}(t)$

Writing:

$$S^{(N)}(t) = \frac{S(0)}{N(0)}\exp\left\{\left(\frac{1}{2}\nu^2-\frac{1}{2}\sigma^2\right)t + \sigma\tilde{W}_1(t) - \nu\tilde{W}_3(t)\right\}$$

Let $X(t) = \left(\frac{1}{2}\nu^2-\frac{1}{2}\sigma^2\right)t + \sigma\tilde{W}_1(t) - \nu\tilde{W}_3(t)$, so $S^{(N)}(t) = S^{(N)}(0)e^{X(t)}$.

Computing:

$$dX(t) = \left(\frac{1}{2}\nu^2-\frac{1}{2}\sigma^2\right)dt + \sigma d\tilde{W}_1(t) - \nu d\tilde{W}_3(t)$$

$$(dX)^2 = \sigma^2 dt - 2\rho\sigma\nu\,dt + \nu^2 dt = \gamma^2 dt, \quad \gamma = \sqrt{\sigma^2-2\rho\sigma\nu+\nu^2}$$

By Itô's lemma:

$$dS^{(N)}(t) = S^{(N)}(t)\left[dX + \frac{1}{2}(dX)^2\right] = S^{(N)}(t)\left[(\nu^2-\rho\sigma\nu)dt + \sigma d\tilde{W}_1(t) - \nu d\tilde{W}_3(t)\right]$$

Define:

$$d\tilde{W}_4(t) = \frac{\sigma}{\gamma}d\tilde{W}_1(t) - \frac{\nu}{\gamma}d\tilde{W}_3(t)$$

Then $\tilde{W}_4(0) = 0$ and $d\tilde{W}_4\,d\tilde{W}_4 = \frac{1}{\gamma^2}(\sigma^2-2\rho\sigma\nu+\nu^2)dt = dt$, so by Lévy's theorem $\tilde{W}_4$ is a Brownian motion under $\tilde{\mathbb{P}}$. Therefore:

$$\frac{dS^{(N)}(t)}{S^{(N)}(t)} = (\nu^2-\rho\sigma\nu)dt + \gamma\,d\tilde{W}_4(t) \quad \blacksquare$$

### (ii) Constructing $\tilde{W}_2(t)$ independent of $\tilde{W}_1(t)$

Define:

$$\tilde{W}_2(t) = \frac{-\rho}{\sqrt{1-\rho^2}}\tilde{W}_1(t) + \frac{1}{\sqrt{1-\rho^2}}\tilde{W}_3(t)$$

**Verification:** $d\tilde{W}_2\,d\tilde{W}_2 = \frac{\rho^2}{1-\rho^2}dt - \frac{2\rho}{1-\rho^2}\rho\,dt + \frac{1}{1-\rho^2}dt = dt$, so $\tilde{W}_2$ is a Brownian motion by Lévy's theorem.

**Independence:** $d\tilde{W}_2\,d\tilde{W}_1 = \frac{-\rho}{\sqrt{1-\rho^2}}dt + \frac{\rho}{\sqrt{1-\rho^2}}dt = 0$, so $\tilde{W}_1$ and $\tilde{W}_2$ are independent.

**Rewriting $dN(t)$:** Inverting, $d\tilde{W}_3 = \rho\,d\tilde{W}_1 + \sqrt{1-\rho^2}\,d\tilde{W}_2$, so:

$$dN(t) = rN(t)dt + \nu N(t)\left[\rho\,d\tilde{W}_1(t) + \sqrt{1-\rho^2}\,d\tilde{W}_2(t)\right] \quad \blacksquare$$

### (iii) Volatility vector of $S^{(N)}(t)$ under $\tilde{\mathbb{P}}^{(N)}$

From (ii), the volatility vectors in the $(\tilde{W}_1, \tilde{W}_2)$ basis are:

$$\boldsymbol{\sigma} = (\sigma, 0), \quad \boldsymbol{\nu} = (\nu\rho,\ \nu\sqrt{1-\rho^2})$$

By Theorem 9.2.2, under $\tilde{\mathbb{P}}^{(N)}$:

$$d\tilde{W}_i^{(N)}(t) = d\tilde{W}_i(t) - \nu_i\,dt, \quad i = 1, 2$$

where $\nu_1 = \nu\rho$ and $\nu_2 = \nu\sqrt{1-\rho^2}$. The volatility vector of $S^{(N)}$ is:

$$(v_1, v_2) = \boldsymbol{\sigma} - \boldsymbol{\nu} = (\sigma - \nu\rho,\ -\nu\sqrt{1-\rho^2})$$

so that:

$$dS^{(N)}(t) = S^{(N)}(t)\left[v_1\,d\tilde{W}_1^{(N)}(t) + v_2\,d\tilde{W}_2^{(N)}(t)\right]$$

**Verification:**

$$\sqrt{v_1^2+v_2^2} = \sqrt{(\sigma-\nu\rho)^2 + \nu^2(1-\rho^2)} = \sqrt{\sigma^2 - 2\rho\sigma\nu + \nu^2\rho^2 + \nu^2 - \nu^2\rho^2}$$

$$= \sqrt{\sigma^2-2\rho\sigma\nu+\nu^2} = \gamma \quad \blacksquare$$

# Exercise 9.4.
From the differential formulas (9.3.14) and (9.3.15) for the stock and discounted exchange rate in terms of the Brownian motions under the domestic risk-neutral measure, derive the differential formulas (9.3.22) and (9.3.23) for the redenominated money market account and stock discounted at the foreign interst rate and written in terms of the Brownian motions under the foreign risk-neutral measure.
$$dS(t) = S(t)[R(t)dt + \sigma_1(t)d\tilde{W}_1(t)] \tag{9.3.14}$$
$$d(M^f(t)Q(t)) = M^f(t)Q(t)[R(t)dt + \sigma_2(t)\rho(t)d\tilde{W}_1(t) + \sigma_2(t)\sqrt{1 - \rho^2(t)}d\tilde{W}_2(t)] $$
$$ = M^f(t)Q(t)[R(t)dt + \sigma_2(t)d\tilde{W}_3(t)] \tag{9.3.15}$$

### Step 1: Differential of $\frac{1}{M^f(t)Q(t)}$

From (9.3.15), $d(M^f Q) = M^f Q\left[R\,dt + \sigma_2\rho\,d\tilde{W}_1 + \sigma_2\sqrt{1-\rho^2}\,d\tilde{W}_2\right]$.

Applying Itô's lemma with $f(x) = \frac{1}{x}$:

$$d\frac{1}{M^fQ} = \frac{1}{M^fQ}\left[(\sigma_2^2-R)dt - \sigma_2\rho\,d\tilde{W}_1 - \sigma_2\sqrt{1-\rho^2}\,d\tilde{W}_2\right]$$

### Step 2: Change of measure to foreign risk-neutral measure

By Theorem 9.2.2, with numeraire $M^f(t)Q(t)$ and volatility vector $\boldsymbol{\sigma}_2 = (\sigma_2\rho,\ \sigma_2\sqrt{1-\rho^2})$:

$$d\tilde{W}_1^f(t) = d\tilde{W}_1(t) - \sigma_2(t)\rho(t)\,dt$$

$$d\tilde{W}_2^f(t) = d\tilde{W}_2(t) - \sigma_2(t)\sqrt{1-\rho^2(t)}\,dt$$

### (9.3.22): Redenominated domestic money market account

$$d\left(\frac{M(t)D^f(t)}{Q(t)}\right) = d\left(M(t)\cdot\frac{1}{M^f(t)Q(t)}\right)$$

By Itô's product rule, using $dM(t) = R(t)M(t)dt$ (no diffusion term):

$$= M(t)\cdot d\frac{1}{M^fQ} + \frac{1}{M^fQ}dM + dM\cdot d\frac{1}{M^fQ}$$

The cross term vanishes ($dM$ has no $d\tilde{W}$ term):

$$= \frac{M D^f}{Q}\left[(\sigma_2^2-R)dt - \sigma_2\rho\,d\tilde{W}_1 - \sigma_2\sqrt{1-\rho^2}\,d\tilde{W}_2 + R\,dt\right]$$

$$= \frac{MD^f}{Q}\left[\sigma_2^2\,dt - \sigma_2\rho\,d\tilde{W}_1 - \sigma_2\sqrt{1-\rho^2}\,d\tilde{W}_2\right]$$

Substituting $d\tilde{W}_i = d\tilde{W}_i^f + \sigma_2\nu_i\,dt$ (where $\nu_1=\rho$, $\nu_2=\sqrt{1-\rho^2}$), the $dt$ terms cancel:

$$d\left(\frac{MD^f}{Q}\right) = \frac{MD^f}{Q}\left[\sigma_2(t)\rho(t)\,d\tilde{W}_1^f(t) + \sigma_2(t)\sqrt{1-\rho^2(t)}\,d\tilde{W}_2^f(t)\right] \quad \blacksquare$$

### (9.3.23): Stock discounted at foreign rate

$$d\left(\frac{D^f(t)S(t)}{Q(t)}\right) = d\left(S(t)\cdot\frac{1}{M^f(t)Q(t)}\right)$$

By Itô's product rule:

$$= \frac{1}{M^fQ}dS + S\cdot d\frac{1}{M^fQ} + dS\cdot d\frac{1}{M^fQ}$$

**Cross term:**

$$dS\cdot d\frac{1}{M^fQ} = S\left[\sigma_1 d\tilde{W}_1\right]\cdot\left[-\sigma_2\rho\,d\tilde{W}_1 - \sigma_2\sqrt{1-\rho^2}\,d\tilde{W}_2\right] = -\sigma_1\sigma_2\rho\,\frac{D^fS}{Q}dt$$

**Combining all $dt$ terms:**

$$\frac{D^fS}{Q}\left[R + \sigma_2^2 - R - \sigma_1\sigma_2\rho\right]dt = \frac{D^fS}{Q}\left[\sigma_2^2 - \sigma_1\sigma_2\rho\right]dt$$

**Combining all $d\tilde{W}$ terms:**

$$\frac{D^fS}{Q}\left[\sigma_1 d\tilde{W}_1 - \sigma_2\rho\,d\tilde{W}_1 - \sigma_2\sqrt{1-\rho^2}\,d\tilde{W}_2\right]$$

Substituting $d\tilde{W}_i = d\tilde{W}_i^f + \sigma_2\nu_i\,dt$, the $dt$ terms cancel, leaving:

$$d\left(\frac{D^fS}{Q}\right) = \frac{D^fS}{Q}\left[(\sigma_1-\sigma_2\rho)\,d\tilde{W}_1^f(t) - \sigma_2\sqrt{1-\rho^2}\,d\tilde{W}_2^f(t)\right] \quad \blacksquare$$

# Exercise 9.5(Quanto option).
A quanto option pays off in one currency the price in another currency of an underlying asset without taking the currency conversion into account. For example, a quanto call on a British asset struck at \$25 would pay \$5 if the price of the asset upon expiration of the option is £30. To compute the payoff of the option, the price 30 is treated as if it were dollars, even though it is pounds sterling.

In this problem we consider a quanto option in the foreign exchange model of Section 9.3. We take the domestic and foreign interest rates to be constants $r$ and $r^f$, respectively, and we assume that $\sigma_1 > 0, \sigma_2 > 0$ and $\rho \in (-1, 1)$ are likewise constant.

(i) From (9.3.14), show that
$$S(t) = S(0)\exp\left\{\sigma_1\tilde{W}_1(t) + \left(r - \frac{1}{2}\sigma_1^2\right)t\right\}.$$
(ii) From (9.3.16), show that
$$Q(t) = Q(0)\exp\left\{\sigma_2\rho\tilde{W}_1(t) + \sigma_2 \sqrt{1 - \rho}\tilde{W}_2(t) + \left(r - r^f - \frac{1}{2} \sigma_2^2\right)t\right\}.$$
(iii) Show that 
$$\frac{S(t)}{Q(t)} = \frac{S(0)}{Q(0)}\exp\left\{\sigma_4\tilde{W}_4(t) + \left(r - a - \frac{1}{2}\sigma_4^2\right)t\right\},$$
where 
$$\sigma_4 = \sqrt{\sigma_1^2 - 2\rho \sigma_1 \sigma_2 + \sigma_2^2},$$
$$a = r - r^f + \rho \sigma_1 \sigma_2 - \sigma_2^2,$$
and 
$$\tilde{W}_4(t) = \frac{\sigma_1 - \sigma_2\rho}{\sigma_4}\tilde{W}_1(t) - \frac{\sigma_2\sqrt{1 - \rho^2}}{\sigma_4} \tilde{W}_2(t)$$
is a Brownian motion.

(iv) Consider a quanto call that pays off
$$\left(\frac{S(T)}{Q(T)} - K\right)^+$$
unites of domestic currency at time $T$. (Note that $\frac{S(T)}{Q(T)}$ is denominated in unites of foreing currency, but in this payoff it is treated as if it is a number of units of domestic currency.) Show that if at time $t \in [0, T]$ we have $\frac{S(t)}{Q(t)} = x$, then the price of the quanto call at this time is
$$q(t, x) = xre^{-ar} N(d_+(\tau, x)) - e^{-r\tau}KN(d_-(\tau, x)),$$
where $\tau = T - t$ and
$$d_{\pm}(\tau, x) = \frac{1}{\sigma_4\sqrt{\tau}}\left[\log\frac{x}{K} + \left(r - a \pm \frac{1}{2} \sigma_4^2\right)\tau\right].$$
(Hint: Argue that this is a case of formula (5.5.12).)

### (i) Formula for $S(t)$

From (9.3.14): $dS(t) = S(t)\left[r\,dt + \sigma_1 d\tilde{W}_1(t)\right]$.

Applying Itô's lemma to $f(x) = \log x$:

$$d\log S(t) = \left(r - \frac{1}{2}\sigma_1^2\right)dt + \sigma_1\,d\tilde{W}_1(t)$$

Integrating:

$$S(t) = S(0)\exp\left\{\sigma_1\tilde{W}_1(t) + \left(r-\frac{1}{2}\sigma_1^2\right)t\right\} \quad \blacksquare$$

### (ii) Formula for $Q(t)$

From (9.3.16): $dQ(t) = Q(t)\left[(r-r^f)dt + \sigma_2\rho\,d\tilde{W}_1(t) + \sigma_2\sqrt{1-\rho^2}\,d\tilde{W}_2(t)\right]$.

Applying Itô's lemma to $f(x) = \log x$:

$$d\log Q(t) = \left(r-r^f-\frac{1}{2}\sigma_2^2\right)dt + \sigma_2\rho\,d\tilde{W}_1(t) + \sigma_2\sqrt{1-\rho^2}\,d\tilde{W}_2(t)$$

Integrating:

$$Q(t) = Q(0)\exp\left\{\sigma_2\rho\tilde{W}_1(t) + \sigma_2\sqrt{1-\rho^2}\tilde{W}_2(t) + \left(r-r^f-\frac{1}{2}\sigma_2^2\right)t\right\} \quad \blacksquare$$

### (iii) Formula for $\frac{S(t)}{Q(t)}$

Dividing (i) by (ii):

$$\frac{S(t)}{Q(t)} = \frac{S(0)}{Q(0)}\exp\left\{(\sigma_1-\sigma_2\rho)\tilde{W}_1(t) - \sigma_2\sqrt{1-\rho^2}\tilde{W}_2(t) + \left(r^f-\frac{1}{2}\sigma_1^2+\frac{1}{2}\sigma_2^2\right)t\right\}$$

Define $\tilde{W}_4(t) = \frac{\sigma_1-\sigma_2\rho}{\sigma_4}\tilde{W}_1(t) - \frac{\sigma_2\sqrt{1-\rho^2}}{\sigma_4}\tilde{W}_2(t)$.

**Verification that $\tilde{W}_4$ is a Brownian motion:**

$$d\tilde{W}_4\,d\tilde{W}_4 = \frac{(\sigma_1-\sigma_2\rho)^2}{\sigma_4^2}dt + \frac{\sigma_2^2(1-\rho^2)}{\sigma_4^2}dt = \frac{\sigma_1^2-2\rho\sigma_1\sigma_2+\sigma_2^2}{\sigma_4^2}dt = dt$$

By Lévy's theorem, $\tilde{W}_4$ is a Brownian motion. $\blacksquare$

**Rewriting the exponent:** The drift term satisfies:

$$r^f - \frac{1}{2}\sigma_1^2 + \frac{1}{2}\sigma_2^2 = r - a - \frac{1}{2}\sigma_4^2$$

since $a = r - r^f + \rho\sigma_1\sigma_2 - \sigma_2^2$ and $\sigma_4^2 = \sigma_1^2 - 2\rho\sigma_1\sigma_2 + \sigma_2^2$. Therefore:

$$\frac{S(t)}{Q(t)} = \frac{S(0)}{Q(0)}\exp\left\{\sigma_4\tilde{W}_4(t) + \left(r-a-\frac{1}{2}\sigma_4^2\right)t\right\} \quad \blacksquare$$

### (iv) Quanto call price $q(t,x)$

The price at time $t$ with $\frac{S(t)}{Q(t)} = x$ is:

$$q(t,x) = \tilde{\mathbb{E}}\left[e^{-r\tau}\left(\frac{S(T)}{Q(T)}-K\right)^+\right]$$

From (iii), writing $\frac{S(T)}{Q(T)} = x\exp\left\{-\sigma_4\sqrt{\tau}Y + \left(r-a-\frac{1}{2}\sigma_4^2\right)\tau\right\}$ where $Y = -\frac{\tilde{W}_4(T)-\tilde{W}_4(t)}{\sqrt{\tau}}\sim\mathcal{N}(0,1)$.

**The condition $\frac{S(T)}{Q(T)} > K$ is equivalent to $Y < d_-(\tau,x)$:**

$$-\sigma_4\sqrt{\tau}Y + \left(r-a-\frac{1}{2}\sigma_4^2\right)\tau > \log\frac{K}{x} \iff Y < \frac{1}{\sigma_4\sqrt{\tau}}\left[\log\frac{x}{K}+\left(r-a-\frac{1}{2}\sigma_4^2\right)\tau\right] = d_-(\tau,x)$$

**Computing the expectation:**

$$q(t,x) = e^{-r\tau}xe^{(r-a-\frac{1}{2}\sigma_4^2)\tau}\int_{-\infty}^{d_-}e^{-\sigma_4\sqrt{\tau}y}\phi(y)\,dy - e^{-r\tau}KN(d_-)$$

**First term** (completing the square, substituting $z = y+\sigma_4\sqrt{\tau}$, upper limit $d_-+\sigma_4\sqrt{\tau}=d_+$):

$$e^{-r\tau}xe^{(r-a-\frac{1}{2}\sigma_4^2)\tau}\cdot e^{\frac{1}{2}\sigma_4^2\tau}\cdot N(d_+) = xe^{-a\tau}N(d_+)$$

**Conclusion:**

$$q(t,x) = xe^{-a\tau}N(d_+(\tau,x)) - e^{-r\tau}KN(d_-(\tau,x)) \quad \blacksquare$$

**Remark:** This is a case of formula (5.5.12) with dividend rate $a = r-r^f+\rho\sigma_1\sigma_2-\sigma_2^2$ and volatility $\sigma_4$, applied to the process $\frac{S(t)}{Q(t)}$ which follows a GBM under $\tilde{\mathbb{P}}$.
