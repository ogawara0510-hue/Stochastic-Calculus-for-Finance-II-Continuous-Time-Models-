# Exercise 6.1.
Consider the stochastic differential equation
$$dX(u) = (a(u) + b(u)X(u))du + (\gamma(u) + \sigma(u)X(u))dW(u), $$
where $W(u)$ is a Brownian motion relative to a filtration $\mathcal{F}(u), u \geq 0,$ and we allow $a(u), b(u),$ and $\sigma(u)$ to be processes adapted to this filtration. Fix an initial time $t \geq 0$ and an initial position $x \in \mathbb{R}.$ Define

$$Z(u) = \exp \left\{ \int_{t}^{u} \sigma(v)dW(v) + \int_{t}^{u}\left(b(v) - \frac{1}{2} \sigma^2(v) \right)dv \right\},$$

$$Y(u) = x + \int_{t}^{u} \frac{a(v) - \sigma(v)\gamma(v)}{Z(v)} dv + \int_{t}^{u} \frac{\gamma(v)}{Z(v)} dW(v).$$
(i) Show that $Z(t) = 14$ and 
$$dZ(u) = b(u)Z(u) du + \sigma(u)Z(u)dW(u), u \geq t.$$
(ii) By its very definition, $Y(u)$ satisfies $Y(t) = x$ and
$$dY(u) = \frac{a(u) - \sigma(u)\gamma(u)}{Z(u)}du + \frac{\gamma(u)}{Z(u)} dW(u), u \geq t.$$
Show that $X(u) = Y(u) Z(u)$ solbes the stochastic differential equation (6.2.4) and satisfies the initial condition $X(t) = x.$

### (i) $Z(t) = 1$ and $dZ(u)$

**Initial condition:**

$$Z(t) = \exp\left\{\int_t^t \sigma(v)dW(v) + \int_t^t\left(b(v)-\frac{1}{2}\sigma^2(v)\right)dv\right\} = e^0 = 1 \quad \blacksquare$$

**SDE for $Z(u)$:** By Itô's lemma applied to $Z(u) = e^{I(u)}$ where $dI(u) = \sigma(u)dW(u) + (b(u)-\frac{1}{2}\sigma^2(u))du$:

$$dZ(u) = Z(u)\,dI(u) + \frac{1}{2}Z(u)(dI)^2$$

$$= Z(u)\left[\left(b(u)-\frac{1}{2}\sigma^2(u)\right)du + \sigma(u)dW(u)\right] + \frac{1}{2}Z(u)\sigma^2(u)du$$

$$= b(u)Z(u)du + \sigma(u)Z(u)dW(u) \quad \blacksquare$$

### (ii) $X(u) = Y(u)Z(u)$ solves the SDE

**Initial condition:** $X(t) = Y(t)Z(t) = x \cdot 1 = x$ 

**SDE:** By Itô's product rule:

$$dX = Y\,dZ + Z\,dY + dY\,dZ$$

**Term 1:**

$$Y\,dZ = Y(u)\left[b(u)Z(u)du + \sigma(u)Z(u)dW(u)\right]$$

**Term 2:**

$$Z\,dY = Z(u)\left[\frac{a(u)-\sigma(u)\gamma(u)}{Z(u)}du + \frac{\gamma(u)}{Z(u)}dW(u)\right] = (a(u)-\sigma(u)\gamma(u))du + \gamma(u)dW(u)$$

**Term 3:**

$$dY\,dZ = \frac{\gamma(u)}{Z(u)}dW\cdot\sigma(u)Z(u)dW = \sigma(u)\gamma(u)dt$$

**Summing all terms:**

$$dX = \left[b(u)Y(u)Z(u) + a(u) - \sigma(u)\gamma(u) + \sigma(u)\gamma(u)\right]du + \left[\sigma(u)Y(u)Z(u) + \gamma(u)\right]dW(u)$$

$$= (a(u) + b(u)X(u))du + (\gamma(u) + \sigma(u)X(u))dW(u) \quad \blacksquare$$

Therefore $X(u) = Y(u)Z(u)$ solves (6.2.4) with initial condition $X(t) = x$.

### (i) Discounted portfolio value

The portfolio satisfies:

$$dX(t) = \Delta_1(t)df(t,R,T_1) + \Delta_2(t)df(t,R,T_2) + R(t)(X(t) - \Delta_1(t)f(t,R,T_1) - \Delta_2(t)f(t,R,T_2))dt$$

By Itô's lemma:

$$df(t,R,T_i) = \left[f_t + \alpha f_r + \frac{1}{2}\gamma^2 f_{rr}\right]dt + \gamma f_r\,dW$$

Computing $d(D(t)X(t)) = D(t)dX(t) - R(t)D(t)X(t)dt$, the $R(t)X(t)$ terms cancel, giving:

$$d(D(t)X(t)) = \sum_{i=1}^{2}\Delta_i(t)D(t)\left[-Rf + f_t + \alpha f_r + \frac{1}{2}\gamma^2 f_{rr}\right]_{T_i}dt + D(t)\gamma\left[\Delta_1 f_r^{T_1} + \Delta_2 f_r^{T_2}\right]dW$$

$$-Rf + f_t + \alpha f_r + \frac{1}{2}\gamma^2 f_{rr} = (\alpha - \beta)f_r$$

Therefore:

$$d(D(t)X(t)) = \Delta_1 D(\alpha-\beta_1)f_r^{T_1}dt + \Delta_2 D(\alpha-\beta_2)f_r^{T_2}dt + D\gamma[\Delta_1 f_r^{T_1} + \Delta_2 f_r^{T_2}]dW \quad \blacksquare$$

### (ii) $\beta$ does not depend on $T$

Choose:

$$\Delta_1(t) = S(t)f_r(t,R,T_2), \quad \Delta_2(t) = -S(t)f_r(t,R,T_1)$$

**Eliminating the $dt$ term:**

$$\Delta_1 f_r^{T_1} + \Delta_2 f_r^{T_2} = S(t)\left[f_r^{T_2}f_r^{T_1} - f_r^{T_1}f_r^{T_2}\right] = 0$$

So the portfolio is instantaneously riskless.

**The $dt$ term becomes:**

$$d(D(t)X(t)) = S(t)D(t)\left[(\alpha-\beta_1)f_r^{T_1}f_r^{T_2} - (\alpha-\beta_2)f_r^{T_2}f_r^{T_1}\right]dt$$

$$= S(t)D(t)(\beta_2 - \beta_1)f_r^{T_1}f_r^{T_2}\,dt$$

$$= D(t)\left|(\beta_2-\beta_1)f_r^{T_1}f_r^{T_2}\right|dt \geq 0$$

Since $X(0) = 0$ and $d(D(t)X(t)) \geq 0$, this is an arbitrage unless $(\beta_2 - \beta_1)f_r^{T_1}f_r^{T_2} = 0$, i.e., $\beta_1 = \beta_2$. Since $T_1, T_2$ are arbitrary, $\beta(t,r,T)$ does not depend on $T$. $\blacksquare$

### (iii) Case $f_r(t,r,T) = 0$

The portfolio satisfies:

$$dX(t) = \Delta(t)df(t,R,T) + R(t)(X(t) - \Delta(t)f(t,R,T))dt$$

Therefore:

$$d(D(t)X(t)) = \Delta(t)D(t)\left[-Rf + f_t + \alpha f_r + \frac{1}{2}\gamma^2 f_{rr}\right]dt + D(t)\Delta(t)\gamma f_r\,dW \quad \blacksquare$$

If $f_r(t,r,T) = 0$, then the $dW$ term vanishes and:

$$d(D(t)X(t)) = \Delta(t)D(t)\left[f_t + \frac{1}{2}\gamma^2 f_{rr} - Rf\right]dt$$

This is a riskless portfolio starting at $X(0) = 0$. To avoid arbitrage, we need:

$$f_t(t,r,T) + \frac{1}{2}\gamma^2(t,r)f_{rr}(t,r,T) = rf(t,r,T) \quad \blacksquare$$

# Exercise 6.3 (Solution of Hull-White model).
This exervise solves the ordinary differential equations (6.5.8) and (6.5.9) to produce the solutions $C(t, T$ and $A(t, T)$ given in (6.5.10) and (6.5.11).

$$C'(t, T) = b(t)C(t, T) - 1, \tag{6.5.8}$$
$$A'(t, T) = -a(t)C(t,T) + \frac{1}{2}\sigma^2(t)C^2(t,T), \tag{6.5.9}$$
$$C(t, T) = \int_{t}^{T}e^{-\int_{t}^{s} b(v)dv} ds, \tag{6.5.10}$$
$$A(t, T) = \int_{t}^{T} \left(a(s)C(s, T) - \frac{1}{2} \sigma^2(s) C^2(s, T)\right) ds. \tag{6.5.11}$$

(i) Use equation (6.5.8) with $s$ replacing $t$ to show that
$$\frac{d}{ds}\left[e^{-\int_{0}^{s}b(v)dv} C(s, T) \right] = - e^{-\int_{0}^{s} b(v)dv}.$$

(ii) Integrate the equation in (i) from $s = t$ to $s = T,$ and use the terminal condition $C(T, T)$ to obtain (6.5.10).

(iii) Replace $t$ by $s$ in (6.5.9), integrate the resulting equation from $s = t$ to $ s = T$, use the termianl condition $A(T, T) = 0,$ and obtain (6.5.11).

### (i) Computing the derivative

$$\frac{d}{ds}\left[e^{-\int_0^s b(v)dv}C(s,T)\right] = -b(s)e^{-\int_0^s b(v)dv}C(s,T) + e^{-\int_0^s b(v)dv}C'(s,T)$$

Substituting $C'(s,T) = b(s)C(s,T) - 1$ from (6.5.8):

$$= e^{-\int_0^s b(v)dv}\left[-b(s)C(s,T) + b(s)C(s,T) - 1\right] = -e^{-\int_0^s b(v)dv} \quad \blacksquare$$

### (ii) Deriving $C(t,T)$

Integrating (i) from $s = t$ to $s = T$:

$$\left[e^{-\int_0^s b(v)dv}C(s,T)\right]_{s=t}^{s=T} = -\int_t^T e^{-\int_0^s b(v)dv}ds$$

$$e^{-\int_0^T b(v)dv}\underbrace{C(T,T)}_{=0} - e^{-\int_0^t b(v)dv}C(t,T) = -\int_t^T e^{-\int_0^s b(v)dv}ds$$

Solving for $C(t,T)$:

$$C(t,T) = e^{\int_0^t b(v)dv}\int_t^T e^{-\int_0^s b(v)dv}ds = \int_t^T e^{-\int_t^s b(v)dv}ds \quad \blacksquare$$

### (iii) Deriving $A(t,T)$

Replacing $t$ by $s$ in (6.5.9) and integrating from $s = t$ to $s = T$:

$$\left[A(s,T)\right]_{s=t}^{s=T} = \int_t^T\left[-a(s)C(s,T) + \frac{1}{2}\sigma^2(s)C^2(s,T)\right]ds$$

Using $A(T,T) = 0$:

$$-A(t,T) = \int_t^T\left[-a(s)C(s,T) + \frac{1}{2}\sigma^2(s)C^2(s,T)\right]ds$$

Therefore:

$$A(t,T) = \int_t^T\left[a(s)C(s,T) - \frac{1}{2}\sigma^2(s)C^2(s,T)\right]ds \quad \blacksquare$$

$$A'(t, T) = \frac{2a \varphi'(t)}{\sigma^2 \varphi(t)}.$$

### (i) Expressing $C$ and $C'$ in terms of $\varphi$

Differentiating $\varphi(t) = \exp\left\{\frac{1}{2}\sigma^2\int_t^T C(u,T)du\right\}$:

$$\varphi'(t) = -\frac{1}{2}\sigma^2 C(t,T)\varphi(t)$$

Therefore:

$$C(t,T) = -\frac{2\varphi'(t)}{\sigma^2\varphi(t)} \quad \blacksquare$$

Differentiating again:

$$C'(t,T) = -\frac{2}{\sigma^2}\cdot\frac{\varphi''\varphi - (\varphi')^2}{\varphi^2} = -\frac{2\varphi''(t)}{\sigma^2\varphi(t)} + \frac{1}{2}\sigma^2 C^2(t,T) \quad \blacksquare$$

### (ii) ODE for $\varphi$

Substituting into (6.5.14): $C'(t,T) = bC(t,T) - 1 + \frac{1}{2}\sigma^2 C^2(t,T)$:

$$-\frac{2\varphi''}{\sigma^2\varphi} + \frac{1}{2}\sigma^2 C^2 = -\frac{2b\varphi'}{\sigma^2\varphi} - 1 + \frac{1}{2}\sigma^2 C^2$$

The $C^2$ terms cancel. Multiplying by $-\frac{\sigma^2\varphi}{2}$:

$$\varphi''(t) - b\varphi'(t) - \frac{1}{2}\sigma^2\varphi(t) = 0 \quad \blacksquare$$

### (iii) General solution for $\varphi(t)$

The characteristic equation $\lambda^2 - b\lambda - \frac{1}{2}\sigma^2 = 0$ has roots:

$$\lambda = \frac{1}{2}b \pm \gamma, \quad \gamma = \frac{1}{2}\sqrt{b^2 + 2\sigma^2}$$

Writing the general solution in terms of $T-t$ by setting $a_1 = \frac{c_1}{\frac{1}{2}b+\gamma}e^{-(\frac{1}{2}b+\gamma)T}$ and $a_2 = -\frac{c_2}{\frac{1}{2}b-\gamma}e^{-(\frac{1}{2}b-\gamma)T}$:

$$\varphi(t) = \frac{c_1}{\frac{1}{2}b+\gamma}e^{-(\frac{1}{2}b+\gamma)(T-t)} - \frac{c_2}{\frac{1}{2}b-\gamma}e^{-(\frac{1}{2}b-\gamma)(T-t)} \quad \blacksquare$$

### (iv) $c_1 = c_2$

Differentiating:

$$\varphi'(t) = c_1 e^{-(\frac{1}{2}b+\gamma)(T-t)} - c_2 e^{-(\frac{1}{2}b-\gamma)(T-t)} \quad \blacksquare$$

Since $C(T,T) = 0$, we have $\varphi'(T) = 0$:

$$\varphi'(T) = c_1 - c_2 = 0 \implies c_1 = c_2 \quad \blacksquare$$

### (v) $\varphi(t)$ in terms of $\sinh$ and $\cosh$, and $C(t,T)$

With $c_1 = c_2$, using $\frac{1}{4}b^2 - \gamma^2 = -\frac{1}{2}\sigma^2$:

$$\varphi(t) = c_1 e^{-\frac{1}{2}b(T-t)}\left[\frac{(\frac{1}{2}b-\gamma)e^{-\gamma(T-t)} - (\frac{1}{2}b+\gamma)e^{\gamma(T-t)}}{\frac{1}{4}b^2-\gamma^2}\right] \quad \blacksquare$$

Using $\sinh z = \frac{e^z-e^{-z}}{2}$ and $\cosh z = \frac{e^z+e^{-z}}{2}$:

$$\varphi(t) = \frac{2c_1}{\sigma^2}e^{-\frac{1}{2}b(T-t)}\left[b\sinh(\gamma(T-t)) + 2\gamma\cosh(\gamma(T-t))\right] \quad \blacksquare$$

$$\varphi'(t) = -2c_1 e^{-\frac{1}{2}b(T-t)}\sinh(\gamma(T-t)) \quad \blacksquare$$

Therefore:

$$C(t,T) = -\frac{2\varphi'(t)}{\sigma^2\varphi(t)} = \frac{4c_1 e^{-\frac{1}{2}b(T-t)}\sinh(\gamma(T-t))}{\sigma^2\cdot\frac{2c_1}{\sigma^2}e^{-\frac{1}{2}b(T-t)}[b\sinh(\gamma(T-t))+2\gamma\cosh(\gamma(T-t))]}$$

$$= \frac{2\sinh(\gamma(T-t))}{b\sinh(\gamma(T-t))+2\gamma\cosh(\gamma(T-t))} \quad \blacksquare$$

### (vi) Deriving $A(t,T)$

From (6.5.15) and (6.9.8): $A'(s,T) = \frac{2a\varphi'(s)}{\sigma^2\varphi(s)}$

Integrating from $s = t$ to $s = T$ and using $A(T,T) = 0$:

### (i) Deriving the PDE (6.9.26) from the martingale property

**Martingale property:** From (6.9.25):

$$e^{-rt}c(t,S(t),V(t)) = \tilde{\mathbb{E}}\left[e^{-rT}(S(T)-K)^+\Big|\mathcal{F}(t)\right]$$

For $s \leq t$, by the tower property:

$$\tilde{\mathbb{E}}\left[e^{-rt}c(t,S(t),V(t))\Big|\mathcal{F}(s)\right] = \tilde{\mathbb{E}}\left[e^{-rT}(S(T)-K)^+\Big|\mathcal{F}(s)\right] = e^{-rs}c(s,S(s),V(s))$$

so $e^{-rt}c(t,S(t),V(t))$ is a $\tilde{\mathbb{P}}$-martingale.

**Itô's formula:** Applying the 2D Itô-Doeblin formula to $e^{-rt}c(t,S(t),V(t))$, using $dS = rS\,dt + \sqrt{V}S\,d\tilde{W}_1$, $dV = (a-bV)dt + \sigma\sqrt{V}d\tilde{W}_2$, and $d\tilde{W}_1\,d\tilde{W}_2 = \rho\,dt$:

$$d(e^{-rt}c) = e^{-rt}\Big[-rc + c_t + rSc_s + (a-bV)c_v + \frac{1}{2}S^2Vc_{ss} + \rho\sigma SVc_{sv} + \frac{1}{2}\sigma^2Vc_{vv}\Big]dt + (\text{Itô integral})$$

Since the left side is a martingale, the $dt$ term must vanish:

$$c_t + rsc_s + (a-bv)c_v + \frac{1}{2}s^2vc_{ss} + \rho\sigma sv c_{sv} + \frac{1}{2}\sigma^2 v c_{vv} = rc \quad \blacksquare$$

### (ii) $c(t,s,v) = sf(t,\log s,v) - e^{-r(T-t)}Kg(t,\log s,v)$ satisfies (6.9.26)

Let $x = \log s$, so $x_s = \frac{1}{s}$. Computing the partial derivatives of $c$:

$$c_t = sf_t - re^{-r(T-t)}Kg - e^{-r(T-t)}Kg_t$$

$$c_s = f + sf_x\cdot\frac{1}{s} - e^{-r(T-t)}Kg_x\cdot\frac{1}{s} = f + f_x - e^{-r(T-t)}\frac{K}{s}g_x$$

$$c_v = sf_v - e^{-r(T-t)}Kg_v$$

$$c_{ss} = \frac{1}{s}\left(f_x + f_{xx}\right) + e^{-r(T-t)}\frac{K}{s^2}(g_x - g_{xx})$$

$$c_{sv} = f_v + f_{xv} - e^{-r(T-t)}\frac{K}{s}g_{xv}$$

$$c_{vv} = sf_{vv} - e^{-r(T-t)}Kg_{vv}$$

Substituting into the left side of (6.9.26) and grouping terms by $f$ and $g$:

**$f$-terms** (coefficient $s$): using (6.9.32),

$$s\left[f_t + \left(r+\frac{1}{2}v\right)f_x + (a-bv+\rho\sigma v)f_v + \frac{1}{2}vf_{xx} + \rho\sigma v f_{xv} + \frac{1}{2}\sigma^2 v f_{vv}\right] = 0$$

**$g$-terms** (coefficient $-e^{-r(T-t)}K$): using (6.9.33),

$$-e^{-r(T-t)}K\left[g_t + \left(r-\frac{1}{2}v\right)g_x + (a-bv)g_v + \frac{1}{2}vg_{xx} + \rho\sigma v g_{xv} + \frac{1}{2}\sigma^2 v g_{vv}\right] = 0$$

The remaining terms collapse to exactly $r(sf - e^{-r(T-t)}Kg) = rc$. Therefore $c(t,s,v)$ satisfies (6.9.26). $\blacksquare$

### (iii) $f(t,x,v)$ satisfies (6.9.32)

By definition $f(t,x,v) = \mathbb{E}^{t,x,v}\mathbb{I}_{\{X(T)\geq\log K\}}$. By the Markov property and tower property, $f(t,X(t),V(t))$ is a $\mathbb{P}$-martingale.

Applying Itô's formula to $f(t,X(t),V(t))$ with $dX = (r+\frac{1}{2}V)dt + \sqrt{V}dW_1$, $dV = (a-bV+\rho\sigma V)dt + \sigma\sqrt{V}dW_2$, and $dW_1\,dW_2 = \rho\,dt$, and setting the $dt$ term to zero:

$$f_t + \left(r+\frac{1}{2}v\right)f_x + (a-bv+\rho\sigma v)f_v + \frac{1}{2}vf_{xx} + \rho\sigma v f_{xv} + \frac{1}{2}\sigma^2 v f_{vv} = 0 \quad \blacksquare$$

**Boundary condition:** At $t=T$, directly from the definition:

$$f(T,x,v) = \mathbb{I}_{\{x\geq\log K\}} \quad \blacksquare$$

### (iv) $g(t,x,v)$ satisfies (6.9.33)

By the same argument as (iii), $g(t,X(t),V(t))$ is a $\mathbb{P}$-martingale where $X(t)$ now follows $dX = (r-\frac{1}{2}V)dt+\sqrt{V}dW_1$ and $V(t)$ follows $dV=(a-bV)dt+\sigma\sqrt{V}dW_2$. Applying Itô's formula and setting the $dt$ term to zero:

$$g_t + \left(r-\frac{1}{2}v\right)g_x + (a-bv)g_v + \frac{1}{2}vg_{xx} + \rho\sigma v g_{xv} + \frac{1}{2}\sigma^2 v g_{vv} = 0 \quad \blacksquare$$

**Boundary condition:**

$$g(T,x,v) = \mathbb{I}_{\{x\geq\log K\}} \quad \blacksquare$$

### (v) $c(t,s,v)$ satisfies the boundary condition (6.9.27)

Setting $t = T$ in (6.9.34) and using the boundary conditions from (iii) and (iv):

$$c(T,s,v) = sf(T,\log s,v) - K g(T,\log s,v) = s\,\mathbb{I}_{\{\log s\geq\log K\}} - K\,\mathbb{I}_{\{\log s\geq\log K\}}$$

$$= (s-K)\mathbb{I}_{\{s\geq K\}} = (s-K)^+ \quad \blacksquare$$

# Exercise 6.8 (Kolmogorov backward equation).
Consider the stochastic differential equation
$$dX(u) = \beta(u,X(u))du + \gamma(u, X(u))dW(u).$$
We assume that, just as with a geometric Brownian motion, if we begin a process at an arbitrary initial positive value $X(t) = x$ at an arbitrary initial time $t$ and evolve it forward using this equation, its value at each time $T > t$ could be any positive number but cannot be less that or equal to zero. For $0 \leq t < T,$ let $p(t, T, x, y)$ be the transition density for the solution to this equation (i.e., if we solve the eqation with the initial condition $X(t) = x,$ then the random variable $X(T)$ has density $p(t, T, x, y)$ in the $y$ variable). We are assuming that $p(t, T; x, y)$ satisfies the Kolmogorov backward equation
$$-p_t(t, T, x, y) = \beta(t, x)p_x(t, T, x, y) + \frac{1}{2}\gamma^2(t, x) p_{xx}(t, T, x, y).$$
(hint: We know from the Feynman-Kac Theorem, Theorem 6.4.1, that, for any function $h(y)$, the function
$$g(t,x) = \mathbb{E}^{t,x} h(X(T)) = \int_{0}^{\infty} h(y) p(t, T, x, y)dy \tag{6.9.44}$$
satisfies the partial differential equation
$$g_t(t, x) + \beta(t, x)g_x(t, x) + \frac{1}{2}\gamma^2(t, x) g_{xx}(t, x) = 0. \tag{6.9.45}$$
Use (6.9.44) to compute $g_t, g_x,$ and $g_{xx}$, and then argue that the only way (6.9.45) can hold regardless of the choice of the function $h(y)$ is for $p(t, T, x, y)$ to satisfy the Kolmogorov backward equation.)

**Proof:**

From (6.9.44), differentiating under the integral sign with respect to $t$ and $x$:

$$g_t(t,x) = \int_0^\infty h(y)\,p_t(t,T,x,y)\,dy$$

$$g_x(t,x) = \int_0^\infty h(y)\,p_x(t,T,x,y)\,dy$$

$$g_{xx}(t,x) = \int_0^\infty h(y)\,p_{xx}(t,T,x,y)\,dy$$

Substituting into (6.9.45):

$$\int_0^\infty h(y)\left[p_t(t,T,x,y) + \beta(t,x)p_x(t,T,x,y) + \frac{1}{2}\gamma^2(t,x)p_{xx}(t,T,x,y)\right]dy = 0$$

This must hold for every function $h(y)$. The only way an integral against an arbitrary test function $h(y)$ can vanish identically is if the bracketed term is itself identically zero (otherwise, choosing $h$ to be concentrated where the bracketed expression is nonzero and of one sign would give a nonzero integral). Therefore:

$$p_t(t,T,x,y) + \beta(t,x)p_x(t,T,x,y) + \frac{1}{2}\gamma^2(t,x)p_{xx}(t,T,x,y) = 0$$

i.e.,

$$-p_t(t,T,x,y) = \beta(t,x)p_x(t,T,x,y) + \frac{1}{2}\gamma^2(t,x)p_{xx}(t,T,x,y) \quad \blacksquare$$

**Remark:** This is the *backward* equation because the differentiation is with respect to the initial variables $(t,x)$, holding the terminal variables $(T,y)$ fixed — in contrast to the Kolmogorov *forward* (Fokker-Planck) equation, which differentiates with respect to $(T,y)$.

# Exercise 6.9 (Kolmogorov forward equation). 
(Also called the Fokker-Planck equation). We begin with the same stochastic differential equation 
$$dX(u) = \beta(u.X(u))du + \gamma(u, X(u))dW(u), \tag{6.9.46}$$
as in Exercise 6.8, use the same notation$p(t, T, x. y)$ for the transition density, and again assume that $p(t, T, x, y) = 0$ for $0 \leq t < T$ and $y \leq 0.$ In this problem, we show that $p(t, T, x, y)$ satisfies the Kolmogorov forward equation
$$\frac{\partial}{\partial T} p(t, T, x, y) = - \frac{\partial}{\partial y} (\beta(t, y)p(t, T, x, y)) + \frac{1}{2} \frac{\partial^2}{\partial y^2} (\gamma^2(T, y)p(t, T, x, y)).$$
In contrast to the Kolmogorov backward equation, in which $T$ and $y$ were held constant and the variables were $t$ and $x$, here $t$ and $x$ are held constant and the variables are $y$ and $T$. The variables $t$ and $x$ are somtimes called the backward variables, and $T$ and $y$ are called the forward variables.

(i) Let $b$ be a positive constant and let $h_b(y)$ be a function with continuous first and second derivatives such that $h_b(x) = 0$ for all $x \leq 0, h'_b(x) = 0$ for all $x \geq b$, and $h_b(b) = h'_b(b) = 0.$ Let $X(u)$ be the solution to the stochastic differential equation with initial condition $X(t) = x \in (0, b),$ and Itô's formula to compute $dh_b(X(u)).$

(ii) Let $0 \leq t < T$ be given, and integrate the equation you obtained in (i) from $t$ to $T$. Take expectations and use the fact that $X(u)$ has density $p(t, u, x, y)$ in the $y$-variable to obtain
$$\int_{0}^{b} h_b(y)p(t, T, x, y)dy = h_b(x) + \int_{t}^{T} \int_{0}^{b} \beta(u, y) p(t, u, x, y)h'_b(y)dy du + \frac{1}{2} \int_{t}^{T} \int_{0}^{b} \gamma^2(u, y)p(t, u, x, y)h''_b(y)dydu. \tag{6.9.48}$$
(iii) Integrate the integrals $\int_{0}^{b} \cdots dy$ on the right-hand side of (6.9.48) by parts to obtain 
$$\int_{0}^{b}h_b(y)p(t, T, x, y)dy $$
$$= h_b(x) - \int_{t}^{T} \int_{0}^{b} \frac{\partial}{\partial y} [\beta(u, y) p(t, u, x, y)] h_b(y) dy du + \frac{1}{2} \int_{t}^{T} \int_{0}^{b} \frac{\partial^2}{\partial y^2} [\gamma^2(u, y)p(t, u, x, y)] h_b(y) dy du. \tag{6.9.49}$$
(iv) Differentiate (6.9.49) with respect to T to obtain
$$\int_{0}^{b} h_b(y) \left[\frac{\partial}{\partial T} p(t, T, x, y) + \frac{\partial}{\partial y} (\beta(T, y)p(t, T, x, y)) - \frac{1}{2} \frac{\partial ^2}{\partial y^2} ( \gamma^2(T, y) p(t, T, x, y)) \right] dy = 0. \tag{6.9.50}$$
(v) Use (6.9.50) to show that there cannot be numbers $0 < y_1 < y_2$ such that 
$$\frac{\partial}{\partial T} p(t, T, x, y) + \frac{\partial}{\partial y} (\beta(T, y) p(t, T, x, y)) - \frac{1}{2}\frac{\partial^2}{\partial y^2} (\gamma^2(T, y)p(t, T, x, y)) > 0 \text{ for all } y \in (y_1, y_2).$$
Similarly, there cannot be numbers $0 < y_1 < y_2$ such that
$$\frac{\partial}{\partial T} p(t, T, x, y) + \frac{\partial}{\partial y} (\beta(T, y)p(t, T, x, y)) - \frac{1}{2} \frac{\partial^2}{\partial y^2} (\gamma^2(T, y)p(t, T, x, y)) < 0 \text{ for all } y \in [y_1, y_2].$$
This is as much as you need to do for this problem. It is now obvious that if
$$\frac{\partial}{\partial T}[(t, T, x, y) + \frac{\partial}{\partial y} (\beta(T, y)p(t, T, x, y)) - \frac{1}{2} \frac{\partial}{\partial y^2} (\gamma^2(T, y)p(t, T, x, y))$$
is a continuous function of $y$, then this expression must be zero for every $y > 0,$ and hence $p(t, T, x, y)$ satisfies the Kolmogorov forward equation stated at the beggining of this problem.

### (i) Itô's formula applied to $h_b(X(u))$

By Itô's formula:

$$dh_b(X(u)) = h_b'(X(u))dX(u) + \frac{1}{2}h_b''(X(u))(dX(u))^2$$

$$= \left[\beta(u,X(u))h_b'(X(u)) + \frac{1}{2}\gamma^2(u,X(u))h_b''(X(u))\right]du + \gamma(u,X(u))h_b'(X(u))dW(u)$$

Note that no indicator function is needed: since $h_b'(y) = h_b''(y) = 0$ for $y \leq 0$ and $y \geq b$, the expression automatically vanishes outside $(0,b)$ wherever $X(u)$ exits that range. $\blacksquare$

### (ii) Integrating and taking expectations

Integrating the equation from (i) from $t$ to $T$:

$$h_b(X(T)) = h_b(x) + \int_t^T\left[\beta(u,X(u))h_b'(X(u)) + \frac{1}{2}\gamma^2(u,X(u))h_b''(X(u))\right]du + \int_t^T \gamma(u,X(u))h_b'(X(u))dW(u)$$

Taking expectations (the Itô integral vanishes) and using the fact that $X(u)$ has density $p(t,u,x,y)$ in the $y$-variable, with $h_b(y) = 0$ outside $(0,b)$:

$$\mathbb{E}[h_b(X(T))] = \int_0^b h_b(y)p(t,T,x,y)dy$$

$$\mathbb{E}\left[\int_t^T \beta(u,X(u))h_b'(X(u))du\right] = \int_t^T\int_0^b \beta(u,y)p(t,u,x,y)h_b'(y)\,dy\,du$$

and similarly for the second-derivative term. This gives:

$$\int_0^b h_b(y)p(t,T,x,y)dy = h_b(x) + \int_t^T\int_0^b \beta(u,y)p(t,u,x,y)h_b'(y)\,dy\,du$$

$$+ \frac{1}{2}\int_t^T\int_0^b \gamma^2(u,y)p(t,u,x,y)h_b''(y)\,dy\,du \quad \blacksquare$$

### (iii) Integration by parts

**First integral** ($\int_0^b \beta p\, h_b'\,dy$): integrating by parts once, with boundary terms vanishing since $h_b(0)=h_b(b)=0$:

$$\int_0^b \beta(u,y)p(t,u,x,y)h_b'(y)\,dy = -\int_0^b \frac{\partial}{\partial y}\left[\beta(u,y)p(t,u,x,y)\right]h_b(y)\,dy$$

**Second integral** ($\int_0^b \gamma^2 p\, h_b''\,dy$): integrating by parts twice, with boundary terms vanishing since $h_b(0) = h_b'(0) = 0$ and $h_b(b) = h_b'(b) = 0$:

$$\int_0^b \gamma^2(u,y)p(t,u,x,y)h_b''(y)\,dy = \int_0^b \frac{\partial^2}{\partial y^2}\left[\gamma^2(u,y)p(t,u,x,y)\right]h_b(y)\,dy$$

Substituting into (6.9.48):

$$\int_0^b h_b(y)p(t,T,x,y)dy = h_b(x) - \int_t^T\int_0^b \frac{\partial}{\partial y}\left[\beta(u,y)p(t,u,x,y)\right]h_b(y)\,dy\,du$$

$$+ \frac{1}{2}\int_t^T\int_0^b \frac{\partial^2}{\partial y^2}\left[\gamma^2(u,y)p(t,u,x,y)\right]h_b(y)\,dy\,du \quad \blacksquare$$

### (iv) Differentiating with respect to $T$

By the Fundamental Theorem of Calculus, differentiating (6.9.49) with respect to $T$:

$$\int_0^b h_b(y)\frac{\partial}{\partial T}p(t,T,x,y)\,dy = -\int_0^b \frac{\partial}{\partial y}\left[\beta(T,y)p(t,T,x,y)\right]h_b(y)\,dy$$

$$+ \frac{1}{2}\int_0^b \frac{\partial^2}{\partial y^2}\left[\gamma^2(T,y)p(t,T,x,y)\right]h_b(y)\,dy$$

Rearranging:

$$\int_0^b h_b(y)\left[\frac{\partial}{\partial T}p(t,T,x,y) + \frac{\partial}{\partial y}\left(\beta(T,y)p(t,T,x,y)\right) - \frac{1}{2}\frac{\partial^2}{\partial y^2}\left(\gamma^2(T,y)p(t,T,x,y)\right)\right]dy = 0 \quad \blacksquare$$

### (v) The bracketed expression must be zero

Suppose, for contradiction, that there exist $0 < y_1 < y_2$ such that the bracketed expression in (6.9.50) is strictly positive on $(y_1,y_2)$. Choose $b > y_2$ and a function $h_b$ satisfying the stated properties with $h_b(y) > 0$ on $(y_1,y_2)$ and $h_b(y) = 0$ outside $(y_1,y_2)$ (this is possible since $h_b$ can be chosen as an arbitrary smooth bump function on $(0,b)$). Then the integral in (6.9.50) would be strictly positive, contradicting (6.9.50) $= 0$.

The same argument with a sign flip rules out the bracketed expression being strictly negative on any subinterval.

Since $h_b(y)$ can be chosen as an arbitrary smooth function vanishing outside $(0,b)$ (with the stated boundary conditions), and (6.9.50) must hold for every such choice, the bracketed expression cannot be positive on any subinterval nor negative on any subinterval. If the bracketed expression is continuous in $y$, it must therefore be identically zero for all $y > 0$:

$$\frac{\partial}{\partial T}p(t,T,x,y) + \frac{\partial}{\partial y}\left(\beta(T,y)p(t,T,x,y)\right) - \frac{1}{2}\frac{\partial^2}{\partial y^2}\left(\gamma^2(T,y)p(t,T,x,y)\right) = 0$$

i.e.,

$$\frac{\partial}{\partial T}p(t,T,x,y) = -\frac{\partial}{\partial y}\left(\beta(T,y)p(t,T,x,y)\right) + \frac{1}{2}\frac{\partial^2}{\partial y^2}\left(\gamma^2(T,y)p(t,T,x,y)\right) \quad \blacksquare$$

This is the Kolmogorov forward equation (Fokker-Planck equation).

$$= -\frac{1}{2}\left[\sigma^2(T,y)y^2\tilde{p}(0,T,x,y)\right]_K^\infty$$

By the assumption that this vanishes as $y\to\infty$:

$$= \frac{1}{2}\sigma^2(T,K)K^2\tilde{p}(0,T,x,K) \quad \blacksquare$$

### (iii) Deriving Dupire's equation

Substituting the Kolmogorov forward equation (6.9.51) into (6.9.53):

$$c_T = -rc + e^{-rT}\int_K^\infty(y-K)\left[-\frac{\partial}{\partial y}(ry\tilde{p}) + \frac{1}{2}\frac{\partial^2}{\partial y^2}(\sigma^2y^2\tilde{p})\right]dy$$

Applying (6.9.54) and (6.9.56):

$$c_T = -rc + e^{-rT}\int_K^\infty ry\tilde{p}(0,T,x,y)\,dy + \frac{1}{2}e^{-rT}\sigma^2(T,K)K^2\tilde{p}(0,T,x,K)$$

Writing $y = (y-K) + K$ in the remaining integral and using (6.9.52):

$$e^{-rT}\int_K^\infty ry\tilde{p}\,dy = r\cdot e^{-rT}\int_K^\infty (y-K)\tilde{p}\,dy + rK\cdot e^{-rT}\int_K^\infty \tilde{p}\,dy = rc + e^{-rT}rK\int_K^\infty\tilde{p}\,dy$$

The $-rc$ and $rc$ terms cancel, giving:

$$c_T(0,T,x,K) = e^{-rT}rK\int_K^\infty\tilde{p}(0,T,x,y)\,dy + \frac{1}{2}e^{-rT}\sigma^2(T,K)K^2\tilde{p}(0,T,x,K)$$

By Exercise 5.9, $c_K = -e^{-rT}\int_K^\infty\tilde{p}\,dy$ and $c_{KK} = e^{-rT}\tilde{p}(0,T,x,K)$, so:

$$c_T(0,T,x,K) = -rKc_K(0,T,x,K) + \frac{1}{2}\sigma^2(T,K)K^2c_{KK}(0,T,x,K) \quad \blacksquare$$

### Remark: Dupire's formula

When $c_{KK}(0,T,x,K) \neq 0$, this equation can be solved for the local volatility:

$$\sigma^2(T,K) = \frac{c_T(0,T,x,K) + rKc_K(0,T,x,K)}{\frac{1}{2}K^2c_{KK}(0,T,x,K)}$$

This is **Dupire's formula**, which allows the entire local volatility surface $\sigma(T,K)$ to be recovered from the market prices of European calls at all strikes $K$ and maturities $T$. In practice, $c_T$, $c_K$, and $c_{KK}$ are estimated by interpolating market option prices and differentiating numerically.
