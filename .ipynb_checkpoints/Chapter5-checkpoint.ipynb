# Exercise 5.1.
Consider the discounted stock price $D(t)S(t)$ of (5.2.19). In this problem, we derive the formula (5.2.20) for $d(D(t)S(t))$ by two methods.

(i) Define $f(x) = S(0)e^x$ and set
$$X(t) = \int_{0}^{t} \sigma(s) dW(s) + \int_{0}^{t} ( \alpha(s) - R(s) - \frac{1}{2} \sigma^2 (s))ds$$
so that $D(t)S(t) = f(X(t)).$ Use the itô-Doeblin formula to compute $df(X(t))$.

(ii) According to Itô's product rule,
$$d(D(t)S(t)) = S(t)dD(t) + D(t)dS(t) + dD(t)dS(t).$$
Use (5.2.15) and (5.2.18) to work out the right-hand side of this equation.
$$dS(t) = \alpha(t)S(t)dt + \sigma(t)S(t)dW(t), 0 \leq t \leq T. \tag{5.2.15}$$
$$dD(t) = -R(t)D(t)dt. \tag{5.2.18}$$
$$D(t)S(t) = S(0)\exp \left\{ \int_{0}^{t} \sigma(s)dW(s) + \int_{0}^{t} \left(\alpha(s) - R(s) - \frac{1}{2} \sigma^2(s)\right) ds \right\}, \tag{5.2.19}$$
$$d(D(t)S(t)) = (\alpha(t) - R(t))D(t)S(t)dt + \sigma(t)D(t)S(t)dW(t) = \sigma(t)D(t)S(t)[\mathcal{\Theta}(t)dt + dW(t)], \tag{5.2.20}$$

### (i) Itô-Doeblin formula

Define $f(x) = S(0)e^x$, so $f'(x) = f''(x) = S(0)e^x$.

$$dX(t) = \sigma(t)dW(t) + \left(\alpha(t) - R(t) - \frac{1}{2}\sigma^2(t)\right)dt, \quad dX\,dX = \sigma^2(t)dt$$

By the Itô-Doeblin formula:

$$df(X(t)) = f'(X(t))dX(t) + \frac{1}{2}f''(X(t))dX\,dX$$

$$= D(t)S(t)\left[\sigma(t)dW(t) + \left(\alpha(t) - R(t) - \frac{1}{2}\sigma^2(t)\right)dt + \frac{1}{2}\sigma^2(t)dt\right]$$

$$= D(t)S(t)\left[(\alpha(t) - R(t))dt + \sigma(t)dW(t)\right] = \sigma(t)D(t)S(t)\left[\mathcal{\Theta}(t)dt + dW(t)\right] \quad \blacksquare$$

### (ii) Itô's product rule

Using (5.2.15) and (5.2.18):

**Term 1:**
$$S(t)dD(t) = -R(t)D(t)S(t)dt$$

**Term 2:**
$$D(t)dS(t) = \alpha(t)D(t)S(t)dt + \sigma(t)D(t)S(t)dW(t)$$

**Term 3:**
$$dD(t)dS(t) = (-R(t)D(t)dt)(\alpha(t)S(t)dt + \sigma(t)S(t)dW(t)) = 0$$

since $dt\cdot dt = 0$ and $dt\cdot dW = 0$.

Summing all terms:

$$d(D(t)S(t)) = (\alpha(t) - R(t))D(t)S(t)dt + \sigma(t)D(t)S(t)dW(t) = \sigma(t)D(t)S(t)[\mathcal{\Theta}(t)dt + dW(t)] \quad \blacksquare$$

# Exercise 5.2 (State price density process).
Show that the risk-neutral pricing formula (5.2.30) may be rewritten as
$$D(t)Z(t)V(t) = \mathbb{E}[D(T)Z(T)V(T)|\mathcal{F}(t)]. \tag{5.9.1}$$
Here $Z(t)$ is the Radon-Nikodym derivative process (5.2.11) when the market price of risk process $\mathcal{\Theta}(t)$ is given by (5.2.21) and the conditional expectation on the right-hand side of (5.9.1) is taken under the actual probability measure $\mathbb{P},$ not the risk-neutral measure $\tilde{\mathbb{P}}.$ In particular, if for some $A \in \mathcal{F}(T)$ a derivative security pays off $\mathbb{I}_A$ (i.e., pays $1$ if $A$ occures and $0$ if $A$ does not occur), then the value of this derivative security at time zero is $\mathbb{E}[D(T)Z(T)\mathbb{I}_A].$ The process $D(t)Z(t)$ appearing in (5.9.1) is called the state price density process.
$$Z(t) = \exp \left\{\int_{0}^{t} \mathcal{\Theta}(u)dW(u) - \frac{1}{2} \int_{0}^{t} \mathcal{\Theta}^2(u)du\right\} \tag{5.2.11}$$
$$ \text{market price of risk : } \mathcal{\Theta}(t) = \frac{\alpha(t) - R(t)}{\sigma(t)}. \tag{5.2.21}$$
$$D(t)V(t) = \tilde{\mathbb{E}}[D(T)V(T)|\mathcal{F}(t)], 0 \leq t \leq T. \tag{5.2.30}$$

**Proof:**

Since $Z(t)$ is the Radon-Nikodym derivative process, by Lemma 5.2.2, for any $\mathcal{F}(T)$-measurable random variable $X$:

$$\tilde{\mathbb{E}}[X|\mathcal{F}(t)] = \frac{1}{Z(t)}\mathbb{E}[Z(T)X|\mathcal{F}(t)]$$

Applying this to $X = D(T)V(T)$, which is $\mathcal{F}(T)$-measurable:

$$\tilde{\mathbb{E}}[D(T)V(T)|\mathcal{F}(t)] = \frac{1}{Z(t)}\mathbb{E}[D(T)V(T)Z(T)|\mathcal{F}(t)]$$

From the risk-neutral pricing formula (5.2.30):

$$D(t)V(t) = \tilde{\mathbb{E}}[D(T)V(T)|\mathcal{F}(t)] = \frac{1}{Z(t)}\mathbb{E}[D(T)V(T)Z(T)|\mathcal{F}(t)]$$

Multiplying both sides by $Z(t)$:

$$D(t)Z(t)V(t) = \mathbb{E}[D(T)Z(T)V(T)|\mathcal{F}(t)] \quad \blacksquare$$

**Remark:** In particular, for a derivative security paying $\mathbb{I}_A$ at time $T$ (where $A \in \mathcal{F}(T)$), its value at time zero is:

$$V(0) = \mathbb{E}[D(T)Z(T)\mathbb{I}_A]$$

This justifies calling $D(t)Z(t)$ the **state price density process**.

# Exercise 5.3.
According to the Black-Scholes-Merton formula, the value at time zero of a European call on a stock whose initial price is $S(0) = x$ is given by 
$$c(0, x) = xN(d_+(T, x)) - K e^{-rT} N(d_-(T, x)), $$
where
$$d_+(T, x) = \frac{1}{\sigma\sqrt{T}} \left[\log \frac{x}{K} + \left(r + \frac{1}{2} \sigma^2 \right) T\right],$$
$$d_-(T, x) = d_+(T, x) - \sigma \sqrt{T}.$$
The stock is modeled as a geometric Brownian motion with constant volatility $\sigma > 0$, the interest rate is constant $r$, the call strike is $K$, and the call expiration time is $T$. This formula is obtained by computing the discounted expected payoff of the call under the risk-neutral measure,
$$c(0, x) = \tilde{\mathbb{E}}[e^{-rT}(S(T) - K)^+] = \tilde{\mathbb{E}}\left[ e^{-rT} \left(x \exp \left\{\sigma \tilde{W}(T) + \left(r - \frac{1}{2}\sigma^2 \right) T \right\} - K\right)^+ \right], \tag{5.9.2}$$
where $\tilde{W}$ is a Brownian motion under the risk-neutral measure $\tilde{\mathbb{P}}$. In exercise 4.9(ii), the delta of this option is computed to be $c_x(0, x) = N(d_+(T, x)).$ This problem provides an alternate way to compute $c_x(0, x).$

(i) We begin with the observation that if $h(s) = (s - K)^+,$ then 
$$h'(s) = \begin{cases} 0 & \text{ if } s < K, \\ 1 & \text{ if } s > K. \end{cases}$$
if $s = K$, then $h'(s)$ is undefined, but that will not matter in what follows because $S(T)$ has zero probability of takeing the value $K$. Using this formula for $h'(s)$, differentiate inside the expected value in (5.9.2) to obtain a formula for $c_x(0, x).$

(ii) Show that the formula you obtained in (i) can be rewritten as 
$$c_x(0,x) = \hat{\mathbb{P}}(S(T) > K),$$
where $\tilde{\mathbb{P}}$ is a probability measure equivalent to $\tilde{\mathbb{P}}.$ Show that
$$\hat{W}(t) = \tilde{W}(t) - \sigma t$$
is a Brownian motion under $\tilde{\mathbb{P}}$.

(iii) Rewrite $S(T)$ in terms of $\hat{W}(T)$, and then show that
$$\hat{\mathbb{P}}\{S(T) > K \} = \hat{\mathbb{P}}\left\{- \frac{\hat{W}(T)}{\sqrt{T}} < d_+(T, x)\right\} = N(d_+(T, x)).$$


### (i) Differentiating inside the expectation

Using $h'(s) = \mathbb{I}_{\{s > K\}}$ and the chain rule:

$$c_x(0,x) = \tilde{\mathbb{E}}\left[e^{-rT}\mathbb{I}_{\{S(T)>K\}}\exp\left\{\sigma\tilde{W}(T) + \left(r-\frac{1}{2}\sigma^2\right)T\right\}\right]$$

Setting $Z = -\frac{\tilde{W}(T)}{\sqrt{T}} \sim \mathcal{N}(0,1)$ under $\tilde{\mathbb{P}}$, and noting $S(T) > K \iff Z < d_-$:

$$c_x(0,x) = \int_{-\infty}^{d_-} e^{-\sigma\sqrt{T}z - \frac{1}{2}\sigma^2 T} \phi(z)dz$$

Substituting $y = z + \sigma\sqrt{T}$ (so $z = y - \sigma\sqrt{T}$, lower limit $d_- + \sigma\sqrt{T} = d_+$... wait, $d_- \to -\infty$ maps to $d_+$):

$$= \int_{-\infty}^{d_+} \phi(y)dy = N(d_+(T,x)) \quad \blacksquare$$

### (ii) Measure change via Radon-Nikodym

From (i):

$$c_x(0,x) = \tilde{\mathbb{E}}\left[\mathbb{I}_{\{S(T)>K\}} \cdot e^{\sigma\tilde{W}(T) - \frac{1}{2}\sigma^2 T}\right]$$

Define the Radon-Nikodym derivative:

$$\frac{d\hat{\mathbb{P}}}{d\tilde{\mathbb{P}}} = e^{\sigma\tilde{W}(T) - \frac{1}{2}\sigma^2 T}$$

Then:

$$c_x(0,x) = \hat{\mathbb{E}}[\mathbb{I}_{\{S(T)>K\}}] = \hat{\mathbb{P}}\{S(T) > K\}$$

By Girsanov's theorem, $\hat{W}(t) = \tilde{W}(t) - \sigma t$ is a Brownian motion under $\hat{\mathbb{P}}$. $\blacksquare$

### (iii) Computing $\hat{\mathbb{P}}\{S(T) > K\}$

Rewriting $S(T)$ in terms of $\hat{W}(T) = \tilde{W}(T) - \sigma T$:

$$S(T) = x\exp\left\{\sigma\tilde{W}(T) + \left(r-\frac{1}{2}\sigma^2\right)T\right\} = x\exp\left\{\sigma(\hat{W}(T) + \sigma T) + \left(r-\frac{1}{2}\sigma^2\right)T\right\}$$

$$= x\exp\left\{\sigma\hat{W}(T) + \left(r+\frac{1}{2}\sigma^2\right)T\right\}$$

$S(T) > K$ is equivalent to:

$$\sigma\hat{W}(T) + \left(r+\frac{1}{2}\sigma^2\right)T > \log\frac{K}{x}$$

Setting $Y = -\frac{\hat{W}(T)}{\sqrt{T}} \sim \mathcal{N}(0,1)$ under $\hat{\mathbb{P}}$:

$$-\sigma\sqrt{T}Y + \left(r+\frac{1}{2}\sigma^2\right)T > \log\frac{K}{x}$$

$$Y < \frac{1}{\sigma\sqrt{T}}\left[\log\frac{x}{K} + \left(r+\frac{1}{2}\sigma^2\right)T\right] = d_+(T,x)$$

Therefore:

$$\hat{\mathbb{P}}\{S(T) > K\} = \hat{\mathbb{P}}\left\{-\frac{\hat{W}(T)}{\sqrt{T}} < d_+(T,x)\right\} = N(d_+(T,x)) \quad \blacksquare$$

# Exercise 5.4 (Black-Sholes-Merton formula for time-varying, non-random interest rate and volatility).
Consider a stock whose price differential is
$$dS(t) = r(t)S(t)dt + \sigma(t)d\tilde{W}(t), $$
where $r(t)$ and $\sigma(t)$ are nonrandom functions of $t$ and $\tilde{W}$ is a Brownian motion under the risk-neutral measure $\tilde{\mathbb{P}}.$ Let $T > 0$ be given, and consider a European call, whose value at time zero is
$$c(0, S(0)) = \mathbb{E}\left[\exp \left\{ - \int_{0}^{T} r(t)dt \right\} (S(T) - K) ^ + \right].$$
(i) Show that $S(T)$ is of the for $S(0)e^X$, where X is a normal random variable, and deterine the mean and varianve of $X$.

(ii) Let 
$$\text{BSM}(T,x; K, R \Sigma) = xN\left(\frac{1}{\Sigma \sqrt{T}} \left[ \log \frac{x}{K} + (R + \Sigma^2/2)T\right] \right) - e^{-RT} K N\left(\frac{1}{\Sigma \sqrt{T}} \left[\log \frac{x}{K} + (R - \Sigma^2/2)T\right] \right)$$
dennote the value at time zero of a European call expiring at time $T$ when the underlying stock has constant volatility $\Sigma$ and the interest rate $R$ is constant. Show that
$$c(0, S(0)) = \text{BSM}\left(S(0), T, \frac{1}{T} \int_{0}^{T} r(t)dt, \sqrt{\frac{1}{T} \int_{0}^{T} \sigma^2(t)dt} \right). $$

### (i) Distribution of $X$

Let $f(x) = \log x$, so $f'(x) = \frac{1}{x}$ and $f''(x) = -\frac{1}{x^2}$.

By Itô's lemma:

$$d\log S(t) = \left(r(t) - \frac{1}{2}\sigma^2(t)\right)dt + \sigma(t)d\tilde{W}(t)$$

Integrating from $0$ to $T$:

$$S(T) = S(0)e^X, \quad X = \int_0^T\left(r(t)-\frac{1}{2}\sigma^2(t)\right)dt + \int_0^T\sigma(t)d\tilde{W}(t)$$

Since $r(t)$ and $\sigma(t)$ are nonrandom, $X$ is normally distributed:

$$X \sim \mathcal{N}\left(\int_0^T\left(r(t)-\frac{1}{2}\sigma^2(t)\right)dt,\ \int_0^T\sigma^2(t)dt\right) \quad \blacksquare$$

### (ii) BSM formula with averaged parameters

Define:

$$RT = \int_0^T r(t)dt, \quad \Sigma^2 T = \int_0^T\sigma^2(t)dt$$

so $R = \frac{1}{T}\int_0^T r(t)dt$ and $\Sigma = \sqrt{\frac{1}{T}\int_0^T\sigma^2(t)dt}$.

Also define:

$$d_\pm = \frac{1}{\Sigma\sqrt{T}}\left[\log\frac{S(0)}{K} + \left(R \pm \frac{\Sigma^2}{2}\right)T\right]$$

Setting $Y = -\frac{X - \int_0^T(r(t)-\frac{1}{2}\sigma^2(t))dt}{\Sigma\sqrt{T}} \sim \mathcal{N}(0,1)$, $S(T) > K$ is equivalent to $Y < d_-$. Therefore:

$$c(0,S(0)) = \int_{-\infty}^{d_-} e^{-RT}(S(0)e^X - K)\phi(y)dy$$

**Second term:**

$$-e^{-RT}K\int_{-\infty}^{d_-}\phi(y)dy = -Ke^{-RT}N(d_-)$$

**First term:**

$$S(0)e^{-RT}\int_{-\infty}^{d_-}e^X\phi(y)dy$$

Since $X = \int_0^T(r(t)-\frac{1}{2}\sigma^2(t))dt - \Sigma\sqrt{T}Y = RT - \frac{1}{2}\Sigma^2 T - \Sigma\sqrt{T}Y$:

$$= S(0)e^{-RT}\cdot e^{RT - \frac{1}{2}\Sigma^2 T}\int_{-\infty}^{d_-}e^{-\Sigma\sqrt{T}y}\phi(y)dy$$

$$= S(0)e^{-\frac{1}{2}\Sigma^2 T}\int_{-\infty}^{d_-}\frac{1}{\sqrt{2\pi}}\exp\left\{-\Sigma\sqrt{T}y - \frac{y^2}{2}\right\}dy$$

$$= S(0)\int_{-\infty}^{d_-}\frac{1}{\sqrt{2\pi}}\exp\left\{-\frac{(y+\Sigma\sqrt{T})^2}{2}\right\}dy$$

Substituting $z = y + \Sigma\sqrt{T}$:

$$= S(0)\int_{-\infty}^{d_+}\phi(z)dz = S(0)N(d_+)$$

**Combining:**

$$c(0,S(0)) = S(0)N(d_+) - Ke^{-RT}N(d_-) = \text{BSM}\left(S(0), T, \frac{1}{T}\int_0^T r(t)dt, \sqrt{\frac{1}{T}\int_0^T\sigma^2(t)dt}\right) \quad \blacksquare$$

# Exercise 5.5.
Prove Corolarry 5.3.2 by the following steps.

(i) Compute the differential of $\frac{1}{Z(t)}$, where $Z(t)$ is given in Corollary 5.3.2.

(ii) Let $\tilde{M}(t), 0 \leq t \leq T,$ be a martingale under $\tilde{\mathbb{P}}.$ Show that $M(T) = Z(t) \tilde{M}(t)$ is a martingale under $\mathbb{P}.$

(iii) According to Theorem 5.3.1, there is an adapted process $\Gamma(u), 0 \leq u \leq T,$ such that
$$M(t) = M(0) + \int_{0}^{T} \Gamma(u) dW(u), 0 \leq t \leq T.$$
Write $\tilde{M}(t) = M(t) \cdot \frac{1}{Z(t)}$ and take its differential using ITô's product rule.

(iv) Show that the differential of $\tilde{M}(t)$ is the sum of an adapted process, which we call $\tilde{\Gamma}(t),$ times $d\tilde{W}(t),$ and zero times dt. Integrate to obtain(5.3.2).

## Exercise 5.5

### (i) Differential of $\frac{1}{Z(t)}$

Since $dZ(t) = -\Theta(t)Z(t)dW(t)$, by Itô's lemma with $f(z) = \frac{1}{z}$:

$$d\left(\frac{1}{Z(t)}\right) = -\frac{1}{Z^2(t)}dZ(t) + \frac{1}{Z^3(t)}(dZ)^2$$

$$= \frac{\Theta(t)}{Z(t)}dW(t) + \frac{\Theta^2(t)}{Z(t)}dt = \frac{\Theta(t)}{Z(t)}\left[dW(t) + \Theta(t)dt\right] = \frac{\Theta(t)}{Z(t)}d\tilde{W}(t) \quad \blacksquare$$

### (ii) $M(t) = Z(t)\tilde{M}(t)$ is a martingale under $\mathbb{P}$

For $s \leq t$, by Lemma 5.2.2:

$$\mathbb{E}[M(t)|\mathcal{F}(s)] = \mathbb{E}[Z(t)\tilde{M}(t)|\mathcal{F}(s)] = Z(s)\tilde{\mathbb{E}}[\tilde{M}(t)|\mathcal{F}(s)] = Z(s)\tilde{M}(s) = M(s) \quad \blacksquare$$

### (iii) Differential of $\tilde{M}(t) = M(t)\cdot\frac{1}{Z(t)}$

By Itô's product rule:

$$d\tilde{M}(t) = \frac{1}{Z(t)}dM(t) + M(t)d\frac{1}{Z(t)} + dM(t)d\frac{1}{Z(t)}$$

$$= \frac{\Gamma(t)}{Z(t)}dW(t) + \frac{M(t)\Theta(t)}{Z(t)}d\tilde{W}(t) + \frac{\Gamma(t)}{Z(t)}dW(t)\cdot\frac{\Theta(t)}{Z(t)}d\tilde{W}(t) \cdot Z(t)$$

Computing the cross term using $dW\cdot d\tilde{W} = dW\cdot(dW + \Theta\,dt) = dt$:

$$dM(t)\cdot d\frac{1}{Z(t)} = \frac{\Gamma(t)\Theta(t)}{Z(t)}dt$$

Therefore:

$$d\tilde{M}(t) = \frac{\Gamma(t)}{Z(t)}dW(t) + \frac{M(t)\Theta(t)}{Z(t)}d\tilde{W}(t) + \frac{\Gamma(t)\Theta(t)}{Z(t)}dt$$

### (iv) Cancellation of $dt$ terms

Substituting $dW(t) = d\tilde{W}(t) - \Theta(t)dt$:

$$\frac{\Gamma(t)}{Z(t)}dW(t) = \frac{\Gamma(t)}{Z(t)}d\tilde{W}(t) - \frac{\Gamma(t)\Theta(t)}{Z(t)}dt$$

Therefore:

$$d\tilde{M}(t) = \frac{\Gamma(t)}{Z(t)}d\tilde{W}(t) - \frac{\Gamma(t)\Theta(t)}{Z(t)}dt + \frac{M(t)\Theta(t)}{Z(t)}d\tilde{W}(t) + \frac{\Gamma(t)\Theta(t)}{Z(t)}dt$$

The $dt$ terms cancel, leaving:

$$d\tilde{M}(t) = \underbrace{\left(\frac{\Gamma(t) + M(t)\Theta(t)}{Z(t)}\right)}_{=:\tilde{\Gamma}(t)}d\tilde{W}(t)$$

Integrating:

$$\tilde{M}(t) = \tilde{M}(0) + \int_0^t \tilde{\Gamma}(u)d\tilde{W}(u) \quad \blacksquare$$

# Exercise 5.6. 
Use the two-dimmensional Levy Theorem, Theorem 4.6.5, to prove the two-dimmensional Girsanov Theorem(i.e., Theorem 5.4.1 widh $d = 2$).

**Proof:**

Define:

$$d\tilde{W}_i(t) = dW_i(t) + \Theta_i(t)dt, \quad i = 1, 2$$

By the two-dimensional Lévy Theorem (Theorem 4.6.5), it suffices to show that each $\tilde{W}_i(t)$ is a $\tilde{\mathbb{P}}$-martingale and $d\tilde{W}_i\,d\tilde{W}_j = \delta_{ij}dt$.

**Step 1: $\tilde{W}_i(t)$ is a $\tilde{\mathbb{P}}$-martingale**

Since $dZ(t) = -Z(t)\sum_{j=1}^2\Theta_j(t)dW_j(t)$, by Itô's product rule:

$$d(\tilde{W}_i Z) = \tilde{W}_i\,dZ + Z\,d\tilde{W}_i + d\tilde{W}_i\,dZ$$

Computing each term:

$$\tilde{W}_i\,dZ = -\tilde{W}_i Z\sum_{j=1}^2\Theta_j\,dW_j \quad \text{(Itô integral only)}$$

$$Z\,d\tilde{W}_i = Z\,dW_i + Z\Theta_i\,dt$$

$$d\tilde{W}_i\,dZ = (dW_i + \Theta_i\,dt)\left(-Z\sum_{j=1}^2\Theta_j\,dW_j\right) = -Z\Theta_i\,dt$$

The $dt$ terms cancel: $Z\Theta_i\,dt - Z\Theta_i\,dt = 0$. Therefore:

$$d(\tilde{W}_i Z) = \sum_{j=1}^2\Gamma_j(t)dW_j(t)$$

for some adapted processes $\Gamma_j$, so $\tilde{W}_i(t)Z(t)$ is a $\mathbb{P}$-martingale. By Lemma 5.2.2, $\tilde{W}_i(t)$ is a $\tilde{\mathbb{P}}$-martingale.

**Step 2: Quadratic covariation**

$$d\tilde{W}_i\,d\tilde{W}_j = (dW_i + \Theta_i\,dt)(dW_j + \Theta_j\,dt) = dW_i\,dW_j = \delta_{ij}dt$$

since $dW_i\,dt = 0$ and $(dt)^2 = 0$.

**Conclusion:**

By the two-dimensional Lévy Theorem, $(\tilde{W}_1(t), \tilde{W}_2(t))$ is a two-dimensional Brownian motion under $\tilde{\mathbb{P}}$. $\blacksquare$

# Exercise 5.7.
$$\mathbb{P}\{X(T) \geq 0\} = 1, \quad \mathbb{P}\{X(T) > 0\} > 0. \tag{5.4.23}$$
$$\mathbb{P}\left\{X(T) \geq \frac{X(0)}{D(T)}\right\} = 1, \mathbb{P}\left\{X(T) > \frac{X(0)}{D(T)} \right\} > 0 \tag{5.4.24}$$
(i) Suppose a multidimensional market model as described in Section 5.4.2 has an arbitrage. In other words, suppose there is a portfolio value process satisfying $X_1(0) = 0$ and 
$$\mathbb{P}\{X_1(T) \geq 0 \} = 1, \quad \mathbb{P}\{X_1(T) > 0 \} > 0, $$
for some positive $T$. Show that if $X_2(0)$ is positive, then there exists a portfolio value porcess $X_2(t)$ starting at $X_2(0)$ and satisfying
$$\mathbb{P} \left\{ X_2(T) \geq \frac{X_2(0)}{D(T)} \right\} = 1, \quad \mathbb{P} \left\{X_2(T) > \frac{X_2(0)}{D(T)} \right\} > 0. $$

(ii) Show that if a multidimensional market model has a portfolio value process $X_2(t)$ such that $X_2(0)$ is positive and (5.4.24) holds, then the model has a portfolio value process $X_1(0)$ such that $X_1(0) = 0$ and (5.4.23) holds.

### Interpretation

This exercise shows that the two definitions of arbitrage are equivalent:
- **Definition 1:** A portfolio starting at zero capital that generates profit with positive probability
- **Definition 2:** A portfolio starting at positive capital that is guaranteed to beat the money market

### (i) From $X_1$ to $X_2$

Given $X_1(t)$ satisfying $X_1(0) = 0$ and (5.4.23), define:

$$X_2(t) = \frac{X_2(0)}{D(t)} + X_1(t)$$

**Initial value:** $X_2(0) = X_2(0) + 0 = X_2(0)$ 

**At time $T$:**

$$X_2(T) = \frac{X_2(0)}{D(T)} + X_1(T)$$

Since $X_1(T) \geq 0$ a.s.:

$$\mathbb{P}\left\{X_2(T) \geq \frac{X_2(0)}{D(T)}\right\} = 1 \quad \blacksquare$$

Since $\mathbb{P}\{X_1(T) > 0\} > 0$:

$$\mathbb{P}\left\{X_2(T) > \frac{X_2(0)}{D(T)}\right\} > 0 \quad \blacksquare$$

### (ii) From $X_2$ to $X_1$

Given $X_2(t)$ satisfying $X_2(0) > 0$ and (5.4.24), define:

$$X_1(t) = X_2(t) - \frac{X_2(0)}{D(t)}$$

**Initial value:** $X_1(0) = X_2(0) - X_2(0) = 0$ 

**At time $T$:**

$$X_1(T) = X_2(T) - \frac{X_2(0)}{D(T)}$$

Since $X_2(T) \geq \frac{X_2(0)}{D(T)}$ a.s. by (5.4.24):

$$\mathbb{P}\{X_1(T) \geq 0\} = 1 \quad \blacksquare$$

Since $\mathbb{P}\{X_2(T) > \frac{X_2(0)}{D(T)}\} > 0$:

$$\mathbb{P}\{X_1(T) > 0\} > 0 \quad \blacksquare$$

### Conclusion

The two definitions of arbitrage are equivalent. Therefore, the First Fundamental Theorem of Asset Pricing (no-arbitrage $\iff$ existence of risk-neutral measure) holds regardless of which definition is used.

# Exercise 5.8 (Every strictly positive asset is a generalized geometric Brownian motion).
Let $(\Omega, \mathcal{F}, \mathbb{P})$ be a probability space on which is defined a Brownian motion $W(T), 0 \leq t \leq T.$ Let $\mathcal{F}(t), 0 \leq t \leq T,$ be the filtration generated by this Brownian motion. Assume there is a unique risk-neutral measure $\tilde{\mathbb{P}}$, and let $\tilde{W}(T), 0 \leq t \leq T,$ be the Brownian motion under $\tilde{\mathbb{P}}$ obtained by an application of Girsanob's Theorem, Theorem 5.2.3.

Corollary 5.3.2 of the Martingale Representation Theorem asserts that every martingale $\tilde{M}(t), 0 \leq t \leq T,$ under $\tilde{\mathbb{P}}$ can be written as a stochastic integral with respect to $\tilde{W}(t), 0 \leq t \leq T.$ In other words, there exists an adapted process $\tilde{\Gamma}(t), 0 \leq t \leq T,$ such that
$$\tilde{M}(t) = \tilde{M}(0) + \int_{0}^{t} \tilde{\Gamma}(u)d\tilde{W}(u), \quad 0 \leq t \leq T.$$
Now let $V(T)$ be an almost surely positive ("almost surely" means with probability one under both $\mathbb{P}$ and $\tilde{\mathbb{P}}$ since these two measures are equivalent), $\mathcal{F}(T)$-measurable random variable. According to the risk-neutral pricing formula (5.2.31), the price at time $t$ of a security paying $V(T)$ at time $T$ is 
$$V(t) = \tilde{\mathbb{E}}\left[e^{-\int_{t}^{T} R(u)du} V(T) |\mathcal{F}(t)\right], \quad 0 \leq t \leq T.$$
(i) Show that there exists an adapted process $\tilde{\Gamma}(t), \quad 0 \leq t \leq T,$ such that
$$dV(t) = R(t)V(t)dt + \frac{\tilde{\Gamma}(t)}{D(t)} d\tilde{W}(t), \quad 0 \leq t \leq T.$$
(ii) Show that for each $t \in [0,T],$ the price of the derivative security $V(t)$ at time $t$ is almost surely positive。

(iii) Conclude from (i) and (ii) that there exists an adapted process $\sigma(t), 0 \leq t \leq T,$ such that
$$dV(t) = R(t)V(t)dt + \sigma(t)V(t)d\tilde{W}(t),  \quad 0 \leq t \leq T.$$
In other words, prior to time $T$, the price of every asset with almost surely positive price at time $T$ follows a generalized (Because the volatility may be random) geometric Brownian motion.

### (i) SDE for $V(t)$

**Step 1:** By the risk-neutral pricing formula, $D(t)V(t)$ is a $\tilde{\mathbb{P}}$-martingale:

$$D(t)V(t) = \tilde{\mathbb{E}}\left[D(T)V(T)|\mathcal{F}(t)\right]$$

**Step 2:** By the Martingale Representation Theorem (Corollary 5.3.2), there exists an adapted process $\tilde{\Gamma}(t)$ such that:

$$d(D(t)V(t)) = \tilde{\Gamma}(t)d\tilde{W}(t)$$

**Step 3:** By Itô's product rule, using $dD(t) = -R(t)D(t)dt$ and $dD\,dV = 0$:

$$d(D(t)V(t)) = D(t)dV(t) + V(t)dD(t) = D(t)dV(t) - R(t)D(t)V(t)dt$$

**Step 4:** Setting the two expressions equal and solving for $dV(t)$:

$$D(t)dV(t) - R(t)D(t)V(t)dt = \tilde{\Gamma}(t)d\tilde{W}(t)$$

$$dV(t) = R(t)V(t)dt + \frac{\tilde{\Gamma}(t)}{D(t)}d\tilde{W}(t) \quad \blacksquare$$

### (ii) $V(t) > 0$ almost surely

Since $D(t) = e^{-\int_0^t R(u)du} > 0$ deterministically, we have:

$$\frac{V(t)}{D(t)} = \tilde{\mathbb{E}}\left[\frac{V(T)}{D(T)}\bigg|\mathcal{F}(t)\right]$$

Since $V(T) > 0$ a.s. and $D(T) > 0$, we have $\frac{V(T)}{D(T)} > 0$ a.s., so the conditional expectation is strictly positive:

$$\frac{V(t)}{D(t)} > 0 \quad \text{a.s.}$$

Therefore $V(t) > 0$ a.s. for all $t \in [0,T]$. $\blacksquare$

### (iii) Generalized geometric Brownian motion

From (ii), $V(t) > 0$ a.s., so we may define:

$$\sigma(t) = \frac{\tilde{\Gamma}(t)}{D(t)V(t)}$$

which is well-defined since $D(t) > 0$ and $V(t) > 0$ a.s. Substituting into (i):

$$dV(t) = R(t)V(t)dt + \sigma(t)V(t)d\tilde{W}(t) \quad \blacksquare$$
