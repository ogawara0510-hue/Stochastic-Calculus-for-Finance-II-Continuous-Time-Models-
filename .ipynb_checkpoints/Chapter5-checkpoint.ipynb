# Exercise 5.1.
Consider the discounted stock price $D(t)S(t)$ of (5.2.19). In this problem, we derive the formula (5.2.20) for $d(D(t)S(t))$ by two methods.

(i) Define $f(x) = S(0)e^x$ and set
$$X(t) = \int_{0}^{t} \sigma(s) dW(s) + \int_{0}^{t} ( \alpha(s) - R(s) - \frac{1}{2} \sigma^2 (s))ds$$
so that $D(t)S(t) = f(X(t)).$ Use the itô-Doeblin formula to compute $df(X(t))$.

(ii) According to Itô's product rule,
$$d(D(t)S(t)) = S(t)dD(t) + D(t)dS(t) + dD(t)dS(t).$$
Use (5.2.15) and (5.2.18) to work out the right-hand side of this equation.
$$dS(t) = \alpha(t)S(t)dt + \sigma(t)S(t)dW(t), 0 \leq t \leq T. \tag{5.2.15}$$
$$dD(t) = -R(t)D(t)dt. \tag{5.2.18}$$
$$D(t)S(t) = S(0)\exp \left\{ \int_{0}^{t} \sigma(s)dW(s) + \int_{0}^{t} \left(\alpha(s) - R(s) - \frac{1}{2} \sigma^2(s)\right) ds \right\}, \tag{5.2.19}$$
$$d(D(t)S(t)) = (\alpha(t) - R(t))D(t)S(t)dt + \sigma(t)D(t)S(t)dW(t) = \sigma(t)D(t)S(t)[\mathcal{\Theta}(t)dt + dW(t)], \tag{5.2.20}$$

### (i) Itô-Doeblin formula

Define $f(x) = S(0)e^x$, so $f'(x) = f''(x) = S(0)e^x$.

$$dX(t) = \sigma(t)dW(t) + \left(\alpha(t) - R(t) - \frac{1}{2}\sigma^2(t)\right)dt, \quad dX\,dX = \sigma^2(t)dt$$

By the Itô-Doeblin formula:

$$df(X(t)) = f'(X(t))dX(t) + \frac{1}{2}f''(X(t))dX\,dX$$

$$= D(t)S(t)\left[\sigma(t)dW(t) + \left(\alpha(t) - R(t) - \frac{1}{2}\sigma^2(t)\right)dt + \frac{1}{2}\sigma^2(t)dt\right]$$

$$= D(t)S(t)\left[(\alpha(t) - R(t))dt + \sigma(t)dW(t)\right] = \sigma(t)D(t)S(t)\left[\mathcal{\Theta}(t)dt + dW(t)\right] \quad \blacksquare$$

### (ii) Itô's product rule

Using (5.2.15) and (5.2.18):

**Term 1:**
$$S(t)dD(t) = -R(t)D(t)S(t)dt$$

**Term 2:**
$$D(t)dS(t) = \alpha(t)D(t)S(t)dt + \sigma(t)D(t)S(t)dW(t)$$

**Term 3:**
$$dD(t)dS(t) = (-R(t)D(t)dt)(\alpha(t)S(t)dt + \sigma(t)S(t)dW(t)) = 0$$

since $dt\cdot dt = 0$ and $dt\cdot dW = 0$.

Summing all terms:

$$d(D(t)S(t)) = (\alpha(t) - R(t))D(t)S(t)dt + \sigma(t)D(t)S(t)dW(t) = \sigma(t)D(t)S(t)[\mathcal{\Theta}(t)dt + dW(t)] \quad \blacksquare$$

# Exercise 5.2 (State price density process).
Show that the risk-neutral pricing formula (5.2.30) may be rewritten as
$$D(t)Z(t)V(t) = \mathbb{E}[D(T)Z(T)V(T)|\mathcal{F}(t)]. \tag{5.9.1}$$
Here $Z(t)$ is the Radon-Nikodym derivative process (5.2.11) when the market price of risk process $\mathcal{\Theta}(t)$ is given by (5.2.21) and the conditional expectation on the right-hand side of (5.9.1) is taken under the actual probability measure $\mathbb{P},$ not the risk-neutral measure $\tilde{\mathbb{P}}.$ In particular, if for some $A \in \mathcal{F}(T)$ a derivative security pays off $\mathbb{I}_A$ (i.e., pays $1$ if $A$ occures and $0$ if $A$ does not occur), then the value of this derivative security at time zero is $\mathbb{E}[D(T)Z(T)\mathbb{I}_A].$ The process $D(t)Z(t)$ appearing in (5.9.1) is called the state price density process.
$$Z(t) = \exp \left\{\int_{0}^{t} \mathcal{\Theta}(u)dW(u) - \frac{1}{2} \int_{0}^{t} \mathcal{\Theta}^2(u)du\right\} \tag{5.2.11}$$
$$ \text{market price of risk : } \mathcal{\Theta}(t) = \frac{\alpha(t) - R(t)}{\sigma(t)}. \tag{5.2.21}$$
$$D(t)V(t) = \tilde{\mathbb{E}}[D(T)V(T)|\mathcal{F}(t)], 0 \leq t \leq T. \tag{5.2.30}$$

**Proof:**

Since $Z(t)$ is the Radon-Nikodym derivative process, by Lemma 5.2.2, for any $\mathcal{F}(T)$-measurable random variable $X$:

$$\tilde{\mathbb{E}}[X|\mathcal{F}(t)] = \frac{1}{Z(t)}\mathbb{E}[Z(T)X|\mathcal{F}(t)]$$

Applying this to $X = D(T)V(T)$, which is $\mathcal{F}(T)$-measurable:

$$\tilde{\mathbb{E}}[D(T)V(T)|\mathcal{F}(t)] = \frac{1}{Z(t)}\mathbb{E}[D(T)V(T)Z(T)|\mathcal{F}(t)]$$

From the risk-neutral pricing formula (5.2.30):

$$D(t)V(t) = \tilde{\mathbb{E}}[D(T)V(T)|\mathcal{F}(t)] = \frac{1}{Z(t)}\mathbb{E}[D(T)V(T)Z(T)|\mathcal{F}(t)]$$

Multiplying both sides by $Z(t)$:

$$D(t)Z(t)V(t) = \mathbb{E}[D(T)Z(T)V(T)|\mathcal{F}(t)] \quad \blacksquare$$

**Remark:** In particular, for a derivative security paying $\mathbb{I}_A$ at time $T$ (where $A \in \mathcal{F}(T)$), its value at time zero is:

$$V(0) = \mathbb{E}[D(T)Z(T)\mathbb{I}_A]$$

This justifies calling $D(t)Z(t)$ the **state price density process**.

# Exercise 5.3.
According to the Black-Scholes-Merton formula, the value at time zero of a European call on a stock whose initial price is $S(0) = x$ is given by 
$$c(0, x) = xN(d_+(T, x)) - K e^{-rT} N(d_-(T, x)), $$
where
$$d_+(T, x) = \frac{1}{\sigma\sqrt{T}} \left[\log \frac{x}{K} + \left(r + \frac{1}{2} \sigma^2 \right) T\right],$$
$$d_-(T, x) = d_+(T, x) - \sigma \sqrt{T}.$$
The stock is modeled as a geometric Brownian motion with constant volatility $\sigma > 0$, the interest rate is constant $r$, the call strike is $K$, and the call expiration time is $T$. This formula is obtained by computing the discounted expected payoff of the call under the risk-neutral measure,
$$c(0, x) = \tilde{\mathbb{E}}[e^{-rT}(S(T) - K)^+] = \tilde{\mathbb{E}}\left[ e^{-rT} \left(x \exp \left\{\sigma \tilde{W}(T) + \left(r - \frac{1}{2}\sigma^2 \right) T \right\} - K\right)^+ \right], \tag{5.9.2}$$
where $\tilde{W}$ is a Brownian motion under the risk-neutral measure $\tilde{\mathbb{P}}$. In exercise 4.9(ii), the delta of this option is computed to be $c_x(0, x) = N(d_+(T, x)).$ This problem provides an alternate way to compute $c_x(0, x).$

(i) We begin with the observation that if $h(s) = (s - K)^+,$ then 
$$h'(s) = \begin{cases} 0 & \text{ if } s < K, \\ 1 & \text{ if } s > K. \end{cases}$$
if $s = K$, then $h'(s)$ is undefined, but that will not matter in what follows because $S(T)$ has zero probability of takeing the value $K$. Using this formula for $h'(s)$, differentiate inside the expected value in (5.9.2) to obtain a formula for $c_x(0, x).$

(ii) Show that the formula you obtained in (i) can be rewritten as 
$$c_x(0,x) = \hat{\mathbb{P}}(S(T) > K),$$
where $\tilde{\mathbb{P}}$ is a probability measure equivalent to $\tilde{\mathbb{P}}.$ Show that
$$\hat{W}(t) = \tilde{W}(t) - \sigma t$$
is a Brownian motion under $\tilde{\mathbb{P}}$.

(iii) Rewrite $S(T)$ in terms of $\hat{W}(T)$, and then show that
$$\hat{\mathbb{P}}\{S(T) > K \} = \hat{\mathbb{P}}\left\{- \frac{\hat{W}(T)}{\sqrt{T}} < d_+(T, x)\right\} = N(d_+(T, x)).$$


### (i) Differentiating inside the expectation

Using $h'(s) = \mathbb{I}_{\{s > K\}}$ and the chain rule:

$$c_x(0,x) = \tilde{\mathbb{E}}\left[e^{-rT}\mathbb{I}_{\{S(T)>K\}}\exp\left\{\sigma\tilde{W}(T) + \left(r-\frac{1}{2}\sigma^2\right)T\right\}\right]$$

Setting $Z = -\frac{\tilde{W}(T)}{\sqrt{T}} \sim \mathcal{N}(0,1)$ under $\tilde{\mathbb{P}}$, and noting $S(T) > K \iff Z < d_-$:

$$c_x(0,x) = \int_{-\infty}^{d_-} e^{-\sigma\sqrt{T}z - \frac{1}{2}\sigma^2 T} \phi(z)dz$$

Substituting $y = z + \sigma\sqrt{T}$ (so $z = y - \sigma\sqrt{T}$, lower limit $d_- + \sigma\sqrt{T} = d_+$... wait, $d_- \to -\infty$ maps to $d_+$):

$$= \int_{-\infty}^{d_+} \phi(y)dy = N(d_+(T,x)) \quad \blacksquare$$

### (ii) Measure change via Radon-Nikodym

From (i):

$$c_x(0,x) = \tilde{\mathbb{E}}\left[\mathbb{I}_{\{S(T)>K\}} \cdot e^{\sigma\tilde{W}(T) - \frac{1}{2}\sigma^2 T}\right]$$

Define the Radon-Nikodym derivative:

$$\frac{d\hat{\mathbb{P}}}{d\tilde{\mathbb{P}}} = e^{\sigma\tilde{W}(T) - \frac{1}{2}\sigma^2 T}$$

Then:

$$c_x(0,x) = \hat{\mathbb{E}}[\mathbb{I}_{\{S(T)>K\}}] = \hat{\mathbb{P}}\{S(T) > K\}$$

By Girsanov's theorem, $\hat{W}(t) = \tilde{W}(t) - \sigma t$ is a Brownian motion under $\hat{\mathbb{P}}$. $\blacksquare$

### (iii) Computing $\hat{\mathbb{P}}\{S(T) > K\}$

Rewriting $S(T)$ in terms of $\hat{W}(T) = \tilde{W}(T) - \sigma T$:

$$S(T) = x\exp\left\{\sigma\tilde{W}(T) + \left(r-\frac{1}{2}\sigma^2\right)T\right\} = x\exp\left\{\sigma(\hat{W}(T) + \sigma T) + \left(r-\frac{1}{2}\sigma^2\right)T\right\}$$

$$= x\exp\left\{\sigma\hat{W}(T) + \left(r+\frac{1}{2}\sigma^2\right)T\right\}$$

$S(T) > K$ is equivalent to:

$$\sigma\hat{W}(T) + \left(r+\frac{1}{2}\sigma^2\right)T > \log\frac{K}{x}$$

Setting $Y = -\frac{\hat{W}(T)}{\sqrt{T}} \sim \mathcal{N}(0,1)$ under $\hat{\mathbb{P}}$:

$$-\sigma\sqrt{T}Y + \left(r+\frac{1}{2}\sigma^2\right)T > \log\frac{K}{x}$$

$$Y < \frac{1}{\sigma\sqrt{T}}\left[\log\frac{x}{K} + \left(r+\frac{1}{2}\sigma^2\right)T\right] = d_+(T,x)$$

Therefore:

$$\hat{\mathbb{P}}\{S(T) > K\} = \hat{\mathbb{P}}\left\{-\frac{\hat{W}(T)}{\sqrt{T}} < d_+(T,x)\right\} = N(d_+(T,x)) \quad \blacksquare$$

# Exercise 5.4 (Black-Sholes-Merton formula for time-varying, non-random interest rate and volatility).
Consider a stock whose price differential is
$$dS(t) = r(t)S(t)dt + \sigma(t)d\tilde{W}(t), $$
where $r(t)$ and $\sigma(t)$ are nonrandom functions of $t$ and $\tilde{W}$ is a Brownian motion under the risk-neutral measure $\tilde{\mathbb{P}}.$ Let $T > 0$ be given, and consider a European call, whose value at time zero is
$$c(0, S(0)) = \mathbb{E}\left[\exp \left\{ - \int_{0}^{T} r(t)dt \right\} (S(T) - K) ^ + \right].$$
(i) Show that $S(T)$ is of the for $S(0)e^X$, where X is a normal random variable, and deterine the mean and varianve of $X$.

(ii) Let 
$$\text{BSM}(T,x; K, R \Sigma) = xN\left(\frac{1}{\Sigma \sqrt{T}} \left[ \log \frac{x}{K} + (R + \Sigma^2/2)T\right] \right) - e^{-RT} K N\left(\frac{1}{\Sigma \sqrt{T}} \left[\log \frac{x}{K} + (R - \Sigma^2/2)T\right] \right)$$
dennote the value at time zero of a European call expiring at time $T$ when the underlying stock has constant volatility $\Sigma$ and the interest rate $R$ is constant. Show that
$$c(0, S(0)) = \text{BSM}\left(S(0), T, \frac{1}{T} \int_{0}^{T} r(t)dt, \sqrt{\frac{1}{T} \int_{0}^{T} \sigma^2(t)dt} \right). $$

### (i) Distribution of $X$

Let $f(x) = \log x$, so $f'(x) = \frac{1}{x}$ and $f''(x) = -\frac{1}{x^2}$.

By Itô's lemma:

$$d\log S(t) = \left(r(t) - \frac{1}{2}\sigma^2(t)\right)dt + \sigma(t)d\tilde{W}(t)$$

Integrating from $0$ to $T$:

$$S(T) = S(0)e^X, \quad X = \int_0^T\left(r(t)-\frac{1}{2}\sigma^2(t)\right)dt + \int_0^T\sigma(t)d\tilde{W}(t)$$

Since $r(t)$ and $\sigma(t)$ are nonrandom, $X$ is normally distributed:

$$X \sim \mathcal{N}\left(\int_0^T\left(r(t)-\frac{1}{2}\sigma^2(t)\right)dt,\ \int_0^T\sigma^2(t)dt\right) \quad \blacksquare$$

### (ii) BSM formula with averaged parameters

Define:

$$RT = \int_0^T r(t)dt, \quad \Sigma^2 T = \int_0^T\sigma^2(t)dt$$

so $R = \frac{1}{T}\int_0^T r(t)dt$ and $\Sigma = \sqrt{\frac{1}{T}\int_0^T\sigma^2(t)dt}$.

Also define:

$$d_\pm = \frac{1}{\Sigma\sqrt{T}}\left[\log\frac{S(0)}{K} + \left(R \pm \frac{\Sigma^2}{2}\right)T\right]$$

Setting $Y = -\frac{X - \int_0^T(r(t)-\frac{1}{2}\sigma^2(t))dt}{\Sigma\sqrt{T}} \sim \mathcal{N}(0,1)$, $S(T) > K$ is equivalent to $Y < d_-$. Therefore:

$$c(0,S(0)) = \int_{-\infty}^{d_-} e^{-RT}(S(0)e^X - K)\phi(y)dy$$

**Second term:**

$$-e^{-RT}K\int_{-\infty}^{d_-}\phi(y)dy = -Ke^{-RT}N(d_-)$$

**First term:**

$$S(0)e^{-RT}\int_{-\infty}^{d_-}e^X\phi(y)dy$$

Since $X = \int_0^T(r(t)-\frac{1}{2}\sigma^2(t))dt - \Sigma\sqrt{T}Y = RT - \frac{1}{2}\Sigma^2 T - \Sigma\sqrt{T}Y$:

$$= S(0)e^{-RT}\cdot e^{RT - \frac{1}{2}\Sigma^2 T}\int_{-\infty}^{d_-}e^{-\Sigma\sqrt{T}y}\phi(y)dy$$

$$= S(0)e^{-\frac{1}{2}\Sigma^2 T}\int_{-\infty}^{d_-}\frac{1}{\sqrt{2\pi}}\exp\left\{-\Sigma\sqrt{T}y - \frac{y^2}{2}\right\}dy$$

$$= S(0)\int_{-\infty}^{d_-}\frac{1}{\sqrt{2\pi}}\exp\left\{-\frac{(y+\Sigma\sqrt{T})^2}{2}\right\}dy$$

Substituting $z = y + \Sigma\sqrt{T}$:

$$= S(0)\int_{-\infty}^{d_+}\phi(z)dz = S(0)N(d_+)$$

**Combining:**

$$c(0,S(0)) = S(0)N(d_+) - Ke^{-RT}N(d_-) = \text{BSM}\left(S(0), T, \frac{1}{T}\int_0^T r(t)dt, \sqrt{\frac{1}{T}\int_0^T\sigma^2(t)dt}\right) \quad \blacksquare$$

# Exercise 5.5.
Prove Corolarry 5.3.2 by the following steps.

(i) Compute the differential of $\frac{1}{Z(t)}$, where $Z(t)$ is given in Corollary 5.3.2.

(ii) Let $\tilde{M}(t), 0 \leq t \leq T,$ be a martingale under $\tilde{\mathbb{P}}.$ Show that $M(T) = Z(t) \tilde{M}(t)$ is a martingale under $\mathbb{P}.$

(iii) According to Theorem 5.3.1, there is an adapted process $\Gamma(u), 0 \leq u \leq T,$ such that
$$M(t) = M(0) + \int_{0}^{T} \Gamma(u) dW(u), 0 \leq t \leq T.$$
Write $\tilde{M}(t) = M(t) \cdot \frac{1}{Z(t)}$ and take its differential using ITô's product rule.

(iv) Show that the differential of $\tilde{M}(t)$ is the sum of an adapted process, which we call $\tilde{\Gamma}(t),$ times $d\tilde{W}(t),$ and zero times dt. Integrate to obtain(5.3.2).

## Exercise 5.5

### (i) Differential of $\frac{1}{Z(t)}$

Since $dZ(t) = -\Theta(t)Z(t)dW(t)$, by Itô's lemma with $f(z) = \frac{1}{z}$:

$$d\left(\frac{1}{Z(t)}\right) = -\frac{1}{Z^2(t)}dZ(t) + \frac{1}{Z^3(t)}(dZ)^2$$

$$= \frac{\Theta(t)}{Z(t)}dW(t) + \frac{\Theta^2(t)}{Z(t)}dt = \frac{\Theta(t)}{Z(t)}\left[dW(t) + \Theta(t)dt\right] = \frac{\Theta(t)}{Z(t)}d\tilde{W}(t) \quad \blacksquare$$

### (ii) $M(t) = Z(t)\tilde{M}(t)$ is a martingale under $\mathbb{P}$

For $s \leq t$, by Lemma 5.2.2:

$$\mathbb{E}[M(t)|\mathcal{F}(s)] = \mathbb{E}[Z(t)\tilde{M}(t)|\mathcal{F}(s)] = Z(s)\tilde{\mathbb{E}}[\tilde{M}(t)|\mathcal{F}(s)] = Z(s)\tilde{M}(s) = M(s) \quad \blacksquare$$

### (iii) Differential of $\tilde{M}(t) = M(t)\cdot\frac{1}{Z(t)}$

By Itô's product rule:

$$d\tilde{M}(t) = \frac{1}{Z(t)}dM(t) + M(t)d\frac{1}{Z(t)} + dM(t)d\frac{1}{Z(t)}$$

$$= \frac{\Gamma(t)}{Z(t)}dW(t) + \frac{M(t)\Theta(t)}{Z(t)}d\tilde{W}(t) + \frac{\Gamma(t)}{Z(t)}dW(t)\cdot\frac{\Theta(t)}{Z(t)}d\tilde{W}(t) \cdot Z(t)$$

Computing the cross term using $dW\cdot d\tilde{W} = dW\cdot(dW + \Theta\,dt) = dt$:

$$dM(t)\cdot d\frac{1}{Z(t)} = \frac{\Gamma(t)\Theta(t)}{Z(t)}dt$$

Therefore:

$$d\tilde{M}(t) = \frac{\Gamma(t)}{Z(t)}dW(t) + \frac{M(t)\Theta(t)}{Z(t)}d\tilde{W}(t) + \frac{\Gamma(t)\Theta(t)}{Z(t)}dt$$

### (iv) Cancellation of $dt$ terms

Substituting $dW(t) = d\tilde{W}(t) - \Theta(t)dt$:

$$\frac{\Gamma(t)}{Z(t)}dW(t) = \frac{\Gamma(t)}{Z(t)}d\tilde{W}(t) - \frac{\Gamma(t)\Theta(t)}{Z(t)}dt$$

Therefore:

$$d\tilde{M}(t) = \frac{\Gamma(t)}{Z(t)}d\tilde{W}(t) - \frac{\Gamma(t)\Theta(t)}{Z(t)}dt + \frac{M(t)\Theta(t)}{Z(t)}d\tilde{W}(t) + \frac{\Gamma(t)\Theta(t)}{Z(t)}dt$$

The $dt$ terms cancel, leaving:

$$d\tilde{M}(t) = \underbrace{\left(\frac{\Gamma(t) + M(t)\Theta(t)}{Z(t)}\right)}_{=:\tilde{\Gamma}(t)}d\tilde{W}(t)$$

Integrating:

$$\tilde{M}(t) = \tilde{M}(0) + \int_0^t \tilde{\Gamma}(u)d\tilde{W}(u) \quad \blacksquare$$

# Exercise 5.6. 
Use the two-dimmensional Levy Theorem, Theorem 4.6.5, to prove the two-dimmensional Girsanov Theorem(i.e., Theorem 5.4.1 widh $d = 2$).

**Proof:**

Define:

$$d\tilde{W}_i(t) = dW_i(t) + \Theta_i(t)dt, \quad i = 1, 2$$

By the two-dimensional Lévy Theorem (Theorem 4.6.5), it suffices to show that each $\tilde{W}_i(t)$ is a $\tilde{\mathbb{P}}$-martingale and $d\tilde{W}_i\,d\tilde{W}_j = \delta_{ij}dt$.

**Step 1: $\tilde{W}_i(t)$ is a $\tilde{\mathbb{P}}$-martingale**

Since $dZ(t) = -Z(t)\sum_{j=1}^2\Theta_j(t)dW_j(t)$, by Itô's product rule:

$$d(\tilde{W}_i Z) = \tilde{W}_i\,dZ + Z\,d\tilde{W}_i + d\tilde{W}_i\,dZ$$

Computing each term:

$$\tilde{W}_i\,dZ = -\tilde{W}_i Z\sum_{j=1}^2\Theta_j\,dW_j \quad \text{(Itô integral only)}$$

$$Z\,d\tilde{W}_i = Z\,dW_i + Z\Theta_i\,dt$$

$$d\tilde{W}_i\,dZ = (dW_i + \Theta_i\,dt)\left(-Z\sum_{j=1}^2\Theta_j\,dW_j\right) = -Z\Theta_i\,dt$$

The $dt$ terms cancel: $Z\Theta_i\,dt - Z\Theta_i\,dt = 0$. Therefore:

$$d(\tilde{W}_i Z) = \sum_{j=1}^2\Gamma_j(t)dW_j(t)$$

for some adapted processes $\Gamma_j$, so $\tilde{W}_i(t)Z(t)$ is a $\mathbb{P}$-martingale. By Lemma 5.2.2, $\tilde{W}_i(t)$ is a $\tilde{\mathbb{P}}$-martingale.

**Step 2: Quadratic covariation**

$$d\tilde{W}_i\,d\tilde{W}_j = (dW_i + \Theta_i\,dt)(dW_j + \Theta_j\,dt) = dW_i\,dW_j = \delta_{ij}dt$$

since $dW_i\,dt = 0$ and $(dt)^2 = 0$.

**Conclusion:**

By the two-dimensional Lévy Theorem, $(\tilde{W}_1(t), \tilde{W}_2(t))$ is a two-dimensional Brownian motion under $\tilde{\mathbb{P}}$. $\blacksquare$

# Exercise 5.7.
$$\mathbb{P}\{X(T) \geq 0\} = 1, \quad \mathbb{P}\{X(T) > 0\} > 0. \tag{5.4.23}$$
$$\mathbb{P}\left\{X(T) \geq \frac{X(0)}{D(T)}\right\} = 1, \mathbb{P}\left\{X(T) > \frac{X(0)}{D(T)} \right\} > 0 \tag{5.4.24}$$
(i) Suppose a multidimensional market model as described in Section 5.4.2 has an arbitrage. In other words, suppose there is a portfolio value process satisfying $X_1(0) = 0$ and 
$$\mathbb{P}\{X_1(T) \geq 0 \} = 1, \quad \mathbb{P}\{X_1(T) > 0 \} > 0, $$
for some positive $T$. Show that if $X_2(0)$ is positive, then there exists a portfolio value porcess $X_2(t)$ starting at $X_2(0)$ and satisfying
$$\mathbb{P} \left\{ X_2(T) \geq \frac{X_2(0)}{D(T)} \right\} = 1, \quad \mathbb{P} \left\{X_2(T) > \frac{X_2(0)}{D(T)} \right\} > 0. $$

(ii) Show that if a multidimensional market model has a portfolio value process $X_2(t)$ such that $X_2(0)$ is positive and (5.4.24) holds, then the model has a portfolio value process $X_1(0)$ such that $X_1(0) = 0$ and (5.4.23) holds.

### Interpretation

This exercise shows that the two definitions of arbitrage are equivalent:
- **Definition 1:** A portfolio starting at zero capital that generates profit with positive probability
- **Definition 2:** A portfolio starting at positive capital that is guaranteed to beat the money market

### (i) From $X_1$ to $X_2$

Given $X_1(t)$ satisfying $X_1(0) = 0$ and (5.4.23), define:

$$X_2(t) = \frac{X_2(0)}{D(t)} + X_1(t)$$

**Initial value:** $X_2(0) = X_2(0) + 0 = X_2(0)$ 

**At time $T$:**

$$X_2(T) = \frac{X_2(0)}{D(T)} + X_1(T)$$

Since $X_1(T) \geq 0$ a.s.:

$$\mathbb{P}\left\{X_2(T) \geq \frac{X_2(0)}{D(T)}\right\} = 1 \quad \blacksquare$$

Since $\mathbb{P}\{X_1(T) > 0\} > 0$:

$$\mathbb{P}\left\{X_2(T) > \frac{X_2(0)}{D(T)}\right\} > 0 \quad \blacksquare$$

### (ii) From $X_2$ to $X_1$

Given $X_2(t)$ satisfying $X_2(0) > 0$ and (5.4.24), define:

$$X_1(t) = X_2(t) - \frac{X_2(0)}{D(t)}$$

**Initial value:** $X_1(0) = X_2(0) - X_2(0) = 0$ 

**At time $T$:**

$$X_1(T) = X_2(T) - \frac{X_2(0)}{D(T)}$$

Since $X_2(T) \geq \frac{X_2(0)}{D(T)}$ a.s. by (5.4.24):

$$\mathbb{P}\{X_1(T) \geq 0\} = 1 \quad \blacksquare$$

Since $\mathbb{P}\{X_2(T) > \frac{X_2(0)}{D(T)}\} > 0$:

$$\mathbb{P}\{X_1(T) > 0\} > 0 \quad \blacksquare$$

### Conclusion

The two definitions of arbitrage are equivalent. Therefore, the First Fundamental Theorem of Asset Pricing (no-arbitrage $\iff$ existence of risk-neutral measure) holds regardless of which definition is used.

# Exercise 5.8 (Every strictly positive asset is a generalized geometric Brownian motion).
Let $(\Omega, \mathcal{F}, \mathbb{P})$ be a probability space on which is defined a Brownian motion $W(T), 0 \leq t \leq T.$ Let $\mathcal{F}(t), 0 \leq t \leq T,$ be the filtration generated by this Brownian motion. Assume there is a unique risk-neutral measure $\tilde{\mathbb{P}}$, and let $\tilde{W}(T), 0 \leq t \leq T,$ be the Brownian motion under $\tilde{\mathbb{P}}$ obtained by an application of Girsanob's Theorem, Theorem 5.2.3.

Corollary 5.3.2 of the Martingale Representation Theorem asserts that every martingale $\tilde{M}(t), 0 \leq t \leq T,$ under $\tilde{\mathbb{P}}$ can be written as a stochastic integral with respect to $\tilde{W}(t), 0 \leq t \leq T.$ In other words, there exists an adapted process $\tilde{\Gamma}(t), 0 \leq t \leq T,$ such that
$$\tilde{M}(t) = \tilde{M}(0) + \int_{0}^{t} \tilde{\Gamma}(u)d\tilde{W}(u), \quad 0 \leq t \leq T.$$
Now let $V(T)$ be an almost surely positive ("almost surely" means with probability one under both $\mathbb{P}$ and $\tilde{\mathbb{P}}$ since these two measures are equivalent), $\mathcal{F}(T)$-measurable random variable. According to the risk-neutral pricing formula (5.2.31), the price at time $t$ of a security paying $V(T)$ at time $T$ is 
$$V(t) = \tilde{\mathbb{E}}\left[e^{-\int_{t}^{T} R(u)du} V(T) |\mathcal{F}(t)\right], \quad 0 \leq t \leq T.$$
(i) Show that there exists an adapted process $\tilde{\Gamma}(t), \quad 0 \leq t \leq T,$ such that
$$dV(t) = R(t)V(t)dt + \frac{\tilde{\Gamma}(t)}{D(t)} d\tilde{W}(t), \quad 0 \leq t \leq T.$$
(ii) Show that for each $t \in [0,T],$ the price of the derivative security $V(t)$ at time $t$ is almost surely positive。

(iii) Conclude from (i) and (ii) that there exists an adapted process $\sigma(t), 0 \leq t \leq T,$ such that
$$dV(t) = R(t)V(t)dt + \sigma(t)V(t)d\tilde{W}(t),  \quad 0 \leq t \leq T.$$
In other words, prior to time $T$, the price of every asset with almost surely positive price at time $T$ follows a generalized (Because the volatility may be random) geometric Brownian motion.

### (i) SDE for $V(t)$

**Step 1:** By the risk-neutral pricing formula, $D(t)V(t)$ is a $\tilde{\mathbb{P}}$-martingale:

$$D(t)V(t) = \tilde{\mathbb{E}}\left[D(T)V(T)|\mathcal{F}(t)\right]$$

**Step 2:** By the Martingale Representation Theorem (Corollary 5.3.2), there exists an adapted process $\tilde{\Gamma}(t)$ such that:

$$d(D(t)V(t)) = \tilde{\Gamma}(t)d\tilde{W}(t)$$

**Step 3:** By Itô's product rule, using $dD(t) = -R(t)D(t)dt$ and $dD\,dV = 0$:

$$d(D(t)V(t)) = D(t)dV(t) + V(t)dD(t) = D(t)dV(t) - R(t)D(t)V(t)dt$$

**Step 4:** Setting the two expressions equal and solving for $dV(t)$:

$$D(t)dV(t) - R(t)D(t)V(t)dt = \tilde{\Gamma}(t)d\tilde{W}(t)$$

$$dV(t) = R(t)V(t)dt + \frac{\tilde{\Gamma}(t)}{D(t)}d\tilde{W}(t) \quad \blacksquare$$

### (ii) $V(t) > 0$ almost surely

Since $D(t) = e^{-\int_0^t R(u)du} > 0$ deterministically, we have:

$$\frac{V(t)}{D(t)} = \tilde{\mathbb{E}}\left[\frac{V(T)}{D(T)}\bigg|\mathcal{F}(t)\right]$$

Since $V(T) > 0$ a.s. and $D(T) > 0$, we have $\frac{V(T)}{D(T)} > 0$ a.s., so the conditional expectation is strictly positive:

$$\frac{V(t)}{D(t)} > 0 \quad \text{a.s.}$$

Therefore $V(t) > 0$ a.s. for all $t \in [0,T]$. $\blacksquare$

### (iii) Generalized geometric Brownian motion

From (ii), $V(t) > 0$ a.s., so we may define:

$$\sigma(t) = \frac{\tilde{\Gamma}(t)}{D(t)V(t)}$$

which is well-defined since $D(t) > 0$ and $V(t) > 0$ a.s. Substituting into (i):

$$dV(t) = R(t)V(t)dt + \sigma(t)V(t)d\tilde{W}(t) \quad \blacksquare$$

# Exercise 5.9 (Implying the risk-neutral distribution).
Let $S(t)$ be the price of an underlying asset, which is not necessarily a geometric Brownian motion(i.e., does not necessarily have constant volatility). With $S(0) = x$, the risk-neutral pricing formula for the price at time zero of a European call on this asset, paying $(S(T) - K)~+$ at time $T$, is
$$c(0, T, x, K) = \tilde{\mathbb{E}}[e^{-rT}(S(T) - K)^+].$$
(Normally we consider this as a function of the current time $0$ and the current stock price $x$, but in this exercise we shall also treat the expiration time T and the strike price K as variable, and for that reason we include them as argument of $c.$) We denote by $\tilde{p}(0, T, x, y)$ the risk-neutral density in the $y$ variable of the distribution of $S(T)$ when $S(0) = x.$ Then we may rewrite the risk-neutral pricing formula as
$$c(0, T, x, K) = e^{-rT} \int_{K}^{\infty} (y - K) \tilde{p}(0,T,x,y)dy.　\tag{5.9.3}$$
Suppose we know the market prices for calls of all strikes (i.e., we knoe $c(0, T, x, K)$ for all $K > 0$). We can then compute $c_K(0, T, x, K)$ and $c_{KK}(0, T, x, K),$ the first and second derivatives of the option price with respect to the strike. Differentiate (5.9.3) twice with respect to $K$ to obtain the equations
$$c_K(0, T, x, K) = -e^{-rT} \int_{K}^{\infty} \tilde{p}(0,T,x,y)dy = -e^{rT}\tilde{\mathbb{P}}\S(T) > K\}, $$
$$c_{KK}(0,T,x,K) = e^{-rT} \tilde{p}(0,T,x,K).$$
The second of these equations provides a formula for the risk-neutral distribution of $S(T)$ in terms of call prices:
$$\tilde{p}(0, T, x, K) = e^{rT}c_{KK}(0,T, x, K) \text{ for all }K > 0.$$

### Leibniz Rule

For an integral of the form $F(K) = \int_{K}^{\infty} f(K, y) dy$, the Leibniz rule states:

$$\frac{d}{dK}\int_{K}^{\infty} f(K,y)dy = -f(K,K) + \int_{K}^{\infty} \frac{\partial}{\partial K}f(K,y)dy$$

where the first term comes from differentiating the lower limit of integration.

### Computing $c_K$

Differentiating (5.9.3) with respect to $K$ using the Leibniz rule:

$$c_K = e^{-rT}\left[-(K-K)\tilde{p}(0,T,x,K) + \int_K^\infty \frac{\partial}{\partial K}(y-K)\tilde{p}(0,T,x,y)dy\right]$$

The first term vanishes since $K - K = 0$, and $\frac{\partial}{\partial K}(y-K) = -1$:

$$c_K = -e^{-rT}\int_K^\infty \tilde{p}(0,T,x,y)dy = -e^{-rT}\tilde{\mathbb{P}}\{S(T) > K\} \quad \blacksquare$$

### Computing $c_{KK}$

Differentiating $c_K$ with respect to $K$ using the Fundamental Theorem of Calculus:

$$c_{KK} = \frac{\partial}{\partial K}\left[-e^{-rT}\int_K^\infty \tilde{p}(0,T,x,y)dy\right] = e^{-rT}\tilde{p}(0,T,x,K) \quad \blacksquare$$

### Conclusion

$$\tilde{p}(0,T,x,K) = e^{rT}c_{KK}(0,T,x,K) \quad \text{for all } K > 0$$

This formula allows us to **recover the risk-neutral distribution directly from market option prices**. In practice, this is used to construct the implied volatility surface and to price exotic derivatives consistently with market prices.

# Exercise 5.10 (Chooser option).
Consider a model with a unique risk-neutral measure $\tilde{\mathbb{P}}$ and constant interest rate $r$. According to the risk-neutral pricing formula, for $0 \leq t \leq T,$ the price at time $t$ of a European call expiring at time $T$ is 
$$C(t) = \tilde{\mathbb{E}} \left[e^{-r(T - t)} (S(T) - K)^+ |\mathcal{F}(t) \right],$$
where $S(T)$ is the underlying asset price at time $T$ and $K$ is the strike price of the call. Similarly, the price at time $t$ of a European put expiring at time $T$ is 
$$P(t) = \tilde{\mathbb{E}}\left[e^{-r(T - t)} (K - S(T))^+|\mathcal{F}(t) \right].$$
Finaly, because $e^{-rt}S(t)$ is a martingale under $\tilde{\mathbb{P}},$ the price at time $t$ of a forward contract for delivery of one share of stock at time $T$ in exchange for a payment of $K$ at time $T$ is
$$F(t) = \tilde{\mathbb{E}}\left[e^{-r(T - t)} (S(T) - K) |\mathcal{F}(t) \right]$$
$$ = e^{rt} \tilde{\mathbb{E}}[e^{-rT}S(T)|\mathcal{F}(t)] - e^{-r(T - t)} K$$
$$ = S(T) - e^{-r(T - t)} K.$$
Because 
$$(S(T) - K)^+ - (K - S(T))^+ = S(T) - K,$$
we have the put-call parity relationship
$$C(t) - P(t) = \tilde{\mathbb{E}}\left[e^{-r(T - t)} (S(T) - K)^+ - e^{-r(T - t)} (K - S(T))^+ |\mathcal{F}(t) \right].$$
$$ = \tilde{\mathbb{E}}\left[e^{-r(T - t)} (S(T) - K) |\mathcal{F}(t) \right] = F(t).$$
Now consider a date $t_0$ between $0$ and $T$, and consider a chooser option, which gives the right at time $t_0$ to choose to own either the call or the put.

(i) Show that at time $t_0$ the value of the chooser option is 
$$C(t_0) + \max\{0, - F(t_0)\} = C(t_0) + \left(e^{-r(T - t_0)} K - S(t_0)\right)^+.$$
(ii) Show that the value of the chooser option at time $0$ is the sum of the value of a call expiring at time $T$ with strike price $K$ and the value of a put expiring at time $t_0$ with strike price $e^{-r(T - t_0)} K.$

### (i) Value of chooser option at time $t_0$

The payoff of the chooser option at time $t_0$ is $\max\{C(t_0), P(t_0)\}$.

By put-call parity $C(t_0) - P(t_0) = F(t_0)$:

$$\max\{C(t_0), P(t_0)\} = \max\{C(t_0), C(t_0) - F(t_0)\}$$

$$= C(t_0) + \max\{0, -F(t_0)\}$$

Since $F(t_0) = S(t_0) - e^{-r(T-t_0)}K$:

$$\max\{0, -F(t_0)\} = \max\{0, e^{-r(T-t_0)}K - S(t_0)\} = \left(e^{-r(T-t_0)}K - S(t_0)\right)^+$$

Therefore the value of the chooser option at time $t_0$ is:

$$C(t_0) + \left(e^{-r(T-t_0)}K - S(t_0)\right)^+ \quad \blacksquare$$

### (ii) Value of chooser option at time $0$

By the risk-neutral pricing formula:

$$\tilde{\mathbb{E}}\left[e^{-rt_0}\left(C(t_0) + \left(e^{-r(T-t_0)}K - S(t_0)\right)^+\right)\right]$$

$$= \tilde{\mathbb{E}}\left[e^{-rt_0}C(t_0)\right] + \tilde{\mathbb{E}}\left[e^{-rt_0}\left(e^{-r(T-t_0)}K - S(t_0)\right)^+\right]$$

**First term:** Since $e^{-rt}C(t)$ is a martingale under $\tilde{\mathbb{P}}$:

$$\tilde{\mathbb{E}}\left[e^{-rt_0}C(t_0)\right] = C(0)$$

which is the price at time $0$ of a **European call expiring at time $T$ with strike $K$**.

**Second term:**

$$\tilde{\mathbb{E}}\left[e^{-rt_0}\left(e^{-r(T-t_0)}K - S(t_0)\right)^+\right]$$

This is the price at time $0$ of a **European put expiring at time $t_0$ with strike price $e^{-r(T-t_0)}K$**.

**Conclusion:** The value of the chooser option at time $0$ is:

$$C(0) + P^*(0)$$

where $C(0)$ is the price of a call expiring at $T$ with strike $K$, and $P^*(0)$ is the price of a put expiring at $t_0$ with strike $e^{-r(T-t_0)}K$. $\blacksquare$

# Exercise 5.11 (Hedging a cash flow).
Let $W(t), 0 \leq t \leq T,$ be a Brownian motion on a probability space $(\Omega, \mathcal{F}, \mathbb{P}),$ and let $\mathcal{F}(t), 0 \leq t \leq T,$ be the filtration generated by this Brownian motion. Let the mean rate of return $\alpha(t),$ the interest rate $R(t),$ and the volatility $\sigma(t)$ be adapted processes, and assume that $\sigma(t)$ is never zero. Consider a stock price process whose differential is given by (5.2.15):
$$dS(t) = \alpha(t)S(t)dt + \sigma(t)S(t)dW(t), 0 \leq t \leq T.$$
Suppose an agent must pay a cash flow at rate $C(t)$ at each time $t$, where $C(t), 0 \leq t \leq T,$ is an adapted process. If the agent holds $\Delta(t)$ shares of stock at each time $t$, then the differential of her portfolio value will be
$$dX(t) = \Delta(t)dS(t) + R(t) ( X(t) - \Delta(t)S(t))dt - C(t)dt.$$
Show that there is a nonrandom value of $X(0)$ and a portfolio process $\Delta(t), 0 \leq t \leq T,$ such that $X(T) = 0$ almost surely. (Hint: Define the risk-neutral measure and apply Corollary 5.3.2 of the Martingale Representation Theorem to the process
$$\tilde{M}(t) = \tilde{\mathbb{E}}\left[\int_{0}^{T} D(u)C(u)du|\mathcal{F}(t)\right], 0 \leq t \leq T,$$
were $D(T)$ is the discount process.)

**Proof:**

**Step 1: Define the risk-neutral measure**

By Girsanov's theorem, define $\tilde{\mathbb{P}}$ with market price of risk $\Theta(t) = \frac{\alpha(t) - R(t)}{\sigma(t)}$, so that:

$$d\tilde{W}(t) = dW(t) + \Theta(t)dt$$

is a Brownian motion under $\tilde{\mathbb{P}}$.

**Step 2: Define the martingale $\tilde{M}(t)$**

$$\tilde{M}(t) = \tilde{\mathbb{E}}\left[\int_0^T D(u)C(u)du\bigg|\mathcal{F}(t)\right]$$

This is a $\tilde{\mathbb{P}}$-martingale by definition.

**Step 3: Apply the Martingale Representation Theorem**

By Corollary 5.3.2, there exists an adapted process $\tilde{\Gamma}(t)$ such that:

$$\tilde{M}(t) = \tilde{M}(0) + \int_0^t \tilde{\Gamma}(u)d\tilde{W}(u)$$

**Step 4: Set initial capital and portfolio process**

Set:

$$X(0) = \tilde{M}(0) = \tilde{\mathbb{E}}\left[\int_0^T D(u)C(u)du\right]$$

This is the present value of all future cash flows — a nonrandom quantity. Define the portfolio process:

$$\Delta(t) = \frac{\tilde{\Gamma}(t)}{\sigma(t)S(t)D(t)}$$

**Step 5: Compute $d(D(t)X(t))$**

From the portfolio SDE and $dD(t) = -R(t)D(t)dt$:

$$d(D(t)X(t)) = -D(t)C(t)dt + \sigma(t)\Delta(t)S(t)D(t)d\tilde{W}(t) = -D(t)C(t)dt + \tilde{\Gamma}(t)d\tilde{W}(t)$$

**Step 6: Integrate and use $\tilde{M}(T)$**

Integrating from $0$ to $T$:

$$D(T)X(T) = D(0)X(0) - \int_0^T D(u)C(u)du + \int_0^T \tilde{\Gamma}(u)d\tilde{W}(u)$$

$$= \tilde{M}(0) - \int_0^T D(u)C(u)du + \tilde{M}(T) - \tilde{M}(0)$$

$$= \tilde{M}(T) - \int_0^T D(u)C(u)du$$

Since $\tilde{M}(T) = \tilde{\mathbb{E}}\left[\int_0^T D(u)C(u)du\bigg|\mathcal{F}(T)\right] = \int_0^T D(u)C(u)du$:

$$D(T)X(T) = 0$$

Since $D(T) > 0$, we conclude $X(T) = 0$ a.s. $\blacksquare$

**Remark:** The initial capital $X(0) = \tilde{\mathbb{E}}\left[\int_0^T D(u)C(u)du\right]$ is exactly the present value of all future cash flows, which is the economically natural result.

Show that

(iv) Show that if $\rho_{ik}(t)$ is not random (although it may still depend on $t$), then for every $t \geq 0,$
$$\mathbb{E}[B_i(t)B_k(t)] = \tilde{\mathbb{E}}[\tilde{B}_i(t)\tilde{B}_k(t)] = \int_{0}^{t} \rho_{ik}(u)du.$$
Since $B_i(t)$ and $B_k(t)$ both have variance $t$ under $\mathbb{P}$ and $\tilde{B}_i(t)$ and $\tilde{B}_k(t)$ both have variance $t$ under $\tilde{\mathbb{P}},$ this shows that the correlation between $B_i(t)$ and $B_k(t)$ under $\mathbb{P}$ is the same as the correlation between $\tilde{B}_i(t)$ and $\tilde{B}_k(t)$ under $\tilde{\mathbb{P}}$. (v) When $\rho_{ik}(t)$ is random, we can have
$$\mathbb{E}[B_i(t)B_k(t)] \neq \tilde{\mathbb{E}}[\tilde{B}_i(t) \tilde{B}_k(t)].$$
Even though instantaneous correlations are unaffected by a change of measure, correlations can be. To see this, we take $m = d = 2$ and let $W_1(t)$ and $W_2(t)$ be independent Brownian motions under $\mathbb{P}.$ Take $\sigma_{11}(t) = \sigma_{21} = 0, \sigma_{12} = t$, and $\sigma_{22} = \text{sign}(W_1(t)),$ where
$$\text{sign}(x) = \begin{cases} 1 & \text{ if } x \geq 0, \\ -1 & \text{ if } x < 0. \end{cases}$$
Then $\sigma_1(t) = 1, \sigma_2(t) = 1, \rho_{11} = 1, \rho_{22} = 1$ and $\rho_{12}(t) = \rho_{21}(t) = \text{sign}(W_1(t)).$ Take $\mathcal{\Theta}_1(t) = 1$ and $\mathcal{\Theta}_2(t)  = 0$, so that $\tilde{W}_1(t) = W_1(t) + t$ and $\tilde{W}_2(t) = W_2(t).$ Then $\gamma_1(t) = \gamma_2(t) = 0.$ We have
$$B_1(t) = W_2(t),  \quad B_2(t) = \int_{0}^{t} \text{sign}(W_1(u))dW_2(u),$$
$$\tilde{B}_1(t) = B_1(t), \quad \tilde{B}_2(t) = B_2(t).$$
Show that 
$$\mathbb{E}[B_1(t)B_2(t)] \neq \tilde{\mathbb{E}}[\tilde{B}_1(t)\tilde{B}_2(t)] \text{ for all } t > 0.$$

### (i) $\tilde{B}_i(t)$ is a Brownian motion under $\tilde{\mathbb{P}}$

Since $B_i(t) = \sum_{j=1}^d \int_0^t \frac{\sigma_{ij}(u)}{\sigma_i(u)}dW_j(u)$:

$$d\tilde{B}_i(t) = dB_i(t) + \gamma_i(t)dt = \sum_{j=1}^d \frac{\sigma_{ij}(t)}{\sigma_i(t)}dW_j(t) + \sum_{j=1}^d \frac{\sigma_{ij}(t)\Theta_j(t)}{\sigma_i(t)}dt$$

$$= \sum_{j=1}^d \frac{\sigma_{ij}(t)}{\sigma_i(t)}\left[dW_j(t) + \Theta_j(t)dt\right] = \sum_{j=1}^d \frac{\sigma_{ij}(t)}{\sigma_i(t)}d\tilde{W}_j(t)$$

Since $\tilde{W}_j(t)$ are independent Brownian motions under $\tilde{\mathbb{P}}$:

$$d\tilde{B}_i\,d\tilde{B}_i = \sum_{j=1}^d \frac{\sigma_{ij}^2(t)}{\sigma_i^2(t)}dt = dt$$

By Lévy's theorem, $\tilde{B}_i(t)$ is a Brownian motion under $\tilde{\mathbb{P}}$. $\blacksquare$

### (ii) SDE for $S_i(t)$ under $\tilde{\mathbb{P}}$

From (5.4.8) and the market price of risk equations $\alpha_i(t) - R(t) = \sigma_i(t)\gamma_i(t)$:

$$dS_i(t) = \alpha_i(t)S_i(t)dt + \sigma_i(t)S_i(t)dB_i(t)$$

$$= R(t)S_i(t)dt + \sigma_i(t)S_i(t)\left[dB_i(t) + \gamma_i(t)dt\right] = R(t)S_i(t)dt + \sigma_i(t)S_i(t)d\tilde{B}_i(t) \quad \blacksquare$$

### (iii) Instantaneous correlation under $\tilde{\mathbb{P}}$

$$d\tilde{B}_i\,d\tilde{B}_k = \left(\sum_{j=1}^d \frac{\sigma_{ij}}{\sigma_i}d\tilde{W}_j\right)\left(\sum_{l=1}^d \frac{\sigma_{kl}}{\sigma_k}d\tilde{W}_l\right) = \sum_{j=1}^d \frac{\sigma_{ij}\sigma_{kj}}{\sigma_i\sigma_k}dt = \rho_{ik}(t)dt \quad \blacksquare$$

### (iv) Correlation under $\mathbb{P}$ and $\tilde{\mathbb{P}}$ when $\rho_{ik}$ is nonrandom

**Under $\mathbb{P}$:** By Itô's product rule and $dB_i\,dB_k = \rho_{ik}(t)dt$:

$$d(B_i B_k) = B_i\,dB_k + B_k\,dB_i + \rho_{ik}(t)dt$$

Taking expectations (Itô integrals vanish):

$$\mathbb{E}[B_i(t)B_k(t)] = \int_0^t \mathbb{E}[\rho_{ik}(u)]du = \int_0^t \rho_{ik}(u)du$$

since $\rho_{ik}$ is nonrandom.

**Under $\tilde{\mathbb{P}}$:** By the same argument with $\tilde{B}_i$:

$$\tilde{\mathbb{E}}[\tilde{B}_i(t)\tilde{B}_k(t)] = \int_0^t \rho_{ik}(u)du \quad \blacksquare$$

### (v) Correlations can differ under $\mathbb{P}$ and $\tilde{\mathbb{P}}$ when $\rho_{ik}$ is random

With the given setup: $B_1(t) = W_2(t)$, $B_2(t) = \int_0^t \text{sign}(W_1(u))dW_2(u)$, $\tilde{B}_1 = B_1$, $\tilde{B}_2 = B_2$.

**Under $\mathbb{P}$:** By Itô's product rule:

$$d(B_1 B_2) = B_1\,dB_2 + B_2\,dB_1 + dB_1\,dB_2 = W_2\,\text{sign}(W_1)dW_2 + B_2\,dW_2 + \text{sign}(W_1)dt$$

Taking expectations:

$$\mathbb{E}[B_1(t)B_2(t)] = \int_0^t \mathbb{E}[\text{sign}(W_1(u))]du = 0$$

since $W_1(u)$ is symmetric about zero under $\mathbb{P}$.

**Under $\tilde{\mathbb{P}}$:** Since $\tilde{W}_1(t) = W_1(t) + t$, we have $W_1(u) = \tilde{W}_1(u) - u$:

$$\tilde{\mathbb{E}}[\tilde{B}_1(t)\tilde{B}_2(t)] = \int_0^t \tilde{\mathbb{E}}[\text{sign}(\tilde{W}_1(u) - u)]du$$

Since $\tilde{W}_1(u) \sim \mathcal{N}(0,u)$ under $\tilde{\mathbb{P}}$, $\tilde{W}_1(u) - u \sim \mathcal{N}(-u, u)$:

$$\tilde{\mathbb{E}}[\text{sign}(\tilde{W}_1(u)-u)] = 2\tilde{\mathbb{P}}\{\tilde{W}_1(u) \geq u\} - 1 = 2\left(1 - N(\sqrt{u})\right) - 1 < 0$$

Therefore $\tilde{\mathbb{E}}[\tilde{B}_1(t)\tilde{B}_2(t)] = \int_0^t\left(2(1-N(\sqrt{u}))-1\right)du \neq 0$.

**Conclusion:**

$$\mathbb{E}[B_1(t)B_2(t)] = 0 \neq \tilde{\mathbb{E}}[\tilde{B}_1(t)\tilde{B}_2(t)] \quad \blacksquare$$

This shows that although instantaneous correlations $\rho_{ik}(t)$ are invariant under change of measure, the correlations $\mathbb{E}[B_i(t)B_k(t)]$ can differ under $\mathbb{P}$ and $\tilde{\mathbb{P}}$ when $\rho_{ik}$ is random.

# Exercise 5.13.
In part (v) of Exercise 5.12, we saw that when we change measures and change Brownian motions, correlations can change if the instantaneous correlations are random. This exercise shows that a change of measure without a change of Brownian motions can change correlations if the market prices of risk are random.

Let $W_1(t)$ and $W_2(t)$ be independent Brownian motions under a probability measure $\tilde{\mathbb{P}}.$ Take $\mathcal{\Theta}_1(t) = 0$ and $\mathcal{\Theta}_2(t) = W_1(t)$ in the multidimensional Girsanov Theorem, Theorem 5.4.1. Then $\tilde{W}_1(t) = W_1(t)$ and $\tilde{W}_2(t) = W_2(t) + \int_{0}^{t} W_1(u)du.$

(i) Because $\tilde{W}_1(t)$ and $\tilde{W}_2(t)$ are Brownian motions under $\tilde{\mathbb{P}},$ the equation $\tilde{\mathbb{E}}\tilde{W}_1(t) = \tilde{\mathbb{E}}W_2(t) = 0$ must hold for all $t \in [0, T].$ Use this equation to conclude that
$$\tilde{\mathbb{E}}W_1(t) = \tilde{\mathbb{E}}\tilde{W}_2(t) = 0 \text{ for all } t \in [0, T].$$
(ii) From Itô's product rule, we have
$$d(W_1(t)W_2(t)) = W_1(t)dW_2(t) + W_2(t)dW_1(t).$$
Use this equation to show that
$$\tilde{\text{Cov}}[W_1(T),W_2(T)] = \tilde{\mathbb{E}}[W_1(T)W_2(T)] = - \frac{1}{2}T^2$$
This is diffrent from 
$$\text{Cov}[W_1(T), W_2(T)] = \mathbb{E}[W_1(T)W_2(T)] = 0.$$

### (i) $\tilde{\mathbb{E}}W_1(t) = \tilde{\mathbb{E}}\tilde{W}_2(t) = 0$

Since $\tilde{W}_1(t) = W_1(t)$ is a Brownian motion under $\tilde{\mathbb{P}}$:

$$\tilde{\mathbb{E}}W_1(t) = \tilde{\mathbb{E}}\tilde{W}_1(t) = 0 \quad \blacksquare$$

Since $\tilde{W}_2(t) = W_2(t) + \int_0^t W_1(u)du$ is a Brownian motion under $\tilde{\mathbb{P}}$:

$$\tilde{\mathbb{E}}\tilde{W}_2(t) = \tilde{\mathbb{E}}W_2(t) + \int_0^t \tilde{\mathbb{E}}W_1(u)du = 0 + 0 = 0 \quad \blacksquare$$

### (ii) $\tilde{\mathbb{E}}[W_1(T)W_2(T)] = -\frac{1}{2}T^2$

By Itô's product rule (since $dW_1\,dW_2 = 0$):

$$d(W_1 W_2) = W_1\,dW_2 + W_2\,dW_1$$

Integrating and taking $\tilde{\mathbb{E}}$:

$$\tilde{\mathbb{E}}[W_1(T)W_2(T)] = \tilde{\mathbb{E}}\left[\int_0^T W_1(t)dW_2(t)\right] + \underbrace{\tilde{\mathbb{E}}\left[\int_0^T W_2(t)dW_1(t)\right]}_{=0 \text{ (Itô integral w.r.t. }\tilde{W}_1\text{)}}$$

Substituting $W_2(t) = \tilde{W}_2(t) - \int_0^t W_1(u)du$ into the first term:

$$\tilde{\mathbb{E}}\left[\int_0^T W_1(t)dW_2(t)\right] = \underbrace{\tilde{\mathbb{E}}\left[\int_0^T W_1(t)d\tilde{W}_2(t)\right]}_{=0 \text{ (Itô integral w.r.t. }\tilde{W}_2\text{)}} - \tilde{\mathbb{E}}\left[\int_0^T W_1^2(t)dt\right]$$

By Fubini's theorem and $\tilde{\mathbb{E}}[W_1^2(t)] = t$:

$$-\tilde{\mathbb{E}}\left[\int_0^T W_1^2(t)dt\right] = -\int_0^T t\,dt = -\frac{T^2}{2}$$

Therefore:

$$\tilde{\mathbb{E}}[W_1(T)W_2(T)] = -\frac{T^2}{2} \neq 0 = \mathbb{E}[W_1(T)W_2(T)] \quad \blacksquare$$

**Remark:** Although $W_1$ and $W_2$ are independent under $\mathbb{P}$ (so $\text{Cov}_{\mathbb{P}}[W_1(T), W_2(T)] = 0$), they become correlated under $\tilde{\mathbb{P}}$ when the market price of risk $\Theta_2(t) = W_1(t)$ is random. This shows that a change of measure with random market prices of risk can alter correlations even without changing the Brownian motions.

# Exercise 5.14 (Cost of carry).
Consider a commodity whose unit price at time $t$ is $S(t).$ Ownership of a unit of this commodity requires payment at a rate $a$ per unit time (cost of carry) for storage. Note that this payment is per unit of comodity, not a fraction of the price of the commodity. Thus, the value of a portfolio that holds $\Delta(t)$ units of the commodity at time $t$ and also invests in a money market account with constant rate of interest $r$ has differential
$$dX(t) = \Delta(t)dS(t) - a \Delta(t)dt + r(X(t) - \Delta(t)S(t))dt. \tag{5.9.6}$$
As with the dividend-paying stock in Section 5.5, we must choose the risk-neutral measure so that the discounted portfolio value $e^{-rT}X(t)$ is a martingale. We shall assume a constant volatility, so in place of (5.5.6) we have
$$dS(t) = rS(t)dt + \sigma S(t)d\tilde{W}(t) + a dt, \tag{5.9.7}$$
where $\tilde{W}(t)$ is a Broanian motion under the risk-neutral measure $\tilde{\mathbb{P}}.$

(i) Show that when $dS(t)$ is given by (5.9.7), then under $\tilde{\mathbb{P}}$ the discounted porfolio value process $e^{-rt} X(t),$ where $X(t)$ is given by (5.9.6), is a martingale.

(ii) Define 
$$Y(t) = \exp \left\{ \sigma \tilde{W}(t) + \left(r - \frac{1}{2} \sigma^2\right)t\right\}.$$
Verify that, for $0 \leq t \leq T,$
$$dY(t) = rY(t)dt + \sigma Y(t) d\tilde{W}(t).$$
that $e^{-rT}Y(t)$ is a martingale under $\tilde{\mathbb{P}},$ and that 
$$S(t) = S(0)Y(t) + Y(t) \int_{0}^{t} \frac{a}{Y(s)} ds$$
satisfies (5.9.7).

(iii) For $0 \leq t \leq T,$ derive a formula for $\tilde{\mathbb{E}}[S(T)|\mathcal{F}(t)]$ in terms of $S(t)$ by writing
$$\tilde{\mathbb{E}}[S(T)|\mathcal{F}(t)] = S(0)\tilde{\mathbb{E}}[Y(T)|\mathcal{F}(t)] + \tilde{\mathbb{E}}[Y(T) |\mathcal{F}(t)] \int_{0}^{t} \frac{a}{Y(s)} ds + a \int_{t}^{T} \tilde{\mathbb{E}}\left[\frac{Y(T)}{Y(s)}|\mathcal{F}(t)\right] ds.$$
and then simplifying the right-hand side of this equation.

(iv) The process $\tilde{\mathbb{E}}[S(T)|\mathcal{F}(t)]$ is the futures price process for the commodity (i.e., $\text{Fut}_S(t, T) = \tilde{\mathbb{E}}[S(T)|\mathcal{F}(t)]$). This must be a martingale under $\tilde{\mathbb{P}}.$ To check the formula you obtained in (iii), differentiate it and verify that $\tilde{\mathbb{E}}[S(T)|\mathcal{F}(t)]$ is a martingale under $\tilde{\mathbb{P}}.$

(v) Let $0 \leq t \leq T$ be given. Consider a forward contract entered at time $t$ to purchase one unit of the commodity at time $T$ for price $K$ paid at time $T$. The value of this contract at time $t$ when it is entered is
$$\tilde{\mathbb{E}}[e^{-r(T-t)}(S(T) - K)|\mathcal{F}(t)]. \tag{5.9.10}$$
The forward price $\text{For}_s(t, T)$ is the value of $K$ that makes the contract value (5.9.10) equal to zero. Show that $\text{For}_S(t, T) = \text{Fut}_S(t, T).$

(vi) Consider an agent who takes a short position in a forward contract at time zero. This costs nothing and generates no income at time zero. The agent hedges this position by borrowing $S(0)$ from the money market account and purchasing one unit of the commodity, which she holds until time $T$. At time $T$, the agent delivers the commodity under the forward contract and receives the forward price $\text{For}_S(0, T)$ set at time zero. Show that this is exactly what the agent needs to cover her debt to the money market account, which has two parts. First of all, at time zero, the agent borrows $S(0)$ from the money market account in order to purchase the unit of the commodity. Second, between times zero and $T$, the agent pays the cost of carry a per unit time, borrowing from the money market account to finance this. (Hint: The value of the agent's portfolio of commodity and money market account begins at $X(0) = 0$ (one unit of the commodity and a money market position of $-S(0)$) and is governed by (5.9.6) with Ll(t) = 1. Write this equation, determine $d(e^{-rT}X(t)),$ integrate both sides from zero to $T,$ and solve for $X(T).$ You will need the fact that $e^{-rT}(dS(t) - rS(t)dt) = d(e^{-rt} S(t)).$ You should get $X(T) = S(T) - \text{For}_S(0, T).$)

### (i) $e^{-rt}X(t)$ is a martingale under $\tilde{\mathbb{P}}$

Computing $d(e^{-rt}X(t))$:

$$d(e^{-rt}X(t)) = -re^{-rt}X(t)dt + e^{-rt}dX(t)$$

Substituting (5.9.6) and (5.9.7):

$$= e^{-rt}\left[\Delta(t)(rS(t)dt + \sigma S(t)d\tilde{W}(t) + adt) - a\Delta(t)dt + r(X(t) - \Delta(t)S(t))dt - rX(t)dt\right]$$

$$= e^{-rt}\sigma\Delta(t)S(t)d\tilde{W}(t)$$

Since this has no $dt$ term, $e^{-rt}X(t)$ is a $\tilde{\mathbb{P}}$-martingale. $\blacksquare$

### (ii) Properties of $Y(t)$

By Itô's lemma applied to $Y(t) = \exp\{\sigma\tilde{W}(t) + (r-\frac{1}{2}\sigma^2)t\}$:

$$dY(t) = rY(t)dt + \sigma Y(t)d\tilde{W}(t)$$

Since this has no $dt$ term after discounting, $e^{-rt}Y(t)$ is a $\tilde{\mathbb{P}}$-martingale.

**Verifying $S(t) = Y(t)\left(S(0) + \int_0^t\frac{a}{Y(s)}ds\right)$ satisfies (5.9.7):**

Let $g(t) = S(0) + \int_0^t\frac{a}{Y(s)}ds$, so $dg(t) = \frac{a}{Y(t)}dt$. By Itô's product rule (noting $dY\,dg = 0$ since $dg$ has no $dW$ term):

$$dS(t) = Y(t)\,dg(t) + g(t)\,dY(t) = Y(t)\cdot\frac{a}{Y(t)}dt + \frac{S(t)}{Y(t)}\left[rY(t)dt + \sigma Y(t)d\tilde{W}(t)\right]$$

$$= adt + rS(t)dt + \sigma S(t)d\tilde{W}(t) = rS(t)dt + \sigma S(t)d\tilde{W}(t) + adt \quad \blacksquare$$

### (iii) Futures price $\tilde{\mathbb{E}}[S(T)|\mathcal{F}(t)]$

Using $S(T) = S(0)Y(T) + Y(T)\int_0^T\frac{a}{Y(s)}ds$:

$$\tilde{\mathbb{E}}[S(T)|\mathcal{F}(t)] = S(0)\tilde{\mathbb{E}}[Y(T)|\mathcal{F}(t)] + \tilde{\mathbb{E}}[Y(T)|\mathcal{F}(t)]\int_0^t\frac{a}{Y(s)}ds + a\int_t^T\tilde{\mathbb{E}}\left[\frac{Y(T)}{Y(s)}\bigg|\mathcal{F}(t)\right]ds$$

**First and second terms:** Since $e^{-rt}Y(t)$ is a martingale, $\tilde{\mathbb{E}}[Y(T)|\mathcal{F}(t)] = e^{r(T-t)}Y(t)$:

$$e^{r(T-t)}Y(t)\left[S(0) + \int_0^t\frac{a}{Y(s)}ds\right] = e^{r(T-t)}Y(t)\cdot\frac{S(t)}{Y(t)} = e^{r(T-t)}S(t)$$

**Third term:** For $s \geq t$, $\frac{Y(T)}{Y(s)}$ is independent of $\mathcal{F}(t)$, so:

$$\tilde{\mathbb{E}}\left[\frac{Y(T)}{Y(s)}\bigg|\mathcal{F}(t)\right] = e^{r(T-s)}$$

Therefore:

$$a\int_t^T e^{r(T-s)}ds = \frac{a}{r}\left(e^{r(T-t)}-1\right)$$

**Combining:**

$$\tilde{\mathbb{E}}[S(T)|\mathcal{F}(t)] = e^{r(T-t)}S(t) + \frac{a}{r}\left(e^{r(T-t)}-1\right) = e^{r(T-t)}\left(S(t) + \frac{a}{r}\right) - \frac{a}{r} \quad \blacksquare$$

### (iv) $\text{Fut}_S(t,T)$ is a $\tilde{\mathbb{P}}$-martingale

Differentiating $\text{Fut}_S(t,T) = e^{r(T-t)}\left(S(t)+\frac{a}{r}\right) - \frac{a}{r}$:

$$d\,\text{Fut}_S = -re^{r(T-t)}\left(S(t)+\frac{a}{r}\right)dt + e^{r(T-t)}dS(t)$$

Substituting (5.9.7):

$$= e^{r(T-t)}\sigma S(t)d\tilde{W}(t)$$

Since there is no $dt$ term, $\text{Fut}_S(t,T)$ is a $\tilde{\mathbb{P}}$-martingale. $\blacksquare$

### (v) Forward price equals futures price

The forward price $\text{For}_S(t,T)$ satisfies:

$$\tilde{\mathbb{E}}[e^{-r(T-t)}(S(T)-K)|\mathcal{F}(t)] = 0$$

$$\implies K = \tilde{\mathbb{E}}[S(T)|\mathcal{F}(t)] = \text{Fut}_S(t,T)$$

Therefore $\text{For}_S(t,T) = \text{Fut}_S(t,T)$. $\blacksquare$

### (vi) Hedging the short forward position

With $\Delta(t) = 1$ and $X(0) = 0$, from (5.9.6):

$$d(e^{-rt}X(t)) = e^{-rt}(dS(t) - rS(t)dt) - ae^{-rt}dt = d(e^{-rt}S(t)) - ae^{-rt}dt$$

Integrating from $0$ to $T$:

$$e^{-rT}X(T) - X(0) = e^{-rT}S(T) - S(0) - \frac{a}{r}(1-e^{-rT})$$

Therefore:

$$X(T) = S(T) - e^{rT}S(0) - \frac{a}{r}(e^{rT}-1) = S(T) - e^{rT}\left(S(0)+\frac{a}{r}\right) + \frac{a}{r} = S(T) - \text{For}_S(0,T) \quad \blacksquare$$

At time $T$, the agent delivers the commodity (worth $S(T)$) and receives $\text{For}_S(0,T)$, exactly covering the money market debt, leaving $X(T) = 0$.
