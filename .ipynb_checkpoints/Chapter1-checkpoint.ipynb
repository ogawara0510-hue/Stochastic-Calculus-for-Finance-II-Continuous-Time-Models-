# Exercise 1.1

**Lemma:** $\mathbb{P}(\emptyset) = 0$

**Proof:** Since $\mathcal{F}$ is a $\sigma$-algebra, $\emptyset = \Omega^c \in \mathcal{F}$. Since $\Omega$ and $\emptyset$ are disjoint and $\Omega \cup \emptyset = \Omega$, by finite additivity:

$$\mathbb{P}(\Omega) = \mathbb{P}(\Omega) + \mathbb{P}(\emptyset) = 1 \implies \mathbb{P}(\emptyset) = 0 \quad \blacksquare$$

**Lemma (Finite Additivity):** Let $A_1, \ldots, A_n \in \mathcal{F}$ be pairwise disjoint. Then:

$$\mathbb{P}\left(\bigcup_{i=1}^{n} A_i\right) = \sum_{i=1}^{n} \mathbb{P}(A_i)$$

**Proof:** Set $A_i = \emptyset$ for $i = n+1, n+2, \ldots$ By countable additivity:

$$\mathbb{P}\left(\bigcup_{i=1}^{\infty} A_i\right) = \sum_{i=1}^{\infty} \mathbb{P}(A_i) = \sum_{i=1}^{n} \mathbb{P}(A_i) + \sum_{i=n+1}^{\infty} \mathbb{P}(\emptyset) = \sum_{i=1}^{n} \mathbb{P}(A_i) \quad \blacksquare$$

### (i)

**Proof:** Decompose $B = A \cup (B \setminus A)$ where $A, B \setminus A \in \mathcal{F}$ are disjoint. By finite additivity:

$$\mathbb{P}(B) = \mathbb{P}(A) + \mathbb{P}(B \setminus A)$$

Since $\mathbb{P}(B \setminus A) \geq 0$, we conclude $\mathbb{P}(A) \leq \mathbb{P}(B)$. $\blacksquare$

### (ii)

**Proof:** Since $A \subset A_n$ for all $n$, by (i):

$$0 \leq \mathbb{P}(A) \leq \mathbb{P}(A_n) \quad \forall n$$

Taking $n \to \infty$ and applying the squeeze theorem:

$$0 \leq \mathbb{P}(A) \leq \lim_{n \to \infty} \mathbb{P}(A_n) = 0$$

Therefore $\mathbb{P}(A) = 0$. $\blacksquare$

# Exercise 1.2.
The infinite coin-toss space $\Omega_{\infty}$ of Example 1.1.4 is uncountable infinite. In other words, we cannot list all its elements in a sequence.
To see that this is impossible, suppose there were such a sequential list of all elements of $\Omega_{\infty}$:
$$\omega^{(1)} = \omega_1^{(1)} \omega_2^{(1)} \omega_3^{(1)} \omega_4^{(1)} \ldots,$$
$$\omega^{(2)} = \omega_1^{(2)} \omega_2^{(2)} \omega_3^{(2)} \omega_4^{(2)} \ldots,$$
$$\omega^{(3)} = \omega_1^{(3)} \omega_2^{(3)} \omega_3^{(3)} \omega_4^{(3)} \ldots,$$
$$\vdots$$
An element that does not appear in this list is the sequence whose first component is $H$ if $\omega_1^{(1)}$ is $T$ and is $T$ if $\omega_1^{(1)}$ is $H$, whose second component is $H$ if $\omega_2^{(2)}$ is $T$ and is $T$ if $\omega_2^{(2)}$ is $H$, whose third component is $H$ if $\omega_3^{(3)}$ is $T$ and is $T$ if $\omega_3^{(3)}$ is $H$, etc. Thus, the list does not include every element of $\Omega_{\infty}$.
Now consider the set of sequences of coin tosses in which the outcome on each even-numberd toss matches the outcome of the toss preceding it, i.e.,
    $$A = \{\omega = \omega_1 \omega_2 \omega_3 \omega_4 \omega_5 \ldots; \omega_1 = \omega_2, \omega_3 = \omega_4, \ldots\}.$$
(i) Show that $A$ is uncountably infinite.
(ii)Show that, when $0<p<1$, we have $\mathbb{P}(A) = 0$.
Uncountably infinite sets can have any probability between zero and one, including zero and one. The uncountability of the set does not help determine its probability.

### (i) $A$ is uncountably infinite

**Proof:** Construct an injection $f: \Omega_\infty \to A$ as follows. For any $\omega = \omega_1\omega_2\omega_3\ldots \in \Omega_\infty$, define:

$$f(\omega) = \omega_1\omega_1\omega_2\omega_2\omega_3\omega_3\ldots$$

This map is:
- **Well-defined:** by construction, $f(\omega)$ satisfies $\omega_{2k-1} = \omega_{2k}$ for all $k$, so $f(\omega) \in A$.
- **Injective:** if $f(\omega) = f(\omega')$, then $\omega_n = \omega_n'$ for all $n$, so $\omega = \omega'$.

Since $\Omega_\infty$ is uncountably infinite and $f$ is an injection from $\Omega_\infty$ into $A$, it follows that $A$ is uncountably infinite. $\blacksquare$

### (ii) $\mathbb{P}(A) = 0$ when $0 < p < 1$

**Proof:** Define $A_n$ to be the event that the first $n$ pairs match:

$$A_n = \{\omega \in \Omega_\infty ; \omega_1 = \omega_2, \omega_3 = \omega_4, \ldots, \omega_{2n-1} = \omega_{2n}\}$$

Since each pair $(\omega_{2k-1}, \omega_{2k})$ is independent, the probability that a single pair matches is:

$$\mathbb{P}(\omega_{2k-1} = \omega_{2k}) = p^2 + (1-p)^2$$

By independence across pairs:

$$\mathbb{P}(A_n) = (p^2 + (1-p)^2)^n$$

Since $0 < p < 1$, by AM-GM inequality:

$$p^2 + (1-p)^2 < (p + (1-p))^2 = 1$$

Note that $A \subset A_n$ for all $n$, so by Exercise 1.1(ii):

$$0 \leq \mathbb{P}(A) \leq \mathbb{P}(A_n) = (p^2 + (1-p)^2)^n \to 0 \quad (n \to \infty)$$

Therefore $\mathbb{P}(A) = 0$. $\blacksquare$

# Exercise 1.3.
Consider the set function $\mathbb{P}$ defined for every subset of $[0,1]$ by the formula that $\mathbb{P}(A) = 0$ if $A$ is a finite set and $\mathbb{P}(A) = {\infty}$ if $A$ is an infinite set. Show that $\mathbb{P}$ satisfies (1.1.3)-(1.1.5), but $\mathbb{P}$ does not have the countable additivity property (1.1.2). We see then the finite additivity property (1.1.5) does not imply the countable additivity property (1.1.2).
The contavle additivity: Whenever $A_1, A_2, \ldots$ is a sequence of disjoint sets in $\mathcal{F}$, then
$$\mathbb{P}(\bigcup_{n=1}^{\infty} A_n) = \sum_{n=1}^{\infty} \mathbb{P}(A_n). \tag{1.1.2}$$
The probability of $\emptyset$
$$\mathbb{P}(\emptyset) = 0. \tag{1.1.3}$$
If A and B are disjoint sets in $\mathcal{F}$,
$$\mathbb{P}(A \cup B) = \mathbb{P}(A) + \mathbb{P}(B). \tag{1.1.4}$$
The finitie additivity: if $A_1, A_2, \ldots, A_N$ are finitely many disjoints sets $\mathcal{F}$, then
$$\mathbb{P}(\bigcup_{n=1}^{N} A_n) = \sum_{n=1}^{N} \mathbb{P}(A_n). \tag{1.1.5}$$

### (1.1.3): $\mathbb{P}(\emptyset) = 0$

**Proof:** $\emptyset$ is a finite set (with zero elements), so by definition $\mathbb{P}(\emptyset) = 0$. $\blacksquare$

### (1.1.4): $\mathbb{P}(A \cup B) = \mathbb{P}(A) + \mathbb{P}(B)$

**Proof:** Let $A, B \in \mathcal{F}$ be disjoint. We consider all cases:

| $A$ | $B$ | $A \cup B$ | LHS | RHS |
|---|---|---|---|---|
| finite | finite | finite | $0$ | $0 + 0 = 0$ |
| finite | infinite | infinite | $\infty$ | $0 + \infty = \infty$ |
| infinite | finite | infinite | $\infty$ | $\infty + 0 = \infty$ |
| infinite | infinite | infinite | $\infty$ | $\infty + \infty = \infty$ |

In all cases, LHS $=$ RHS. $\blacksquare$

### (1.1.5): Finite Additivity

**Proof:** By induction on $N$. The base case $N = 2$ is exactly (1.1.4). Assume the result holds for $N = k$. Then:

$$\mathbb{P}\left(\bigcup_{n=1}^{k+1} A_n\right) = \mathbb{P}\left(\bigcup_{n=1}^{k} A_n\right) + \mathbb{P}(A_{k+1}) = \sum_{n=1}^{k} \mathbb{P}(A_n) + \mathbb{P}(A_{k+1}) = \sum_{n=1}^{k+1} \mathbb{P}(A_n)$$

Thus the result holds for $N = k+1$, completing the induction. $\blacksquare$

### (1.1.2) does not hold: Counterexample

Let $\{x_1, x_2, \ldots\}$ be a countably infinite subset of $[0,1]$ and define $A_n = \{x_n\}$ for each $n$. The sets $A_n$ are pairwise disjoint and each is finite, so:

$$\sum_{n=1}^{\infty} \mathbb{P}(A_n) = \sum_{n=1}^{\infty} 0 = 0$$

However, $\bigcup_{n=1}^{\infty} A_n$ is infinite, so:

$$\mathbb{P}\left(\bigcup_{n=1}^{\infty} A_n\right) = \infty \neq 0$$

Therefore countable additivity (1.1.2) fails. $\blacksquare$

### Conclusion
This example shows that finite additivity (1.1.5) does not imply countable additivity (1.1.2).

# Exercise 1.4. 
(i) Construct a standard normal random variable $Z$ on the probability space $(\Omega_{\infty}, \mathcal{F}_{\infty}, \mathbb{P})$ of Example 1.1.4 under the assumption that the probability for head is $p = \frac{1}{2}$. 
(ii) Define a sequence of random variables $\{Z_n\}_{n=1}^{\infty}$ on $\Omega_{\infty}$ such that
$$\lim_{n \to \infty} Z_n (\omega) = Z(\omega) \text{ for every } \omega \in \Omega_{\infty}$$
and, for each $n, Z_n$ depends only on the first $n$ coin tosses. (This gives us a procedure for approximating a standard normal random variable by random variables generated by a finite number of coin tosses, a useful algorithm for Monte Carlo simulation.)

### (i) Construction of a Standard Normal Random Variable

Define the outcome of the $k$-th coin toss as:

$$X_k(\omega) = \begin{cases} 1 & \omega_k = H \\ -1 & \omega_k = T \end{cases}$$

Since $p = \frac{1}{2}$, the $X_k$ are i.i.d. with:

$$\mathbb{E}[X_k] = 0, \quad \text{Var}(X_k) = 1$$

Define $U(\omega)$ as the binary expansion of a uniform random variable on $[0,1]$:

$$U(\omega) = \sum_{k=1}^{\infty} \frac{X_k(\omega) + 1}{2^{k+1}}$$

Since each coin toss contributes an independent binary digit, $U(\omega)$ is uniformly distributed on $[0,1]$. By the inverse transform method, if $\Phi^{-1}$ denotes the inverse CDF of the standard normal distribution, then:

$$Z(\omega) = \Phi^{-1}(U(\omega)) = \Phi^{-1}\left(\sum_{k=1}^{\infty} \frac{X_k(\omega) + 1}{2^{k+1}}\right)$$

is a standard normal random variable on $(\Omega_\infty, \mathcal{F}_\infty, \mathbb{P})$. $\blacksquare$

### (ii) Approximating Sequence

For each $n$, define:

$$Z_n(\omega) = \Phi^{-1}\left(\sum_{k=1}^{n} \frac{X_k(\omega) + 1}{2^{k+1}}\right)$$

Then:
- $Z_n$ depends only on the first $n$ coin tosses. 
- Since $U_n(\omega) = \sum_{k=1}^{n} \frac{X_k(\omega)+1}{2^{k+1}} \to U(\omega)$ for every $\omega \in \Omega_\infty$, and $\Phi^{-1}$ is continuous, we have:

$$\lim_{n \to \infty} Z_n(\omega) = \Phi^{-1}(U(\omega)) = Z(\omega) \quad \text{for every } \omega \in \Omega_\infty \quad \blacksquare$$

### Code: Monte Carlo Simulation


```python
def simulate_Z_n(n, num_simulations=10000):
    # Coin tosses: H=1, T=0
    tosses = np.random.randint(0, 2, size=(num_simulations, n))
    
    # U_n = sum_{k=1}^{n} toss_k / 2^k
    weights = np.array([1 / 2**k for k in range(1, n+1)])
    U_n = tosses @ weights
    
    # クリップして0と1を避ける
    U_n = np.clip(U_n, 1e-10, 1 - 1e-10)
    
    # Z_n = Phi^{-1}(U_n)
    Z_n = norm.ppf(U_n)
    return Z_n

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, n in zip(axes, [5, 20, 100]):
    Z_n = simulate_Z_n(n)
    ax.hist(Z_n, bins=50, density=True, alpha=0.7, label=f'$Z_{{{n}}}$')
    x = np.linspace(-4, 4, 100)
    ax.plot(x, norm.pdf(x), 'r-', label='$\\mathcal{N}(0,1)$')
    ax.set_title(f'$n = {n}$')
    ax.legend()

plt.suptitle('Approximation of Standard Normal via Coin Tosses')
plt.tight_layout()
plt.show()
```

## Exercise 1.5.
When dealing with double Lebesgue integrals, just as with double Reimann integrals, the order of integration can be reversed. The only assumption required is that the function being integrated be either nonnnegative or integrable. Here is an application of this fact.
Let X be a nonnegative random variable with cumulative distribution function $F(x) = \mathbb{P}\{X \leq x \}.$ Show that
$$\mathbb{E}X = \int_{0}^{\infty} (1 - F(x)) dx$$
by showing that 
$$\int_{\Omega} \int_{0}^{\infty} \mathbb{I}_{[0, X(\omega))}(x)dxd\mathbb{P}(\omega)$$
is equal to both $\mathbb{E}X$ and $\int_{0}^{\infty}(1-F(x))dx$.



### Step 1: Evaluate the inner integral

Since $\mathbb{I}_{[0, X(\omega))}(x) = 1$ if and only if $0 \leq x < X(\omega)$:

$$\int_0^{\infty} \mathbb{I}_{[0, X(\omega))}(x) dx = \int_0^{X(\omega)} 1 \, dx = X(\omega)$$

### Step 2: Equality with $\mathbb{E}X$

$$\int_\Omega \int_0^{\infty} \mathbb{I}_{[0, X(\omega))}(x) dx \, d\mathbb{P}(\omega) = \int_\Omega X(\omega) \, d\mathbb{P}(\omega) = \mathbb{E}X$$

### Step 3: Exchange order of integration

Since $\mathbb{I}_{[0,X(\omega))}(x) \geq 0$, by Fubini's theorem we may exchange the order of integration:

$$\int_\Omega \int_0^{\infty} \mathbb{I}_{[0, X(\omega))}(x) dx \, d\mathbb{P}(\omega) = \int_0^{\infty} \int_\Omega \mathbb{I}_{[0, X(\omega))}(x) \, d\mathbb{P}(\omega) \, dx$$

### Step 4: Evaluate the inner integral

Fix $x$. Since $\mathbb{I}_{[0, X(\omega))}(x) = 1$ if and only if $X(\omega) > x$:

$$\int_\Omega \mathbb{I}_{[0, X(\omega))}(x) \, d\mathbb{P}(\omega) = \mathbb{P}(X > x) = 1 - F(x)$$

### Step 5: Equality with $\int_0^\infty (1-F(x))dx$

$$\int_0^{\infty} \int_\Omega \mathbb{I}_{[0, X(\omega))}(x) \, d\mathbb{P}(\omega) \, dx = \int_0^{\infty} (1 - F(x)) \, dx$$

### Conclusion

Combining Step 2 and Step 5:

$$\mathbb{E}X = \int_\Omega \int_0^\infty \mathbb{I}_{[0,X(\omega))}(x) \, dx \, d\mathbb{P}(\omega) = \int_0^\infty (1-F(x)) \, dx \quad \blacksquare$$

# Exercise 1.6.
Let $u$ be a fixed number in $\mathbb{R}$, and define the convex function $\varphi(x) = e^{ux}$ for all $x \in \mathbb{R}$. Let $X$ be a normal random variable with mean $\mu = \mathbb{E}X$ and standard deviation $\sigma = [\mathbb{E}(X - \mu)^2]^\frac{1}{2}$, i.e., with density
$$f(x) = \frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{(x-\mu)^2}{2\sigma^2}}$$
(i) Verify that
$$\mathbb{E}e^{uX} = e^{u\mu + \frac{1}{2}u^2 \mu^2}.$$
(ii) Verify that Jensen's inequality holds(as it must):
$$\mathbb{E}\varphi(x) \geq \varphi(\mathbb{E}X).$$

### (i) Moment Generating Function of Normal Distribution

$$\mathbb{E}e^{uX} = \int_{-\infty}^{\infty} e^{ux} \cdot \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{(x-\mu)^2}{2\sigma^2}} dx$$

Combining the exponents:

$$ux - \frac{(x-\mu)^2}{2\sigma^2} = -\frac{1}{2\sigma^2}\left[x^2 - 2(\mu + u\sigma^2)x + \mu^2\right]$$

Completing the square:

$$= -\frac{(x - (\mu + u\sigma^2))^2}{2\sigma^2} + u\mu + \frac{1}{2}u^2\sigma^2$$

Substituting back:

$$\mathbb{E}e^{uX} = e^{u\mu + \frac{1}{2}u^2\sigma^2} \cdot \int_{-\infty}^{\infty} \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{(x-(\mu+u\sigma^2))^2}{2\sigma^2}} dx$$

The integral is the total probability of a normal distribution, which equals 1. Therefore:

$$\mathbb{E}e^{uX} = e^{u\mu + \frac{1}{2}u^2\sigma^2} \quad \blacksquare$$

### (ii) Jensen's Inequality

**Theorem (Jensen's Inequality):** If $\varphi$ is convex, then $\mathbb{E}[\varphi(X)] \geq \varphi(\mathbb{E}[X])$.

**Proof:** Since $\varphi(x) = e^{ux}$ is convex ($\varphi''(x) = u^2 e^{ux} \geq 0$), by the supporting hyperplane property of convex functions, for any $x_0$:

$$\varphi(x) \geq \varphi(x_0) + \varphi'(x_0)(x - x_0) \quad \forall x$$

Setting $x_0 = \mu = \mathbb{E}[X]$:

$$\varphi(X(\omega)) \geq \varphi(\mu) + \varphi'(\mu)(X(\omega) - \mu)$$

Taking expectations of both sides:

$$\mathbb{E}[\varphi(X)] \geq \varphi(\mu) + \varphi'(\mu)(\mathbb{E}[X] - \mu)$$

Since $\mathbb{E}[X] = \mu$, the second term vanishes:

$$\mathbb{E}[\varphi(X)] \geq \varphi(\mu) = \varphi(\mathbb{E}[X]) \quad \blacksquare$$

**Verification:** From (i), $\mathbb{E}e^{uX} = e^{u\mu + \frac{1}{2}u^2\sigma^2}$, and $\varphi(\mathbb{E}[X]) = e^{u\mu}$. Since $e^{\frac{1}{2}u^2\sigma^2} \geq 1$, we confirm:

$$\mathbb{E}e^{uX} = e^{u\mu + \frac{1}{2}u^2\sigma^2} \geq e^{u\mu} = \varphi(\mathbb{E}[X]) \quad \blacksquare$$

# Exercise 1.7.
For each positive integer $n$, define $f_n$ to be the normal density with mean zero and variance $n$, i.e.,
$$f_n(x) = \frac{1}{\sqrt{2n\pi}} e^{-\frac{x^2}{2n}}.$$
(i) What is the function $f(x) = \lim_{n \to \infty} f_n(x)$ ?
(ii) What is $\lim_{n \to \infty} \int_{-\infty}^{\infty} f_n(x) dx$ ?
(iii) Note that 
$$\lim_{n \to \infty} \int_{-\infty}^{\infty} f_n(x) dx \neq \int_{-\infty}^{\infty} f(x) dx.$$
Explain why this does not violate the Monotone Convergence Theorem, Theorem 1.4.5.

### (i) Pointwise limit of $f_n$

As $n \to \infty$:
- $\frac{1}{\sqrt{2n\pi}} \to 0$
- $e^{-\frac{x^2}{2n}} \to 1$

Therefore:

$$f(x) = \lim_{n \to \infty} f_n(x) = 0 \quad \forall x \in \mathbb{R}$$

### (ii) Limit of the integrals

Since $f_n$ is a normal density for every $n$:

$$\int_{-\infty}^{\infty} f_n(x) dx = 1 \quad \forall n$$

Therefore:

$$\lim_{n \to \infty} \int_{-\infty}^{\infty} f_n(x) dx = 1$$

### (iii) Why this does not violate the Monotone Convergence Theorem

The Monotone Convergence Theorem requires $f_n$ to be monotone increasing, i.e.:

$$0 \leq f_1(x) \leq f_2(x) \leq \cdots$$

We determine where $f_n(x) \geq f_{n+1}(x)$. Taking the log of both sides of $f_n(x) \geq f_{n+1}(x)$ and rearranging:

$$x^2 \leq n(n+1)\ln\frac{n+1}{n}$$

Therefore:

- For $|x| \leq \sqrt{n(n+1)\ln\frac{n+1}{n}}$: $f_n(x) \geq f_{n+1}(x)$ (monotone **decreasing** near the center)
- For $|x| > \sqrt{n(n+1)\ln\frac{n+1}{n}}$: $f_n(x) < f_{n+1}(x)$ (monotone **increasing** in the tails)

Since $f_n$ is neither monotone increasing nor monotone decreasing in any region, the hypothesis of the Monotone Convergence Theorem is not satisfied, and therefore no contradiction arises. $\blacksquare$

# Exercise 1.8 (Moment-generating function).
Let X be a nonnegative random variable, and assume that
$$\varphi(t) = \mathbb{E}e^{tX}$$
is finite for every $t \in \mathbb{R}$. Assume further that $\mathbb{E}[Xe^{tX}] < \infty$ for every $t \in \mathbb{R}$. The purpose of this exercise is to show that $\varphi'(t) = \mathbb{E}[Xe^{tX}]$ and, in particular, $\varphi'(0)=\mathbb{E} X.$

We recall the definition of derivative:
$$\varphi'(t) = \lim_{s \to t} \frac{\varphi(t) - \varphi(s)}{t - s} = \lim_{s \to t} \frac{\mathbb{E}e^{tX} - \mathbb{E}e^{sX}}{t - s} = \lim_{s \to t}\mathbb{E}[\frac{e^{tX} - e^{sX}}{t - s}].$$
The limit above is taken over a continuous variable $s$, but we can choose a sequence of numbers $\{s_n\}_{n=1}^{\infty}$ converging to $t$ and compute
$$\lim_{s_n \to t} \mathbb{E}[\frac{e^{tX}-e^{s_n X}}{t - s_n}],$$
where now we are taking a limit of the expectations of the sequence of random variables
$$Y_n = \frac{e^{tX}-e^{s_n X}}{t - s_n}.$$
If this limit turns out to be the same, regardless of how we choose the sequence ${s_n}_{n=1}^{\infty}$ that converges to $t$, then this limit is also the same as $\lim_{s \to t} \mathbb{E}[\frac{e^{tX} - e^{sX}}{t - s}]$ and is $\varphi'(t)$.

The Mean Value Theorem from calculus states that if $f(t)$ is a diffrentiable function, then for any two numbers $s$ and $t$, there is a number $\theta$ between $s$ and $t$ such that
$$f(t) - f(s) = f'(\theta)(t - s).$$
If we fix $\omega \in \Omega$ and define $f(t) = e^{tX(\omega)}$, then this becomes
$$e^{tX(\omega)} - e^{sX(\omega)} = (t - s)X(\omega)e^{\theta(\omega) X(\omega)}, \tag{1.9.1}$$
where $\theta(\omega)$ is a number depending on $\omega$(i.e., a random variable lying between $t$ and $s$).

(i) Use the Dominated Convergence Theorem and equation (1.9.1) to show that 
$$\lim_{n \to \infty} \mathbb{E}Y_n = \mathbb{E}[\lim_{n \to \infty} Y_n] = \mathbb{E}[Xe^{tX}].$$
This establishes the desired formula $\varphi'(t) = \mathbb{E}[Xe^{tX}].$

(ii) Suppose the random variable $X$ can take both positive and negative values and $\mathbb{E}e^{tX} < \infty$ and $\mathbb{E}[|X|e^{tX}] < \infty$ for every $t \in \mathbb{R}$. Show that onve again $\varphi'(t)=\mathbb{E}[Xe^{tX}]$.

### (i) Applying the Dominated Convergence Theorem

**Step 1: Pointwise limit**

For fixed $\omega$:

$$\lim_{n \to \infty} Y_n(\omega) = \lim_{n \to \infty} \frac{e^{tX(\omega)} - e^{s_n X(\omega)}}{t - s_n} = \frac{d}{dt}e^{tX(\omega)} = X(\omega)e^{tX(\omega)}$$

**Step 2: Construct a dominating function**

By the Mean Value Theorem, there exists $\theta_n(\omega)$ between $s_n$ and $t$ such that:

$$Y_n(\omega) = X(\omega)e^{\theta_n(\omega)X(\omega)}$$

For $n$ sufficiently large, $\theta_n(\omega) \in [t-1, t+1]$, so:

$$|Y_n(\omega)| = |X(\omega)|e^{\theta_n(\omega)X(\omega)} \leq |X(\omega)| \cdot \max(e^{(t+1)X(\omega)}, e^{(t-1)X(\omega)}) =: Z(\omega)$$

Since $\mathbb{E}[Xe^{tX}] < \infty$ for every $t \in \mathbb{R}$ by assumption:

$$\mathbb{E}[Z] \leq \mathbb{E}[|X|e^{(t+1)X}] + \mathbb{E}[|X|e^{(t-1)X}] < \infty$$

so $Z$ is an integrable dominating function.

**Step 3: Apply the Dominated Convergence Theorem**

Since $|Y_n| \leq Z$ and $\mathbb{E}[Z] < \infty$, by the Dominated Convergence Theorem:

$$\lim_{n \to \infty} \mathbb{E}[Y_n] = \mathbb{E}\left[\lim_{n \to \infty} Y_n\right] = \mathbb{E}[Xe^{tX}] \quad \blacksquare$$

This establishes $\varphi'(t) = \mathbb{E}[Xe^{tX}]$, and in particular $\varphi'(0) = \mathbb{E}X$.

### (ii) $X$ takes both positive and negative values

Decompose $X = X^+ - X^-$ where $X^+ = \max(X, 0)$ and $X^- = \max(-X, 0)$.

By the Mean Value Theorem, there exists $\theta_n(\omega)$ between $s_n$ and $t$ such that:

$$|Y_n(\omega)| = |X(\omega)|e^{\theta_n(\omega)X(\omega)} \leq |X(\omega)| \cdot \max(e^{(t+1)X(\omega)}, e^{(t-1)X(\omega)}) =: Z(\omega)$$

This bound holds regardless of the sign of $X$. By assumption $\mathbb{E}[|X|e^{tX}] < \infty$ for every $t \in \mathbb{R}$, so:

$$\mathbb{E}[Z] \leq \mathbb{E}[|X|e^{(t+1)X}] + \mathbb{E}[|X|e^{(t-1)X}] < \infty$$

Since $Z$ is an integrable dominating function, the Dominated Convergence Theorem applies exactly as in (i):

$$\varphi'(t) = \mathbb{E}[Xe^{tX}] \quad \blacksquare$$

# Exercise 1.9.
Suppose $X$ is a random variable on some probability space $(\Omega, \mathcal{F}, \mathbb{P}), A$ is a set in $\mathcal{F}$, and for every Borel subset $B$ of $\mathbb{R}$, we have
$$\int_{A}\mathbb{I}_{B}(X(\omega))d\mathbb{P}(\omega) = \mathbb{P}(A)\cdot \mathbb{P}\{X \in B\}.$$
Then we say that $X$ is independent of the event $A$.
Show that if X is independent on an event $A$, then
$$\int_{A}g(X(\omega))d\mathbb{P}(\omega) = \mathbb{P}(A) \cdot \mathbb{E}(X)$$
for every nonnegative, Borel-measurable function $g$.

### Step 1: Indicator functions

By assumption, for every Borel set $B$:

$$\int_A \mathbb{I}_B(X(\omega)) d\mathbb{P}(\omega) = \mathbb{P}(A) \cdot \mathbb{P}(X \in B) = \mathbb{P}(A) \cdot \mathbb{E}[\mathbb{I}_B(X)]$$

So the result holds for $g = \mathbb{I}_B$.

### Step 2: Simple functions

Let $g = \sum_{k=1}^{n} c_k \mathbb{I}_{B_k}$ where $c_k \geq 0$ and $B_k$ are Borel sets. By linearity of integration and Step 1:

$$\int_A g(X(\omega)) d\mathbb{P}(\omega) = \sum_{k=1}^{n} c_k \int_A \mathbb{I}_{B_k}(X(\omega)) d\mathbb{P}(\omega) = \sum_{k=1}^{n} c_k \cdot \mathbb{P}(A) \cdot \mathbb{E}[\mathbb{I}_{B_k}(X)] = \mathbb{P}(A) \cdot \mathbb{E}[g(X)]$$

### Step 3: Nonnegative Borel-measurable functions

For any nonnegative Borel-measurable function $g$, define the Borel sets:

$$B_{k,n} = \left\{x \,;\, \frac{k}{2^n} \leq g(x) < \frac{k+1}{2^n}\right\}, \quad k = 0, 1, \ldots, 4^n - 1$$

and the simple functions:

$$g_n(x) = \sum_{k=0}^{4^n - 1} \frac{k}{2^n} \mathbb{I}_{B_{k,n}}(x)$$

The partition points for $n+1$ include all partition points for $n$ plus their midpoints, so:

$$0 \leq g_1 \leq g_2 \leq \cdots \leq g$$

For every $x \in \mathbb{R}$, the approximation error satisfies $0 \leq g(x) - g_n(x) \leq \frac{1}{2^n}$, so:

$$\lim_{n \to \infty} g_n(x) = g(x) \quad \forall x \in \mathbb{R}$$

By Step 2, for each $n$:

$$\int_A g_n(X(\omega)) d\mathbb{P}(\omega) = \mathbb{P}(A) \cdot \mathbb{E}[g_n(X)]$$

Since $0 \leq g_n \nearrow g$, the Monotone Convergence Theorem applies to both sides:

$$\int_A g(X(\omega)) d\mathbb{P}(\omega) = \lim_{n \to \infty} \int_A g_n(X(\omega)) d\mathbb{P}(\omega) = \lim_{n \to \infty} \mathbb{P}(A) \cdot \mathbb{E}[g_n(X)] = \mathbb{P}(A) \cdot \mathbb{E}[g(X)] \quad \blacksquare$$

# Exercise 1.10.
Let $\mathbb{P}$ be the uniform (Lebesgue) measure on $\Omega = [0,1].$ Define 
$$Z(\omega) = \begin {cases} 0 & \text{ if } 0 \leq \omega \leq \frac{1}{2}, \\ 2 & \text{ if } \frac{1}{2} \leq \omega \leq 1. \end {cases}$$
For $A \in \mathcal{B}[0,1],$ define
$$\tilde{\mathbb{P}}(A) = \int_{A} Z(\omega) d\mathbb{P}(\omega).$$
(i) Show that $\tilde{\mathbb{P}}$ is a probability measure.

(ii) Show that if $\mathbb{P}(A) = 0$, then $\tilde{\mathbb{P}}(A) = 0$. We say that $\tilde{\mathbb{P}}$ is absolutely continuous with respect to $\mathbb{P}$.

(iii)Show that there is a set $A$ for which $\tilde{\mathbb{P}}(A) = 0$ but $\mathbb{P}(A) > 0$. In other words, $\tilde{\mathbb{P}}$ and $\mathbb{P}$ are not equivalent.


### (i) $\tilde{\mathbb{P}}$ is a probability measure

**$\tilde{\mathbb{P}}(\Omega) = 1$:**

$$\tilde{\mathbb{P}}(\Omega) = \int_0^1 Z(\omega) d\mathbb{P}(\omega) = \int_0^{1/2} 0 \, d\mathbb{P}(\omega) + \int_{1/2}^1 2 \, d\mathbb{P}(\omega) = 0 + 2 \cdot \frac{1}{2} = 1$$

**Countable additivity:** Let $A_1, A_2, \ldots \in \mathcal{B}[0,1]$ be pairwise disjoint. Then:

$$\tilde{\mathbb{P}}\left(\bigcup_{n=1}^{\infty} A_n\right) = \int_{\bigcup_{n=1}^{\infty} A_n} Z(\omega) d\mathbb{P}(\omega) = \sum_{n=1}^{\infty} \int_{A_n} Z(\omega) d\mathbb{P}(\omega) = \sum_{n=1}^{\infty} \tilde{\mathbb{P}}(A_n)$$

Therefore $\tilde{\mathbb{P}}$ is a probability measure. $\blacksquare$

### (ii) $\tilde{\mathbb{P}} \ll \mathbb{P}$

Suppose $\mathbb{P}(A) = 0$. Then:

$$\tilde{\mathbb{P}}(A) = \int_A Z(\omega) d\mathbb{P}(\omega) = 0$$

since the integral of any function over a set of measure zero is zero. Therefore:

$$\mathbb{P}(A) = 0 \implies \tilde{\mathbb{P}}(A) = 0$$

which means $\tilde{\mathbb{P}} \ll \mathbb{P}$ ($\tilde{\mathbb{P}}$ is absolutely continuous with respect to $\mathbb{P}$). $\blacksquare$

### (iii) $\tilde{\mathbb{P}}$ and $\mathbb{P}$ are not equivalent

Let $A = \left[0, \frac{1}{4}\right]$. Then:

$$\mathbb{P}(A) = \frac{1}{4} > 0$$

$$\tilde{\mathbb{P}}(A) = \int_0^{1/4} Z(\omega) d\mathbb{P}(\omega) = \int_0^{1/4} 0 \, d\mathbb{P}(\omega) = 0$$

Since $\tilde{\mathbb{P}}(A) = 0$ but $\mathbb{P}(A) > 0$. Therefore $\tilde{\mathbb{P}}$ and $\mathbb{P}$ are not equivalent. $\blacksquare$

# Exercise 1.11.
In example 1.6.6, we began with a standard normal random  variable $X$ under a measure $\mathbb{P}$. According to Exercise 1.6, this random variable has the moment-generating function
$$\mathbb{E}e^{uX} = e^{\frac{1}{2}u^2} \forall u \in \mathbb{R}.$$
The moment-generating function of a random variable determines its distributin. In particular, any random variable that has moment-generating function $e^{\frac{1}{2}u^2}$ must be standar normal.

In Example 1.6.6, we also defined $Y = X + \theta$, where $\theta$ is a constant, we set $Z = e^{-\theta X - \frac{1}{2} \theta^2}$, and we defined $\tilde{\mathbb{P}}$ by te formula (1.6.9):

$$\tilde{\mathbb{P}}(A) = \int_{A} Z(\omega) d\mathbb{P}(\omega) \quad \forall A \in \mathcal{F}.$$
We showed by considering its cumulative distribution function that $Y$ is a standard normal random variable under $\tilde{\mathbb{P}}$. Give another proof that $Y$ is standard normal under $\tilde{\mathbb{P}}$ by verifying the moment-generating function formula 
$$\tilde{\mathbb{E}}e^{uY} = e^{\frac{1}{2}u^2} \quad \forall u \in \mathbb{R}.$$

**Proof:**

Using $d\tilde{\mathbb{P}} = Z(\omega)d\mathbb{P}$, $Y = X + \theta$, and $Z = e^{-\theta X - \frac{1}{2}\theta^2}$:

$$\tilde{\mathbb{E}}e^{uY} = \int_\Omega e^{uY} d\tilde{\mathbb{P}} = \int_\Omega e^{u(X+\theta)} Z(\omega) d\mathbb{P}$$

$$= \int_\Omega e^{u(X+\theta)} e^{-\theta X - \frac{1}{2}\theta^2} d\mathbb{P}$$

$$= e^{u\theta - \frac{1}{2}\theta^2} \int_\Omega e^{(u-\theta)X} d\mathbb{P}$$

$$= e^{u\theta - \frac{1}{2}\theta^2} \cdot \mathbb{E}e^{(u-\theta)X}$$

Since $X$ is standard normal under $\mathbb{P}$, by Exercise 1.6:

$$\mathbb{E}e^{(u-\theta)X} = e^{\frac{1}{2}(u-\theta)^2}$$

Therefore:

$$\tilde{\mathbb{E}}e^{uY} = e^{u\theta - \frac{1}{2}\theta^2} \cdot e^{\frac{1}{2}(u-\theta)^2} = e^{u\theta - \frac{1}{2}\theta^2 + \frac{1}{2}u^2 - u\theta + \frac{1}{2}\theta^2} = e^{\frac{1}{2}u^2} \quad \blacksquare$$

Since the moment-generating function of $Y$ under $\tilde{\mathbb{P}}$ is $e^{\frac{1}{2}u^2}$, $Y$ is standard normal under $\tilde{\mathbb{P}}$.

# Exercise 1.12
In Example 1.6.6, we began with a standard normal random variable $X$ on a probability space $(\Omega, \mathcal{F}, \mathbb{P})$ and defined the random variable $Y = X + \theta$, where $\theta$ is a constant. We also defined $Z = e^{-\theta X - \frac{1}{2}\theta^2}$ and used $Z$ as the Radon-Nikodym derivative to construct the probability measure $\tilde{\mathbb{P}}$ by the formula (1.6.9):
$$\tilde{\mathbb{P}}(A) = \int_{A} Z(\omega) d\mathbb{P}(\omega) \quad \forall A \in \mathcal{F}.$$
Under $\tilde{\mathbb{P}}$, the random variable $Y$ was shown to be standard normal.

We now have a standard normal random variable $Y$ on the probability space $(\Omega, \mathcal{F}, \tilde{\mathbb{P}})$, and $X$ is related to $Y$ by $X = Y - \theta$. By what we have just stated, with $X$ replaced by $Y$ and $\theta$ replaced by $-\theta$, we could define $\hat{Z} = e^{\theta Y - \frac{1}{2} \theta^2}$ and then use $\hat{Z}$ as a Radon-Nikodym derivative to construct a probability measure $\hat{\mathbb{P}}$ by the formula
$$\hat{\mathbb{P}}(A) = \int_{A}\hat{Z}(\omega)d\tilde{\mathbb{P}} \quad \forall A \in \mathcal{F},$$
so that, under $\hat{\mathbb{P}}$, the random variable $X$ is standard normal. Show that $\hat{Z} = \frac{1}{Z}$ and $\hat{\mathbb{P}} = \mathbb{P}$.
**Proof:**

First, compute $\hat{Z} \cdot Z$:

$$\hat{Z} \cdot Z = e^{\theta Y - \frac{1}{2}\theta^2} \cdot e^{-\theta X - \frac{1}{2}\theta^2}$$

Substituting $Y = X + \theta$:

$$= e^{\theta(X+\theta) - \frac{1}{2}\theta^2} \cdot e^{-\theta X - \frac{1}{2}\theta^2} = e^{\theta X + \theta^2 - \frac{1}{2}\theta^2 - \theta X - \frac{1}{2}\theta^2} = e^0 = 1$$

Therefore $\hat{Z} = \frac{1}{Z}$. $\blacksquare$

Next, using $d\tilde{\mathbb{P}} = Z(\omega)d\mathbb{P}$:

$$\hat{\mathbb{P}}(A) = \int_A \hat{Z}(\omega) d\tilde{\mathbb{P}} = \int_A \hat{Z}(\omega) Z(\omega) d\mathbb{P} = \int_A 1 \, d\mathbb{P} = \mathbb{P}(A)$$

Therefore $\hat{\mathbb{P}} = \mathbb{P}$. $\blacksquare$

# Exercise 1.13 (Change of measure for a normal random variable).
A nonrigorous but informative deviation of the fomula for the Radon-Nikodym derivative $Z(\omega)$ in Example 1.6.6 is provided by this exercise. As in that example, let $X$ be a standard normmal random variable on some probability space $(\Omega, \mathcal{F}, \mathbb{P})$, and let $Y = X + \theta$. Our goal is to define a strictly positive random variable $Z(\omega)$ so that when we set 
$$\tilde{\mathbb{P}}(A) = \int_{A} Z(\omega) d\mathbb{P}(\omega) \quad \forall A \in \mathcal{F},$$
the random variable $Y$ under $\tilde{\mathbb{P}}$ is standard normal. If we fix $\bar{\omega} \in \Omega$ and choose a set $A$ that contains $\bar{\omega}$ and is "small", then (1.9.4) gives 
$$\tilde{\mathbb{P}}(A) \approx Z(\bar{\omega})\mathbb{P}(A),$$
where the symbol $\approx$ means "is approximately equal to." Dividing by $\mathbb{P}(A)$, we see that
$$\frac{\tilde{\mathbb{P}}(A)}{\mathbb{P}} \approx Z(\bar{\omega})$$
for "small" sets $A$ containing $\bar{\omega}$. We use this observatio to identify $Z(\bar{\omega}).$

With $\bar{\omega}$ fixed, let $x = X(\bar{\omega})$. For $\epsilon > 0$, we define $B(x, \epsilon) = [x - \frac{\epsilon}{2}, x + \frac{\epsilon}{2}]$ to be the closed interval centerd at $x$ and having length $\epsilon$. Let $y = x + \theta$ and $B(y, \epsilon) = [y - \frac{\epsilon}{2}, y + \frac{\epsilon}{2}].$

(i) Show that
$$\frac{1}{\epsilon} \mathbb{P}\{X \in B(x, \epsilon)\} \approx \frac{1}{\sqrt{2\pi}} \exp \{-\frac{X^2(\bar{\omega})}{2}\}.$$
(ii) In order for $Y$ to be a standar normal random variable under $\tilde{\mathbb{P}}$, show that we must have 
$$\frac{1}{\epsilon} \tilde{\mathbb{P}}\{Y \in B(y, \epsilon)\} \approx \frac{1}{\sqrt{2\pi}} \exp \{-\frac{Y^2(\bar{\omega})}{2}\}.$$
(iii) Show that $\{X \in B(x, \epsilon)\}$ and $\{Y \in B(y, \epsilon)\}$ are the same set, which we call $A(\bar{\omega}, \epsilon)$. This set containes $\bar{\omega}$ and is "small" when $\epsilon > 0$ is small.

(iv) Show that 
$$\frac{\tilde{\mathbb{P}}(A)}{\mathbb{P}(A)} \approx \exp \{-\theta X(\bar{\omega}) - \frac{1}{2} \theta^2\}.$$
The right-hand side is the value we obtained for $Z(\bar{\omega})$ in Example 1.6.6.
### (i)

Since $X$ is standard normal with density $\frac{1}{\sqrt{2\pi}}e^{-\frac{x^2}{2}}$, for small $\epsilon$:

$$\mathbb{P}\{X \in B(x, \epsilon)\} = \int_{x-\frac{\epsilon}{2}}^{x+\frac{\epsilon}{2}} \frac{1}{\sqrt{2\pi}} e^{-\frac{t^2}{2}} dt \approx \epsilon \cdot \frac{1}{\sqrt{2\pi}} e^{-\frac{x^2}{2}}$$

Dividing both sides by $\epsilon$:

$$\frac{1}{\epsilon}\mathbb{P}\{X \in B(x, \epsilon)\} \approx \frac{1}{\sqrt{2\pi}} \exp\left\{-\frac{X^2(\bar{\omega})}{2}\right\} \quad \blacksquare$$

### (ii)

If $Y$ is standard normal under $\tilde{\mathbb{P}}$ with density $\frac{1}{\sqrt{2\pi}}e^{-\frac{y^2}{2}}$, for small $\epsilon$:

$$\tilde{\mathbb{P}}\{Y \in B(y, \epsilon)\} \approx \epsilon \cdot \frac{1}{\sqrt{2\pi}} e^{-\frac{y^2}{2}}$$

Dividing both sides by $\epsilon$:

$$\frac{1}{\epsilon}\tilde{\mathbb{P}}\{Y \in B(y, \epsilon)\} \approx \frac{1}{\sqrt{2\pi}} \exp\left\{-\frac{Y^2(\bar{\omega})}{2}\right\} \quad \blacksquare$$

### (iii)

Since $Y = X + \theta$ and $y = x + \theta$:

$$\{Y \in B(y, \epsilon)\} = \left\{\omega : y - \frac{\epsilon}{2} \leq Y(\omega) \leq y + \frac{\epsilon}{2}\right\}$$

$$= \left\{\omega : x + \theta - \frac{\epsilon}{2} \leq X(\omega) + \theta \leq x + \theta + \frac{\epsilon}{2}\right\}$$

$$= \left\{\omega : x - \frac{\epsilon}{2} \leq X(\omega) \leq x + \frac{\epsilon}{2}\right\} = \{X \in B(x, \epsilon)\}$$

Therefore both sets are equal to $A(\bar{\omega}, \epsilon)$, which contains $\bar{\omega}$ and is small when $\epsilon > 0$ is small. $\blacksquare$

### (iv)

Using (i), (ii), and (iii):

$$\frac{\tilde{\mathbb{P}}(A)}{\mathbb{P}(A)} = \frac{\tilde{\mathbb{P}}\{Y \in B(y,\epsilon)\}}{\mathbb{P}\{X \in B(x,\epsilon)\}} \approx \frac{\frac{1}{\sqrt{2\pi}}\exp\left\{-\frac{Y^2(\bar{\omega})}{2}\right\}}{\frac{1}{\sqrt{2\pi}}\exp\left\{-\frac{X^2(\bar{\omega})}{2}\right\}} = \exp\left\{-\frac{Y^2(\bar{\omega})}{2} + \frac{X^2(\bar{\omega})}{2}\right\}$$

Substituting $Y(\bar{\omega}) = X(\bar{\omega}) + \theta$:

$$-\frac{(X(\bar{\omega})+\theta)^2}{2} + \frac{X^2(\bar{\omega})}{2} = -\theta X(\bar{\omega}) - \frac{\theta^2}{2}$$

Therefore:

$$\frac{\tilde{\mathbb{P}}(A)}{\mathbb{P}(A)} \approx \exp\left\{-\theta X(\bar{\omega}) - \frac{1}{2}\theta^2\right\} = Z(\bar{\omega}) \quad \blacksquare$$

This confirms the Radon-Nikodym derivative $Z(\bar{\omega}) = e^{-\theta X(\bar{\omega}) - \frac{1}{2}\theta^2}$ obtained in Example 1.6.6.

# Exercise 1.14 (Change of measure for an exponential random variable).
Let $X$ be a nonnegative random variable defined on a probability space $(\Omega, \mathcal{F}, \mathbb{P})$ with the exponential distribution, which is 
$$\mathbb{P}\{X \leq a\} = 1 - e^{-\lambda a}, a \leq 0,$$
where $\lambda$ is a positive constant. Let $\tilde{\lambda}$ be another positive constant, and defin
$$Z = \frac{\tilde{\lambda}}{\lambda} e^{-(\tilde{\lambda} - \lambda)X},$$
Define $\tilde{\mathbb{P}}$ by
$$\tilde{\mathbb{P}}(A) = \int_{A} Z  d\mathbb{P}, \quad \forall A \in \mathcal{F}.$$
(i) Show that $\tilde{\mathbb{P}}(\Omega) = 1.$

(ii) Compute the cumulative distribution function 
$$\tilde{\mathbb{P}}\{X \leq a \}, \quad \forall a \geq 0$$
for the random variable $X$ under the probability measure $\tilde{\mathbb{P}}$.
### (i) $\tilde{\mathbb{P}}(\Omega) = 1$

$$\tilde{\mathbb{P}}(\Omega) = \int_\Omega d\tilde{\mathbb{P}} = \int_\Omega Z \, d\mathbb{P} = \int_0^\infty \frac{\bar{\lambda}}{\lambda} e^{-(\bar{\lambda}-\lambda)x} \cdot \lambda e^{-\lambda x} dx = \bar{\lambda} \int_0^\infty e^{-\bar{\lambda}x} dx = \bar{\lambda} \cdot \frac{1}{\bar{\lambda}} = 1 \quad \blacksquare$$

### (ii) Cumulative distribution function of $X$ under $\tilde{\mathbb{P}}$

$$\tilde{\mathbb{P}}\{X \leq a\} = \int_{\{X \leq a\}} d\tilde{\mathbb{P}} = \int_0^a \frac{\bar{\lambda}}{\lambda} e^{-(\bar{\lambda}-\lambda)x} \cdot \lambda e^{-\lambda x} dx = \bar{\lambda} \int_0^a e^{-\bar{\lambda}x} dx = 1 - e^{-\bar{\lambda}a}, \quad a \geq 0 \quad \blacksquare$$

Therefore, under $\tilde{\mathbb{P}}$, $X$ follows an exponential distribution with parameter $\bar{\lambda}$. The change of measure transforms the distribution of $X$ from $\text{Exp}(\lambda)$ to $\text{Exp}(\bar{\lambda})$.

# Exercise 1.15 (Provided by Alexander Ng).
Let $X$ be a random variable on a probability space $(\Omega, \mathcal{F}, \mathbb{P})$, and assume $X$ has a density functin $f(x)$ that is positive for every $x \in \mathbb{R}$. Let $g$ be a strictly increasing, differentiable function satisfying
$$\lim_{y \to -\infty} g(y) = -\infty, \lim_{y \to \infty} g(y) = \infty,$$
and define the random variable $Y = g(X).$

Let $h(y)$ be an arbitrary nonnegative function satisfying $\int_{-\infty}^{\infty} h(y) dy = 1.$

We want to change the probability measure so that $h(y)$ is the density function for the random variable $Y$. To do this, we define
$$Z = \frac{h(g(X))g'(X)}{f(X)}.$$
(i) Show that $Z$ is nonnegative and $\mathbb{E}Z = 1.$
Now define $\tilde{\mathbb{P}}$ by 
$$\tilde{\mathbb{P}}(A) = \int_{A} Z d\tilde{\mathbb{P}} \quad \forall A \in \mathcal{F}.$$
(ii) Show that $Y$ has density $h$ under $\tilde{\mathbb{P}}.$

### (i) $Z \geq 0$ and $\mathbb{E}Z = 1$

**Nonnegativity:**

Since $h(y) \geq 0$, $g'(x) > 0$ ($g$ is strictly increasing and differentiable), and $f(x) > 0$ for all $x \in \mathbb{R}$:

$$Z = \frac{h(g(X))g'(X)}{f(X)} \geq 0 \quad \blacksquare$$

**$\mathbb{E}Z = 1$:**

$$\mathbb{E}Z = \int_{-\infty}^{\infty} \frac{h(g(x))g'(x)}{f(x)} f(x) dx = \int_{-\infty}^{\infty} h(g(x))g'(x) dx$$

Substituting $y = g(x)$, so $dy = g'(x)dx$, and using $\lim_{x \to \pm\infty} g(x) = \pm\infty$:

$$= \int_{-\infty}^{\infty} h(y) dy = 1 \quad \blacksquare$$

### (ii) $Y$ has density $h$ under $\tilde{\mathbb{P}}$

For any $a \in \mathbb{R}$:

$$\tilde{\mathbb{P}}\{Y \leq a\} = \int_{\{Y \leq a\}} Z \, d\mathbb{P} = \int_{\{g(X) \leq a\}} \frac{h(g(X))g'(X)}{f(X)} f(X) \, dx$$

Since $g$ is strictly increasing, $\{g(X) \leq a\} = \{X \leq g^{-1}(a)\}$:

$$= \int_{-\infty}^{g^{-1}(a)} h(g(x))g'(x) \, dx$$

Substituting $y = g(x)$, so $dy = g'(x)dx$:

$$= \int_{-\infty}^{a} h(y) \, dy$$

This is the cumulative distribution function with density $h$. Therefore, under $\tilde{\mathbb{P}}$, $Y$ has density $h$. $\blacksquare$
