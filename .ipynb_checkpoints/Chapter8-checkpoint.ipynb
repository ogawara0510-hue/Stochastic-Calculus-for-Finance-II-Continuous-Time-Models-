# Exercise 8.1 (Determination of $L_*$ by smooth pasting).
Consider the function $v_L(x)$ in (8.3.11). The first line in formula (8.3.11) implies that the left-hand derivative of $v_L(x)$ at $x = L$ is $v_L(L-) = -1.$ Use the second line in formula (8.3.11) to compute the right-hand derivative $v'_L(L+).$ Show that the smooth-pasting condition
$$v'_{L_*}(L_*-) = v'_{L_*}(L_* +)$$
is satisfied only by $L_*$ given by (8.3.12).

### Left-hand derivative at $x = L$

From the first line, $v_L(x) = K-x$ for $x \leq L$, so:

$$v'_{L}(L-) = -1$$

### Right-hand derivative at $x = L$

From the second line, differentiating with respect to $x$:

$$v'_L(x) = (K-L)\cdot\left(-\frac{2r}{\sigma^2}\right)\cdot\frac{1}{L}\cdot\left(\frac{x}{L}\right)^{-\frac{2r}{\sigma^2}-1}$$

Evaluating at $x = L$:

$$v'_L(L+) = (K-L)\cdot\left(-\frac{2r}{\sigma^2}\right)\cdot\frac{1}{L}$$

### Smooth-pasting condition

Setting $v'_{L_*}(L_*-) = v'_{L_*}(L_*+)$:

$$(K-L_*)\cdot\left(-\frac{2r}{\sigma^2}\right)\cdot\frac{1}{L_*} = -1$$

$$\frac{2r}{\sigma^2}\cdot\frac{K-L_*}{L_*} = 1$$

$$\frac{2r}{\sigma^2}(K-L_*) = L_*$$

$$\frac{2r}{\sigma^2}K = L_*\left(1+\frac{2r}{\sigma^2}\right) = L_*\cdot\frac{2r+\sigma^2}{\sigma^2}$$

$$L_* = \frac{2r}{2r+\sigma^2}K \quad \blacksquare$$

# Exercise 8.2.
Consider two perpetual American puts on the geometric Brownian motion (8.3.1). Suppose the puts have different strike prices, $K_1$ and $K_2$, where $0 < K_1 < K_2.$ Let $v_1(x)$ and $v_2(x)$ denote their respective prices,
as determined in Section 8.3.2. Show that $v_2(x)$ satisfies the first two linear complementarity conditions,
$$v_2(x) \geq (K_1 - x)^+ \text{ for all } x \geq 0, \tag{8.8.1}$$
$$rv_2(x) = rxv'_2(x) - \frac{1}{2} \sigma^2 x^2 v''_2(x) \geq 0 \text{ for all } x \geq 0, \tag{8.8.2}$$
for the perpetual American put price with strike $K_1$ but that $v_2(x)$ does not satisfy the third linear complementarity condition:
$$\text{for each } x \geq 0, \text{ equality holds in either (8.8.1) or (8.8.2) or both.} \tag{8.8.3}$$


### Proof
Recall that for strike $K_i$, the optimal stopping level is $L_i = \frac{2r}{2r+\sigma^2}K_i$, and since $K_1 < K_2$:

$$L_1 < L_2$$

The price of the perpetual American put with strike $K_2$ is:

$$v_2(x) = \begin{cases} K_2 - x & 0 \leq x \leq L_2 \\ (K_2-L_2)\left(\frac{x}{L_2}\right)^{-\frac{2r}{\sigma^2}} & x > L_2 \end{cases}$$

### Verification of (8.8.1): $v_2(x) \geq (K_1-x)^+$

The complementarity condition for strike $K_2$ gives $v_2(x) \geq (K_2-x)^+$ for all $x \geq 0$. Since $K_1 < K_2$:

$$(K_2-x)^+ \geq (K_1-x)^+ \quad \text{for all } x \geq 0$$

Therefore:

$$v_2(x) \geq (K_2-x)^+ \geq (K_1-x)^+ \quad \blacksquare$$

### Verification of (8.8.2): $rv_2(x) - rxv_2'(x) - \frac{1}{2}\sigma^2x^2v_2''(x) \geq 0$

The complementarity condition for strike $K_2$ already guarantees:

$$rv_2(x) - rxv_2'(x) - \frac{1}{2}\sigma^2x^2v_2''(x) \geq 0 \quad \text{for all } x \geq 0 \quad \blacksquare$$

### Failure of (8.8.3)

The condition (8.8.3) requires that for each $x \geq 0$, equality holds in at least one of (8.8.1) or (8.8.2).

Consider $x \in (L_1, L_2)$ (the stopping region for $K_2$ but the continuation region for $K_1$).

**Equality in (8.8.2) fails:** Since $x \leq L_2$, we are in the stopping region of $v_2$, where $v_2(x) = K_2 - x$. In this region:

$$rv_2 - rxv_2' - \frac{1}{2}\sigma^2x^2v_2'' = r(K_2-x) - rx(-1) - 0 = rK_2 > 0$$

So strict inequality holds in (8.8.2).

**Equality in (8.8.1) fails:** Since $x > L_1$ and $L_1 = \frac{2r}{2r+\sigma^2}K_1 < K_1$, for $x \in (L_1, K_1)$:

$$v_2(x) = K_2 - x > K_1 - x = (K_1-x)^+$$

So strict inequality holds in (8.8.1).

**Conclusion:** For any $x \in (L_1, L_2)$, strict inequality holds in both (8.8.1) and (8.8.2), so (8.8.3) fails. $\blacksquare$

**Remark:** This shows that $v_2(x)$ is not the price of the perpetual American put with strike $K_1$. The third complementarity condition (8.8.3) is the key distinguishing condition that identifies the correct price: it captures the fact that at the optimal stopping boundary, the agent is indifferent between exercising and holding.

# Exercise 8.3 (Solving the linear complementarity conditions).
Suppose $v(x)$ is a bounded continuous function having a continuous derivative and satisfying the linear complementarity conditions (8.3.18)-(8.3.20). This exercise shows that $v(x)$ must be the function $v_{L_*}(x)$ given by (8.3.13) with $L_*$ given by (8.3.12). We assume that $K$ is strictly positive.

(i) First consider an interval of $x$-values in which $v(x)$ satisfies (8.3.19) with equality, i.e., where 
$$rv(x) - rxv'(x) - \frac{1}{2}\sigma^2 x^2 v''(x) = 0. \tag{8.8.4}$$
Equation (8.8.4) is a linear, second-order ordinary differenetial equation, and it has two solutions of the form $x^p$ into (8.8.4) and show that the only values of $p$ that cause $x^p$ to satisfy (8.8.4) are $p = -\frac{2r}{\sigma^2}$ and $p = 1$.

(ii) The functions $x^{-\frac{2r}{\sigma^2}}$ and $x$ are said to be linearly independent solutions of (8.8.4), and every function that (8.8.4) on an interval must be of the form $x^p$, the solutions differing because of different value of $p$. Substitute $x^p$ into (8.8.4) on an interval must be of the form
$$f(x) = Ax^{-\frac{2r}{\sigma^2}} + Bx$$
for some constants $A$ and $B$. Use this fact and the fact that both $v(x)$ and $v'(x)$ are continuous to show that there cannot be an interval $[x_1, x_2]$, where $0 < x_1 < x_2 < \infty,$ such that $v(x)$ satisfies (8.3.19) with equality on $[x_1, x_2]$ and satisfies (8.3.18) with equality for x at and immediately to the lefy of $x_1$ and for $x$ at and immediately to the right of $x_2$ unless $v(x)$ is identically zero on $[x_1, x_2].$

(iii) Use the fact that $v(0)$ must equal $K$ to show that there cannot be a number $x_2 > 0$ such that $v(x)$ satisfies (8.3.19) with equality on $[0, x_2].$

(iv) Explain why $v(x)$ cannot satisfy (8.3.19) with equality for all $x \geq 0.$

(v) Explain why $v(x)$ cannot satisfy (8.3.18) with equality for all $x \geq 0.$

(vi) From (iv) and (v) and (8.3.20), we see that $v(x)$ sometimes satisfies (8.3.18) with equality and sometimes does not satisfy (8.3.18) with equality, in which case it must satisfy (8.3.19) with equality. From (ii) and (iii) we see that the region in which $v(x)$ does not satisfy (8.3.18) with equality and satisfies (8.3.19) with equality is not an interval $[x_1, x_2],$ where $0 \leq x_1 < x_2 < \infty,$ nor can this region be a union of disjoint intervals of this form. Therefore, it must be a half-line $[x_1, \infty),$ where $x_1 > 0$. In the region $[0, x_1], v(x)$ satisfies (8.3.18) with equality. Show that $x_1$ must equal $L_*$ given by (8.3.12) and $v(x)$ must be $v_{L_*}(x)$ given by (8.3.13).

### (i) Solutions of the form $x^p$

Substituting $v(x) = x^p$ into (8.8.4):

$$rx^p - rpx^p - \frac{1}{2}\sigma^2p(p-1)x^p = 0$$

Dividing by $x^p > 0$:

$$r - rp - \frac{1}{2}\sigma^2p(p-1) = 0$$

$$-\frac{1}{2}\sigma^2p^2 + \left(\frac{1}{2}\sigma^2 - r\right)p + r = 0$$

$$\sigma^2p^2 - (\sigma^2 - 2r)p - 2r = 0$$

Factoring: $(\sigma^2p + 2r)(p - 1) = 0$, giving:

$$p = -\frac{2r}{\sigma^2} \quad \text{or} \quad p = 1 \quad \blacksquare$$

### (ii) No bounded interval $[x_1, x_2]$ can be the continuation region

Suppose $v(x) = Ax^{-\frac{2r}{\sigma^2}} + Bx$ on $[x_1, x_2]$, and $v(x) = K-x$ immediately to the left of $x_1$ and immediately to the right of $x_2$.

By continuity of $v$ and $v'$ at $x_1$ and $x_2$:

$$Ax_1^{-\frac{2r}{\sigma^2}} + Bx_1 = K - x_1, \quad -\frac{2r}{\sigma^2}Ax_1^{-\frac{2r}{\sigma^2}-1} + B = -1$$

$$Ax_2^{-\frac{2r}{\sigma^2}} + Bx_2 = K - x_2, \quad -\frac{2r}{\sigma^2}Ax_2^{-\frac{2r}{\sigma^2}-1} + B = -1$$

Subtracting the two $v'$ equations:

$$-\frac{2r}{\sigma^2}A\left(x_1^{-\frac{2r}{\sigma^2}-1} - x_2^{-\frac{2r}{\sigma^2}-1}\right) = 0$$

Since $x_1 \neq x_2$, we have $x_1^{-\frac{2r}{\sigma^2}-1} \neq x_2^{-\frac{2r}{\sigma^2}-1}$, so $A = 0$. Then $v(x) = Bx$, and the continuity equations give:

$$Bx_1 = K - x_1 \implies B = \frac{K}{x_1} - 1, \quad Bx_2 = K - x_2 \implies B = \frac{K}{x_2} - 1$$

Since $x_1 \neq x_2$, this is a contradiction unless $A = B = 0$, i.e., $v \equiv 0$ on $[x_1, x_2]$. $\blacksquare$

### (iii) No $x_2 > 0$ such that (8.8.4) holds on $[0, x_2]$

If (8.8.4) holds on $[0, x_2]$, then $v(x) = Ax^{-\frac{2r}{\sigma^2}} + Bx$. Since $v$ is bounded, $A = 0$, giving $v(x) = Bx$. But then $v(0) = 0$, contradicting the boundary condition $v(0) = K > 0$. $\blacksquare$

### (iv) $v(x)$ cannot satisfy (8.3.19) with equality for all $x \geq 0$

If (8.3.19) held with equality for all $x \geq 0$, then in particular it would hold on $[0, x_2]$ for any $x_2 > 0$, contradicting (iii). $\blacksquare$

### (v) $v(x)$ cannot satisfy (8.3.18) with equality for all $x \geq 0$

Suppose $v(x) = (K-x)^+$ for all $x \geq 0$. By (8.3.20), for each $x$, equality holds in exactly one of (8.3.18) or (8.3.19). If (8.3.18) holds with equality for all $x$, then (8.3.19) must be strict for all $x$. However, for $x \geq K$, $v(x) = 0$ and substituting into (8.3.19):

$$r\cdot 0 - rx\cdot 0 - \frac{1}{2}\sigma^2x^2\cdot 0 = 0$$

So (8.3.19) holds with equality for $x \geq K$, contradicting (8.3.20). $\blacksquare$

### (vi) $v(x) = v_{L_*}(x)$ with $x_1 = L_*$

By (ii)-(v), the continuation region must be a half-line $[x_1,\infty)$ and the stopping region $[0,x_1]$. Therefore:

$$v(x) = K - x \quad \text{for } 0 \leq x \leq x_1$$

$$v(x) = Ax^{-\frac{2r}{\sigma^2}} + Bx \quad \text{for } x \geq x_1$$

Since $v$ is bounded as $x\to\infty$, we need $B = 0$, giving $v(x) = Ax^{-\frac{2r}{\sigma^2}}$.

**Applying continuity of $v$ and $v'$ at $x = x_1$:**

$$K - x_1 = Ax_1^{-\frac{2r}{\sigma^2}}$$

$$-1 = -\frac{2r}{\sigma^2}Ax_1^{-\frac{2r}{\sigma^2}-1}$$

From the second equation: $Ax_1^{-\frac{2r}{\sigma^2}} = \frac{\sigma^2}{2r}x_1$. Substituting into the first:

$$K - x_1 = \frac{\sigma^2}{2r}x_1 \implies K = x_1\left(1 + \frac{\sigma^2}{2r}\right) = x_1\cdot\frac{2r+\sigma^2}{2r}$$

$$x_1 = \frac{2r}{2r+\sigma^2}K = L_* \quad \blacksquare$$

Therefore $v(x) = v_{L_*}(x)$ as given by (8.3.13). $\blacksquare$

# Exercise 8.4.
It was asserted at the end of Subsection 8.3.3 and established in Exercise 8.3 that $v_{L_*}(x)$ given by (8.3.13) is the only bounded continuous function having a continuous derivative and satisfying the linear complementarity conditions (8.3.18)-(8.3.20). There are, however, unbounded functions that satisfy these conditions. Let $0 < L < K$ be given, and assume
$$\frac{2r}{2r+\sigma^2}K > L. \tag{8.8.5}$$

(i) Show that, for any constats $A$ and $B$, the function 
$$f(x) = Ax^{-\frac{2r}{\sigma^2}}+Bx \tag{8.8.6}$$
satisfies the differential equation
$$rf(x) - rxf'(x) - \frac{1}{2} \sigma^2 x^2 f''(x) = 0 \text{ for all } x \geq 0. \tag{8.8.7}$$
(ii) Show that the constatns $A$ and $B$ can be chosen so that
$$f(L) = K - L, f'(L) = -1. \tag{8.8.8}$$
(iii) With the constants $A$ and $B$ you chose in (ii), show that $f(x) \geq (K - x)^+$ for all $x \geq L.$

(iv) Define
v(x) = \begin{cases}K - x, & 0 \geq x \geq L, \\ f(x), & x \geq L.\end{cases}.$$
Show that $v(x)$ satisfies the linear complementarity conditions (8.3.18)(8.3.20), but $v(x)$ is not the function $v_{L_*}(x)$ given by (8.3.13).

(v) Every solution of the differential equation (8.8.7) is of the form (8.8.6). In order to have a bounded solution, we must have $B = 0$. Show that in order to hace $B = 0$, we must have $L = \frac{2r}{2r+\sigma^2}K,$ and in this case $v(x)$ agrees with the function $v_{L_*}(x)$ of (8.3.13).

### (i) $f(x)$ satisfies (8.8.7)

Computing derivatives of $f(x) = Ax^{-\frac{2r}{\sigma^2}} + Bx$:

$$f'(x) = -\frac{2r}{\sigma^2}Ax^{-\frac{2r}{\sigma^2}-1} + B$$

$$f''(x) = \frac{2r}{\sigma^2}\left(\frac{2r}{\sigma^2}+1\right)Ax^{-\frac{2r}{\sigma^2}-2}$$

Substituting into the left-hand side of (8.8.7):

$$rf - rxf' - \frac{1}{2}\sigma^2x^2f''$$

$$= r\left(Ax^{-\frac{2r}{\sigma^2}}+Bx\right) - rx\left(-\frac{2r}{\sigma^2}Ax^{-\frac{2r}{\sigma^2}-1}+B\right) - \frac{1}{2}\sigma^2x^2\cdot\frac{2r}{\sigma^2}\left(\frac{2r}{\sigma^2}+1\right)Ax^{-\frac{2r}{\sigma^2}-2}$$

$$= rAx^{-\frac{2r}{\sigma^2}} + rBx + \frac{2r^2}{\sigma^2}Ax^{-\frac{2r}{\sigma^2}} - rBx - r\left(\frac{2r}{\sigma^2}+1\right)Ax^{-\frac{2r}{\sigma^2}}$$

$$= A x^{-\frac{2r}{\sigma^2}}\left[r + \frac{2r^2}{\sigma^2} - r\left(\frac{2r}{\sigma^2}+1\right)\right] = 0 \quad \blacksquare$$

### (ii) Determining $A$ and $B$ from (8.8.8)

From $f(L) = K-L$ and $f'(L) = -1$:

$$AL^{-\frac{2r}{\sigma^2}} + BL = K - L \tag{1}$$

$$-\frac{2r}{\sigma^2}AL^{-\frac{2r}{\sigma^2}-1} + B = -1 \tag{2}$$

From (2): $B = -1 + \frac{2r}{\sigma^2}AL^{-\frac{2r}{\sigma^2}-1}$. Substituting into (1) and solving:

$$AL^{-\frac{2r}{\sigma^2}}\left(1 + \frac{2r}{\sigma^2}\right) = K - L + L = K - L\cdot\frac{\sigma^2}{\sigma^2}... $$

More directly, from (2): $\frac{2r}{\sigma^2}AL^{-\frac{2r}{\sigma^2}} = (B+1)L$. Substituting into (1):

$$\frac{\sigma^2}{2r}(B+1)L + BL = K-L$$

$$L\left[\frac{\sigma^2}{2r}B + \frac{\sigma^2}{2r} + B\right] = K - L$$

$$L\cdot B\cdot\frac{2r+\sigma^2}{2r} = K - L - \frac{\sigma^2}{2r}L = K - \frac{2r+\sigma^2}{2r}L$$

$$B = \frac{2r}{2r+\sigma^2}\cdot\frac{K}{L} - 1$$

Substituting back into (2):

$$A = \frac{\sigma^2}{2r+\sigma^2}L^{\frac{2r}{\sigma^2}}K \quad \blacksquare$$

### (iii) $f(x) \geq (K-x)^+$ for all $x \geq L$

Since $B = \frac{2r}{2r+\sigma^2}\frac{K}{L} - 1 \geq 0$ by (8.8.5) (which states $L \leq \frac{2r}{2r+\sigma^2}K$), and $A > 0$:

$$f(x) = Ax^{-\frac{2r}{\sigma^2}} + Bx \geq Bx = \left(\frac{2r}{2r+\sigma^2}\frac{K}{L}-1\right)x \geq \left(\frac{2r}{2r+\sigma^2}\frac{K}{L}-1\right)L = \frac{2r}{2r+\sigma^2}K - L$$

More directly, for $x \geq L$:

$$f(x) \geq Bx \geq \frac{2r}{2r+\sigma^2}K - x \geq K - x$$

where the last inequality uses $\frac{2r}{2r+\sigma^2}K \geq K$ is false... rather:

Since $f(L) = K-L$ and $f'(L) = -1 = (K-x)'$, and $f$ is convex while $K-x$ is linear, $f(x) \geq K-x$ for $x \geq L$. Also $f(x) \geq 0$ since $A,B \geq 0$ (using (8.8.5)). Therefore:

$$f(x) \geq \max\{K-x, 0\} = (K-x)^+ \quad \text{for all } x \geq L \quad \blacksquare$$

### (iv) $v(x)$ satisfies (8.3.18)-(8.3.20) but $v(x) \neq v_{L_*}(x)$

**On $[0, L]$:** $v(x) = K-x$, so (8.3.18) holds with equality. Since $rv-rxv'-\frac{1}{2}\sigma^2x^2v'' = rK > 0$, (8.3.19) holds with strict inequality. By (8.3.20), equality holds in (8.3.18). ✓

**On $[L, \infty)$:** $v(x) = f(x)$ satisfies (8.8.7) with equality, so (8.3.19) holds with equality. By (iii), $f(x) \geq (K-x)^+$, so (8.3.18) holds. By (8.3.20), equality holds in (8.3.19). ✓

**$v(x) \neq v_{L_*}(x)$:** Since $L \neq L_* = \frac{2r}{2r+\sigma^2}K$ (by assumption (8.8.5) which gives $L < L_*$), the constants $A$ and $B$ differ from those in (8.3.13), so $v(x) \neq v_{L_*}(x)$. $\blacksquare$

### (v) $B = 0$ implies $L = L_*$

Setting $B = 0$:

$$\frac{2r}{2r+\sigma^2}\cdot\frac{K}{L} - 1 = 0 \implies L = \frac{2r}{2r+\sigma^2}K = L_* \quad \blacksquare$$

In this case $A = \frac{\sigma^2}{2r+\sigma^2}L_*^{\frac{2r}{\sigma^2}}K$, and for $x \geq L_*$:

$$v(x) = Ax^{-\frac{2r}{\sigma^2}} = \frac{\sigma^2}{2r+\sigma^2}L_*^{\frac{2r}{\sigma^2}}K\cdot x^{-\frac{2r}{\sigma^2}} = (K-L_*)\left(\frac{x}{L_*}\right)^{-\frac{2r}{\sigma^2}}$$

since $K - L_* = K - \frac{2r}{2r+\sigma^2}K = \frac{\sigma^2}{2r+\sigma^2}K$. This agrees with $v_{L_*}(x)$ of (8.3.13). $\blacksquare$

# Exercise 8.5 (Perpetual American put paying dividends).
Consider a perpetual American put on a geometric Brownian motion asset price paying dividends at a constant rate $a > 0.$ The differential of this asset is 
$$dS(t) = (r - a)S(t)dt + \sigma S(t) d\tilde{W}(t), \tag{8.8.9}$$
where $\tilde{W}(t)$ is a Brownian motion under a risk-neutral measure $\tilde{\mathbb{P}}.$ (Equation (8.8.9) can be obtained by computing the differential in (5.5.8).)

(i) Suppose we adopt the strategy of exercising the put the first time the asset price is at or below $L.$ What is the risk-neutral expected discounted payoff of this strategy? Write this as a function $v_L(x)$ of the initial asset price $x.$ (Hint: Define the positive constant
$$\gamma = \frac{1}{2}\left(r - a - \frac{1}{2} \sigma^2\right) + \frac{1}{\sigma} \sqrt{\frac{1}{\sigma^2}\left(r - a -\frac{1}{2} \sigma^2 \right)^2 + 2r}$$
and write $v_L(x)$ using $\gamma$.)

(ii) Determine $L_*$, the value of $L$ that maximizes the risk-neutral expected discounted payoff computed in (i).

(iii) Show that, for any initial asset price $S(0) = x,$ the process $e^{-rt}v_{L_*}(S(t))$ is a supermartingale under $\tilde{\mathbb{P}}.$ Show that if $S(0) = x > L_*$ and $e^{-rt} v_{L_*}(S(t))$ is stopped the first time the asset price reaches $L_*,$ then the stopped supermartingale is a martingale. (Hint: Show that
$$r + (r - a)\gamma - \frac{1}{2}\sigma^2\gamma(\gamma + 1) = 0. \tag{8.8.10}$$
)

(iv) Show that, for any initial asset price $S(0)=x$,
$$v_{L_*}(x) = \max_{\tau \in T} \tilde{\mathbb{E}}[e^{-r\tau}(K - S(\tau))]. \tag{8.8.11}$$


### (i) Risk-neutral expected discounted payoff $v_L(x)$

The asset price satisfies:

$$S(t) = x\exp\left\{\left(r-a-\frac{\sigma^2}{2}\right)t + \sigma\tilde{W}(t)\right\}$$

Define:

$$\mu = -\frac{1}{\sigma}\left(r-a-\frac{\sigma^2}{2}\right), \quad m = \frac{1}{\sigma}\log\frac{x}{L}, \quad \lambda = r$$

By Theorem 8.3.2, $\tilde{\mathbb{E}}[e^{-\lambda\tau_m}] = e^{-m(\mu+\sqrt{\mu^2+2\lambda})}$ for $\lambda > 0$, where:

$$-\mu + \sqrt{\mu^2+2\lambda} = \sigma\gamma, \quad \gamma = \frac{1}{\sigma^2}\left(r-a-\frac{1}{2}\sigma^2\right) + \frac{1}{\sigma}\sqrt{\frac{1}{\sigma^2}\left(r-a-\frac{1}{2}\sigma^2\right)^2+2r}$$

Therefore, for $x \geq L$:

$$v_L(x) = (K-L)\tilde{\mathbb{E}}[e^{-r\tau_L}] = (K-L)\left(\frac{x}{L}\right)^{-\gamma}$$

For $0 \leq x \leq L$: $v_L(x) = (K-x)^+$

### (ii) Optimal stopping level $L_*$

For $x \geq L$, write $v_L(x) = g(L)\cdot x^{-\gamma}$ where $g(L) = (K-L)L^\gamma$.

We maximize $g(L)$:

$$g'(L) = -L^\gamma + \gamma(K-L)L^{\gamma-1} = L^{\gamma-1}\left[\gamma K - (1+\gamma)L\right] = 0$$

$$L_* = \frac{\gamma}{1+\gamma}K \quad \blacksquare$$

Note $g(0)=0$, $\lim_{L\to\infty}g(L)=-\infty$, and $g(L_*)>0$, confirming $L_*$ is a maximum.

### (iii) $e^{-rt}v_{L_*}(S(t))$ is a supermartingale

Let $v = v_{L_*}$. By Itô's lemma:

$$d(e^{-rt}v) = e^{-rt}\left[-rv + (r-a)Sv' + \frac{1}{2}\sigma^2S^2v''\right]dt + e^{-rt}\sigma Sv'\,d\tilde{W}(t)$$

Let $A = -rv + (r-a)xv' + \frac{1}{2}\sigma^2x^2v''$.

**On $[0, L_*]$:** $v = K-x$, $v'=-1$, $v''=0$:

$$A = -r(K-x) + (r-a)x(-1) = -rK + ax \leq -rK + aL_* \leq 0$$

since $a \in [0,1]$ and $L_* \leq K$.

**On $[L_*, \infty)$:** $v = (K-L_*)\left(\frac{x}{L_*}\right)^{-\gamma}$, $v' = -(K-L_*)\frac{\gamma}{x}\left(\frac{x}{L_*}\right)^{-\gamma}$, $v'' = (K-L_*)\frac{\gamma(\gamma+1)}{x^2}\left(\frac{x}{L_*}\right)^{-\gamma}$:

$$A = -\left(\frac{x}{L_*}\right)^{-\gamma}(K-L_*)\left[r + (r-a)\gamma - \frac{1}{2}\sigma^2\gamma(\gamma+1)\right]$$

**Proof of (8.8.10):** Since $\gamma$ satisfies $r + (r-a-\frac{1}{2}\sigma^2)\gamma - \frac{1}{2}\sigma^2\gamma^2 = 0$:

$$r + (r-a)\gamma - \frac{1}{2}\sigma^2\gamma - \frac{1}{2}\sigma^2\gamma^2 = 0 \implies r + (r-a)\gamma - \frac{1}{2}\sigma^2\gamma(\gamma+1) = 0 \quad \blacksquare$$

Therefore $A = 0$ for $x \geq L_*$.

**Conclusion:** The $dt$ term of $d(e^{-rt}v)$ is $\leq 0$ everywhere, so $e^{-rt}v_{L_*}(S(t))$ is a supermartingale. On $[L_*,\infty)$ the $dt$ term is exactly $0$, so the process stopped at $\tau_{L_*}$ is a martingale. $\blacksquare$

### (iv) $v_{L_*}(x) = \max_{\tau\in\mathcal{T}}\tilde{\mathbb{E}}[e^{-r\tau}(K-S(\tau))]$

**Upper bound:** Since $e^{-rt}v_{L_*}(S(t))$ is a supermartingale, by the Optional Sampling Theorem, for any $\tau \in \mathcal{T}$:

$$v_{L_*}(x) \geq \tilde{\mathbb{E}}\left[e^{-r(t\wedge\tau)}v_{L_*}(S(t\wedge\tau))\right]$$

Since $v_{L_*}(S(t\wedge\tau))$ is bounded, letting $t\to\infty$ by the Dominated Convergence Theorem:

$$v_{L_*}(x) \geq \tilde{\mathbb{E}}\left[e^{-r\tau}v_{L_*}(S(\tau))\right] \geq \tilde{\mathbb{E}}\left[e^{-r\tau}(K-S(\tau))^+\right]$$

where the last inequality uses $v_{L_*}(x) \geq (K-x)^+$ (condition (8.3.18)). Therefore:

$$v_{L_*}(x) \geq \max_{\tau\in\mathcal{T}}\tilde{\mathbb{E}}\left[e^{-r\tau}(K-S(\tau))\right]$$

**Lower bound:** Since $e^{-r(t\wedge\tau_{L_*})}v_{L_*}(S(t\wedge\tau_{L_*}))$ is a martingale:

$$v_{L_*}(x) = \tilde{\mathbb{E}}\left[e^{-r(t\wedge\tau_{L_*})}v_{L_*}(S(t\wedge\tau_{L_*}))\right]$$

Letting $t\to\infty$ by the Dominated Convergence Theorem:

$$v_{L_*}(x) = \tilde{\mathbb{E}}\left[e^{-r\tau_{L_*}}v_{L_*}(S(\tau_{L_*}))\right]$$

When $\tau_{L_*} < \infty$, $S(\tau_{L_*}) = L_*$, so $v_{L_*}(S(\tau_{L_*})) = K-L_* = K-S(\tau_{L_*})$. When $\tau_{L_*} = \infty$, both sides are interpreted as zero. Therefore:

$$v_{L_*}(x) = \tilde{\mathbb{E}}\left[e^{-r\tau_{L_*}}(K-S(\tau_{L_*}))\right] \leq \max_{\tau\in\mathcal{T}}\tilde{\mathbb{E}}\left[e^{-r\tau}(K-S(\tau))\right]$$

**Conclusion:**

$$v_{L_*}(x) = \max_{\tau\in\mathcal{T}}\tilde{\mathbb{E}}\left[e^{-r\tau}(K-S(\tau))\right] \quad \blacksquare$$

There is a second part to Theorem 8.2.4 (optional sampling), which says the following.

## Theorem 8.8.1 (Optional sampling - Part II).
Let $X(t), t \geq 0,$ be a submartingale, and let $\tau$ be a stopping time. Then $\mathbb{E}X(t \wedge \tau) \leq \mathbb{E}X(t).$ If $X(t)$ is a supermartingale, then $\mathbb{E}X(t \wedge \tau) \geq \mathbb{E}X(t).$ If $X(t)$ is a martingale, then $\mathbb{E}X(t \wedge \tau) = \mathbb{E}X(t).$

The proof is technical and is omitted. The idea behind the statement about submartingales is the following . Submartingales tend to go up. Since $t \wedge \tau \leq t,$ we would expect this upward trend to result in the inequality $\mathbb{E}X(t \wedge \tau) \leq \mathbb{E}X(t).$ Where $\tau$ is a stopping time, this intuition is correct. Once we have Theorem 8.8.1 for submartingales, we easily obtain it for supermartingales by using the fact that the negative of a supermartingale is a submartingale. Since a martingale is both a submartingale and a supermartingale, we obtain the equality $\mathbb{E}X(t \wedge \tau) = \mathbb{E}X(t)$ for martingales

Use Theorem 8.8.1 and Lemma 8.5.1 to show in the context of Subsection 8.5.1 that 
$$\tilde{\mathbb{E}}[e^{-rT}(S(T) - K)^+] = \max_{\tau \in \mathcal{T}_{0, t}} \tilde{\mathbb{E}}[e^{-r\tau}(S(\tau) - K)^+], \tag{8.8.12}$$
where as usual we interpret $e^{-r\tau}(S(\tau) - K)^+$ to be zero if $\tau = \infty.$ The right-hand side is the American call price analogous to Definition 8.4.1 for the American put price. The left-hand side is the European call pruce.

**Proof:**

**Step 1: $e^{-rt}(S(t)-K)^+$ is a submartingale**

By Lemma 8.5.1, $h(x) = (x-K)^+$ is convex with $h(0) = 0$, so $e^{-rt}(S(t)-K)^+$ is a submartingale under $\tilde{\mathbb{P}}$.

**Step 2: Upper bound via Theorem 8.8.1**

For any $\tau \in \mathcal{T}_{0,T}$, since $e^{-rt}(S(t)-K)^+$ is a submartingale, Theorem 8.8.1 gives:

$$\tilde{\mathbb{E}}[e^{-r(T\wedge\tau)}(S(T\wedge\tau)-K)^+] \leq \tilde{\mathbb{E}}[e^{-rT}(S(T)-K)^+]$$

Since $\tau \leq T$, we have $T\wedge\tau = \tau$, so:

$$\tilde{\mathbb{E}}[e^{-r\tau}(S(\tau)-K)^+] \leq \tilde{\mathbb{E}}[e^{-rT}(S(T)-K)^+]$$

Since this holds for all $\tau \in \mathcal{T}_{0,T}$:

$$\max_{\tau\in\mathcal{T}_{0,T}}\tilde{\mathbb{E}}[e^{-r\tau}(S(\tau)-K)^+] \leq \tilde{\mathbb{E}}[e^{-rT}(S(T)-K)^+]$$

**Step 3: Lower bound — equality attained at $\tau = T$**

Since $\tau = T$ is itself a valid stopping time in $\mathcal{T}_{0,T}$:

$$\max_{\tau\in\mathcal{T}_{0,T}}\tilde{\mathbb{E}}[e^{-r\tau}(S(\tau)-K)^+] \geq \tilde{\mathbb{E}}[e^{-rT}(S(T)-K)^+]$$

**Conclusion:**

Combining Steps 2 and 3:

$$\tilde{\mathbb{E}}[e^{-rT}(S(T)-K)^+] = \max_{\tau\in\mathcal{T}_{0,T}}\tilde{\mathbb{E}}[e^{-r\tau}(S(\tau)-K)^+] \quad \blacksquare$$

**Remark:** This shows that for an American call on a non-dividend-paying stock, early exercise is never optimal — the American call price equals the European call price. The key insight is that the discounted payoff $e^{-rt}(S(t)-K)^+$ is a submartingale, so waiting until expiry $T$ is always at least as good as stopping early.

# Exercise 8.7.
A function $f(x)$ defined for $x \geq 0$ is said to be convex if, for every $0 \leq x_1 \leq x_2$ and every $0 \leq \lambda \leq 1,$ the inequality
$$f((1 - \lambda)x_1 + \lambda x_2) \leq (1 - \lambda)f(x_1) + \lambda f(x_2)$$
holds. Suppose $f(x)$ and $g(x)$ are convex functions defined for $x \geq 0.$ Show that
$$h(x) = \max \{f(x), g(x)\}$$
is also convex。

**Proof:**

Let $0 \leq x_1 \leq x_2$ and $0 \leq \lambda \leq 1$. Since $h(x) = \max\{f(x),g(x)\}$, either $h = f$ or $h = g$ at the point $(1-\lambda)x_1+\lambda x_2$. By symmetry, assume $h((1-\lambda)x_1+\lambda x_2) = f((1-\lambda)x_1+\lambda x_2)$.

Since $f$ is convex and $h(x_i) \geq f(x_i)$ for $i = 1, 2$:

$$h((1-\lambda)x_1+\lambda x_2) = f((1-\lambda)x_1+\lambda x_2) \leq (1-\lambda)f(x_1)+\lambda f(x_2) \leq (1-\lambda)h(x_1)+\lambda h(x_2) \quad \blacksquare$$

The case $h((1-\lambda)x_1+\lambda x_2) = g((1-\lambda)x_1+\lambda x_2)$ follows by the same argument with $g$ in place of $f$.
